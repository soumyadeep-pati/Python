## 1. What is a variable?

A **variable** is a name that refers to an object (a value) in Python. Assignment creates or updates a name binding.

For example, in `age = 20`:

- `age` is the name.
- `20` is an integer object.
- `=` is the assignment operator.

Python does not require you to declare a variable's type before assigning a value.

```python
age = 20
print(age)
print(type(age))
```

## 2. Assignment and reassignment

Use `=` to assign a value to a name. Assigning a new value to the same name rebinds that name.

This is **reassignment**. It does not necessarily edit the original object.

```python
score = 75
print("First score:", score)

score = 92
print("Updated score:", score)
```

### Assigning an expression

Python evaluates the expression on the right-hand side, then assigns the resulting value to the name on the left.

```python
price = 120
quantity = 3
total_cost = price * quantity

print("Total cost:", total_cost)
```

## 3. A name can refer to values of different types

Python is dynamically typed: a name can be rebound to an object of a different type. The object has a type; the name is not permanently locked to one type.

This flexibility is useful, but consistent use and clear naming make code easier to understand.

```python
value = 42
print(value, type(value))

value = "forty-two"
print(value, type(value))

value = 42.0
print(value, type(value))
```

## 4. Assigning multiple variables

### Multiple assignment

You can assign values to several names in one statement. Values are matched from left to right.

```python
name, age, city = "Mira", 21, "Kolkata"

print("Name:", name)
print("Age:", age)
print("City:", city)
```

### Chained assignment

A chained assignment binds multiple names to the same object. For mutable objects, that can mean multiple names refer to one shared object.

```python
first = second = third = 100

print(first, second, third)
print(first == second == third)
```

### Unpacking an iterable

When the number of names matches the number of items, Python can unpack an iterable into separate names.

```python
coordinates = (10, 25)
x, y = coordinates

print("x =", x)
print("y =", y)
```

If the number of names does not match the number of values, Python raises a `ValueError`. The next cell keeps the error example commented out so the notebook runs cleanly.

```python
# Uncomment to observe a ValueError:
left, middle = (1, 2, 3)
```

## 5. Swapping two variables

Python allows you to swap two names without creating a temporary variable. The right-hand side is evaluated before the assignments are completed.

```python
a = 5
b = 12

print("Before swap:", a, b)
a, b = b, a
print("After swap:", a, b)
```

## 6. Names refer to objects

Think of a name as a label referring to an object. Two names can refer to the same object.

`id()` returns an identity value for an object during its lifetime. Use it to explore identity, but do not rely on a particular numeric ID or assume it will stay the same across runs.

```python
number = 500
another_name = number

print(number)
print(another_name)
print("Same value:", number == another_name)
print("Same object:", number is another_name)
print("IDs:", id(number), id(another_name))
```

### Rebinding one name

Reassigning one name does not automatically reassign another name that previously referred to the same object.

```python
original = 10
alias = original

original = 20

print("original:", original)
print("alias:", alias)
```

## 7. Reassignment versus mutation

These are different operations:

- **Reassignment:** a name is made to refer to another object.
- **Mutation:** a mutable object is changed in place.

Lists are mutable, so they help demonstrate the difference.

```python
numbers = [1, 2, 3]
other_name = numbers

numbers.append(4)

print("numbers:", numbers)
print("other_name:", other_name)
print("Same object:", numbers is other_name)
```

Now compare that with rebinding a name to a new list.

```python
numbers = [1, 2, 3]
other_name = numbers

numbers = [10, 20]

print("numbers:", numbers)
print("other_name:", other_name)
print("Same object:", numbers is other_name)
```

## 8. Variable naming rules

A valid Python identifier:

- Can contain letters, digits, and underscores.
- Cannot start with a digit.
- Cannot be a Python keyword such as `if`, `class`, or `for`.
- Is case-sensitive: `score`, `Score`, and `SCORE` are different names.

Prefer descriptive `snake_case` names for ordinary variables.

```python
student_score = 88
maximum_attempts = 3
is_exam_passed = True

print(student_score)
print(maximum_attempts)
print(is_exam_passed)
```

The following are valid names. Invalid examples are comments so the cell remains executable.

```python
valid_name = "valid"
_name = "also valid"
value2 = 2
studentScore = "valid, but snake_case is usually preferred"

# Invalid examples:
# 2value = 2       # Cannot start with a digit
# first-name = ""  # Hyphen is an operator, not part of an identifier
# class = "CSE"    # 'class' is a Python keyword
```

## 9. Constants by convention

Python does not enforce ordinary constants. By convention, names intended to remain unchanged are written in uppercase with underscores.

```python
PI = 3.141592653589793
MAX_RETRIES = 3

print("PI:", PI)
print("Maximum retries:", MAX_RETRIES)
```

## 10. Common mistakes

### Mistake 1: confusing `=` with `==`

- `=` assigns a value.
- `==` compares two values and produces a Boolean result.

```python
count = 5
print(count == 5)
print(count == 10)
```

### Mistake 2: using a name before assigning it

A name must be bound before it is read in the current scope. Uncomment the line to observe a `NameError`.

```python
# Uncomment to observe the error:
# print(undefined_variable)
```

### Mistake 3: assuming input is numeric

`input()` returns a string. Convert it when you need a number. This example uses a fixed string so it can run automatically in a notebook.

```python
user_age_text = "20"
user_age = int(user_age_text)

print(user_age, type(user_age))
```

## Key takeaways

- Assignment binds a name to an object.
- Reassignment changes what a name refers to; it does not necessarily mutate an object.
- Multiple assignment and unpacking can make code concise and readable.
- Two names may refer to the same object.
- Mutating a shared mutable object is visible through every name referring to it.
- Use descriptive `snake_case` names for ordinary variables.
- `=` assigns; `==` compares values.
- Python constants are generally indicated by uppercase naming convention rather than enforced by the language.
