## 1. What is a string?

A **string** (`str`) is a sequence of text characters. You can write strings using single quotes or double quotes. Use triple quotes for text that spans multiple lines.

Strings are immutable: string operations create new strings rather than changing the original string in place.

```python
single_quoted = 'Hello'
double_quoted = "Python"
multiline = """This is a string
that spans multiple lines."""

print(single_quoted, type(single_quoted))
print(double_quoted, type(double_quoted))
print(multiline)
```

### Quotes inside strings

Choose the quote style that makes the text readable, or use an escape sequence when needed.

```python
sentence = "It's a beautiful day."
quoted_word = 'She said, "Hello!"'
escaped_quote = "It\'s Python."

print(sentence)
print(quoted_word)
print(escaped_quote)
```

## 2. Escape sequences

An escape sequence begins with a backslash (`\`) and represents a special character. Common examples include:

- `\n`: newline
- `\t`: tab
- `\\`: a literal backslash
- `\'` and `\"`: quote characters

Use `print()` to see how escape sequences are displayed.

```python
print("First line\nSecond line")
print("Name:\tAsha")
print("Folder path: C:\\Users\\Asha")
print("She said: \"Hi!\"")
```

### Raw strings

A raw string, written with an `r` prefix, treats backslashes as ordinary characters for most escape-sequence purposes. Raw strings are often handy for Windows-style paths and regular expressions.

```python
normal_path = "C:\\new\\test"
raw_path = r"C:\new\test"

print("Normal string:", normal_path)
print("Raw string:", raw_path)
```

## 3. String length and indexing

`len(text)` returns the number of characters in a string. Python uses zero-based indexing: the first character is at index `0`, and negative indices count from the end.

```python
language = "Python"

print("Length:", len(language))
print("First character:", language[0])
print("Second character:", language[1])
print("Last character:", language[-1])
print("Second-last character:", language[-2])
```

Indexing outside the valid range raises `IndexError`. Keep the example commented out so the notebook runs cleanly.

```python
# Uncomment to observe IndexError:
print(language[100])
```

## 4. String slicing

Slicing extracts a portion of a string with `text[start:stop:step]`. The `stop` index is excluded. If omitted, `start` and `stop` use their default bounds.

```python
text = "Python Programming"

print("First six characters:", text[:6])
print("From index 7:", text[7:])
print("Characters 7 through 17:", text[7:17])
print("Every second character:", text[::2])
print("Reversed:", text[::-1])
```

## 5. Strings are immutable

You cannot replace one character of a string by assigning to an index. Instead, build a new string using slicing, concatenation, or a method.

```python
word = "cat"
new_word = "b" + word[1:]

print("Original:", word)
print("New string:", new_word)

# Uncomment to observe TypeError:
# word[0] = "b" 
```

## 6. Common string methods

String methods return useful results, often as new strings. Methods such as `.lower()`, `.upper()`, `.title()`, `.strip()`, `.replace()`, `.startswith()`, and `.endswith()` are commonly used for text processing.

```python
message = "  Welcome to Python Programming  "

print("lower:", message.lower())
print("upper:", message.upper())
print("title:", message.title())
print("strip:", message.strip())
print("replace:", message.replace("Python", "Data Science"))
print("starts with spaces?", message.startswith("  "))
print("ends with spaces?", message.endswith("  "))
```

### Splitting and joining strings

`.split()` breaks a string into a list of pieces. `.join()` combines strings from an iterable using a separator.

```python
csv_line = "Asha,20,Kolkata"
parts = csv_line.split(",")

print("Parts:", parts)
print("Joined with |:", " | ".join(parts))
print("Words:", "Python makes programming practical".split())
```

### Checking string content

Methods such as `.isalpha()`, `.isdigit()`, `.isalnum()`, and `.isspace()` can help validate text. They check the entire string, and an empty string usually returns `False` for these checks.

```python
samples = ["Python", "2026", "Room42", " ", ""]

for sample in samples:
    print(
        repr(sample),
        "alpha:", sample.isalpha(),
        "digit:", sample.isdigit(),
        "alnum:", sample.isalnum(),
        "space:", sample.isspace()
    )
```

## 7. String comparison and membership

Strings can be compared using operators such as `==`, `!=`, `<`, and `>`. Ordering comparisons are lexicographical (based on Unicode code points), not a measure of linguistic meaning. Use `in` to check whether a substring occurs in another string.

```python
print("python" == "Python")
print("apple" < "banana")
print("Py" in "Python")
print("Java" in "Python")
print("cat" != "dog")
```

For case-insensitive comparisons, normalize both strings first. `.casefold()` is designed for caseless matching and can be more thorough than `.lower()` for some languages.

```python
user_answer = " YES "
expected_answer = "yes"

normalized_answer = user_answer.strip().casefold()
print(normalized_answer == expected_answer.casefold())
```

## 8. String formatting with f-strings

F-strings let you place variables and expressions inside a string using `{}`. They are readable and support formatting numbers, widths, and precision.

```python
name = "Riya"
score = 92
average = 87.456

print(f"{name} scored {score} marks.")
print(f"Average: {average:.2f}")
print(f"Score out of 100: {score:3d}/100")
print(f"Half the score: {score / 2:.1f}")
```

## 9. What is a Boolean?

A Boolean (`bool`) represents one of two logical values: `True` or `False`. Comparison operators such as `==`, `!=`, `<`, `<=`, `>`, and `>=` produce Boolean results.

```python
age = 20
minimum_age = 18

print(age == minimum_age)
print(age != minimum_age)
print(age >= minimum_age)
print(age < minimum_age)
print(type(age >= minimum_age))
```

## 10. Logical operators: `and`, `or`, and `not`

- `and` requires both conditions to be truthy.
- `or` requires at least one condition to be truthy.
- `not` reverses the truth value.

Parentheses can make more complex conditions easier to read.

```python
age = 21
has_student_id = True
is_member = False

can_get_student_discount = age < 25 and has_student_id
can_enter = age >= 18 or is_member
must_register = not is_member

print("Student discount:", can_get_student_discount)
print("Can enter:", can_enter)
print("Must register:", must_register)
```

### Precedence and parentheses

`not` has higher precedence than `and`, which has higher precedence than `or`. Even when you know the precedence, parentheses can make your intention clearer.

```python
a = True
b = False
c = False

print(a or b and c)       # Equivalent to a or (b and c)
print((a or b) and c)
print(not a or b)
print(not (a or b))
```

## 11. Truthy and falsy values

In a Boolean context, Python treats some values as false and most others as true.

Common falsy values include `False`, numeric zero, `None`, and empty collections or strings. A non-empty string such as `"False"` is truthy because it is not empty.

```python
values = [False, True, 0, 1, 0.0, "", "False", "0", None, [], [0], {}, {"x": 0}]

for value in values:
    print(f"{value!r:10} -> bool(value) is {bool(value)}")
```

## 12. Short-circuit evaluation

`and` and `or` short-circuit:

- `A and B` does not evaluate `B` if `A` is falsy.
- `A or B` does not evaluate `B` if `A` is truthy.

In Python, `and` and `or` return one of their operands, not necessarily a Boolean. This behaviour is useful but can surprise beginners.

```python
def show(label, value):
    print("Evaluated:", label)
    return value

print("Result of and:")
result_and = show("left side", False) and show("right side", True)
print("Returned:", result_and)

print("\nResult of or:")
result_or = show("left side", "ready") or show("right side", "fallback")
print("Returned:", result_or)
```

### Safe conditional expressions

Short-circuiting can protect an operation from running when a prerequisite is missing. For example, check that a string is non-empty before reading its first character.

```python
username = ""

if username and username[0].isalpha():
    print("Username starts with a letter.")
else:
    print("Username is empty or does not start with a letter.")
```

## 13. Boolean expressions in real programs

Combine comparisons and logical operators to express rules. Keep conditions readable and test boundary cases, such as values exactly equal to the minimum or maximum allowed.

```python
marks = 78
attendance_percentage = 82

passed_exam = marks >= 40
meets_attendance = attendance_percentage >= 75
eligible_for_exam_credit = passed_exam and meets_attendance

print("Passed exam:", passed_exam)
print("Meets attendance:", meets_attendance)
print("Eligible:", eligible_for_exam_credit)
```

## 14. Common mistakes

- Comparing strings with different capitalization without normalizing them.
- Forgetting that string methods return a result and do not mutate the original string.
- Using `bool("False")` expecting `False`.
- Writing a complicated Boolean condition without parentheses.
- Using `and` or `or` and assuming the result must be `True` or `False`.
- Using `is` to compare ordinary string or numeric values. Use `==` for value equality; reserve `is` for identity checks such as `value is None`.

```python
answer = "False"

print("bool(answer):", bool(answer))
print("answer == 'False':", answer == "False")
print("answer.casefold() == 'false':", answer.casefold() == "false")
```
