## 1. What is a type annotation?

A type annotation is a hint about the kind of value a variable or function is expected to use. It makes code easier for people and static-analysis tools to understand.

Python does **not** generally enforce annotations automatically when the program runs. A type checker such as Pyright or mypy can inspect your code and report mismatches.

```python
age: int = 21
name: str = "Asha"
is_student: bool = True

print(age, type(age))
print(name, type(name))
print(is_student, type(is_student))
```

## 2. Annotations do not automatically enforce types

Python allows a value that disagrees with an annotation at runtime. That does not make the mismatch a good idea; it means the annotation is descriptive unless another tool or your own code checks it.

```python
score: int = 95
score = "excellent"  # Runtime Python allows this assignment.

print(score)
print(type(score))
```

## 3. Annotating function parameters and return values

Write parameter annotations after parameter names and a return annotation after `->`. The return annotation describes the intended result.

```python
def add(a: int, b: int) -> int:
    return a + b

def format_score(name: str, score: float) -> str:
    return f"{name}: {score:.1f}"

print(add(3, 4))
print(format_score("Asha", 91.5))
```

## 4. What happens if a caller passes the wrong type?

The interpreter does not automatically reject the call. The function may still work, produce an unexpected result, or raise an error depending on its implementation.

```python
def repeat_text(text: str, times: int) -> str:
    return text * times

print(repeat_text("ha", 3))

# This call violates the annotation. Python does not reject it because of
# the annotation; the function itself raises an error for this operation.
try:
    print(repeat_text("ha", 2.5))
except TypeError as error:
    print(type(error).__name__ + ":", error)
```

## 5. Common built-in types

Use annotations that describe the values a function or variable is intended to hold.

| Annotation | Intended value |
|---|---|
| `int` | Integer |
| `float` | Floating-point number |
| `str` | Text |
| `bool` | Boolean |
| `None` | No return value / `None` |
| `list[str]` | List of strings |
| `tuple[int, int]` | Two integers |
| `dict[str, int]` | String keys and integer values |
| `set[str]` | Set of strings |

The annotation documents the expected structure; it does not validate every item automatically.

```python
student_name: str = "Asha"
marks: list[int] = [82, 91, 76]
coordinates: tuple[int, int] = (10, 20)
subject_credits: dict[str, int] = {"Math": 4, "Python": 3}
skills: set[str] = {"Python", "SQL"}

print(student_name)
print(marks)
print(coordinates)
print(subject_credits)
print(skills)
```

## 6. Annotating collections

With `list[str]`, the annotation communicates that the list is intended to contain strings. It does not stop you from appending a number at runtime.

```python
languages: list[str] = ["Python", "Java"]
languages.append("SQL")
print(languages)

# Python permits this at runtime, though it violates the annotation.
languages.append(42)
print(languages)
```

## 7. Values that may be `None`

Use `str | None` when a value may be a string or `None`. This is useful for optional names, configuration values, and lookup results.

```python
middle_name: str | None = None

def display_middle_name(value: str | None) -> str:
    if value is None:
        return "No middle name provided."
    return f"Middle name: {value}"

print(display_middle_name(middle_name))
print(display_middle_name("Kumar"))
```

## 8. Functions that return no useful value

Annotate a function that intentionally returns no value with `-> None`. The function may still perform an action such as printing or updating an object.

```python
def show_message(message: str) -> None:
    print(f"Message: {message}")

result = show_message("Keep practicing")
print("Return value:", result)
```

## 9. Type aliases

A type alias gives a reusable name to a more complex type. In modern Python, a simple assignment is enough for many aliases.

```python
StudentRecord = dict[str, str | int]

student: StudentRecord = {
    "name": "Asha",
    "year": 4,
    "department": "CSE",
}

print(student)
```

## 10. A basic `TypedDict`

A normal dictionary can hold many shapes of data. `TypedDict` lets type checkers understand a dictionary with particular keys and expected value types. It is primarily a static typing aid, not automatic runtime validation.

```python
from typing import TypedDict

class Student(TypedDict):
    name: str
    age: int
    enrolled: bool

student: Student = {
    "name": "Asha",
    "age": 21,
    "enrolled": True,
}

print(student)
print(student["name"])
```

## 11. Constants in Python

Python does not have a special keyword that makes a variable constant. By convention, names that should not be reassigned are written in `UPPER_SNAKE_CASE`.

A constant is a promise made by the programmer and communicated through naming; the interpreter does not prevent reassignment.

```python
PI = 3.141592653589793
MAX_LOGIN_ATTEMPTS = 5
DEFAULT_LANGUAGE = "en"

print(PI)
print(MAX_LOGIN_ATTEMPTS)
print(DEFAULT_LANGUAGE)

# Python permits reassignment, but avoid doing this to a constant by convention.
MAX_LOGIN_ATTEMPTS = 10
print("Reassigned value:", MAX_LOGIN_ATTEMPTS)
```

## 12. Constants versus ordinary variables

Use ordinary `snake_case` names for values that are expected to change during program execution. Use `UPPER_SNAKE_CASE` for module-level values intended to remain fixed.

```python
tax_rate = 0.08
subtotal = 100.0
TAX_RATE = 0.08

total = subtotal + subtotal * TAX_RATE
print(f"Total: {total:.2f}")

subtotal = 150.0
total = subtotal + subtotal * TAX_RATE
print(f"Updated total: {total:.2f}")
```

## 13. `Final` for static type checkers

`typing.Final` communicates that a name should not be reassigned. It is checked by compatible static type checkers, but it does not enforce immutability at runtime.

```python
from typing import Final

APP_NAME: Final[str] = "StudyTracker"
MAX_RETRIES: Final = 3

print(APP_NAME)
print(MAX_RETRIES)

# A static type checker should flag reassignment, but Python itself permits it.
MAX_RETRIES = 4
print("Runtime still permits reassignment:", MAX_RETRIES)
```

## 14. Annotations can be inspected

Python stores annotations on functions and annotated module variables. `__annotations__` lets you inspect some of this metadata.

```python
def calculate_area(width: float, height: float) -> float:
    return width * height

print(calculate_area.__annotations__)
```

## 15. Choosing useful annotations

Good annotations clarify intent. They should be accurate and simple enough to maintain. Avoid adding annotations that claim something your code does not actually guarantee.

For example, if a function may return `None`, its return annotation should say so rather than promising a string every time.

```python
def find_user(user_id: int) -> str | None:
    users = {1: "Asha", 2: "Rohan"}
    return users.get(user_id)

print(find_user(1))
print(find_user(99))
```

## 16. Common mistakes

1. **Thinking annotations enforce types at runtime.** They generally do not.
2. **Writing an inaccurate return annotation.** The annotation should match the possible return values.
3. **Forgetting `None` in an optional type.** Use `str | None` if both are valid.
4. **Assuming `Final` makes an object deeply immutable.** It is a static typing marker, not a runtime lock.
5. **Using uppercase names for every variable.** Reserve uppercase names for values intended to be constants.
6. **Assuming `list[int]` checks every element.** Runtime Python does not automatically validate collection contents.

```python
from typing import Final

DEFAULT_PORT: Final[int] = 8000

def get_port(configured_port: int | None) -> int:
    if configured_port is None:
        return DEFAULT_PORT
    return configured_port

print(get_port(None))
print(get_port(9000))
```
