### Exercise 2.3 — Parse a Boolean string safely

Do not use `bool("False")`, because every non-empty string is truthy. Write a function that accepts `"true"` or `"false"` (case-insensitive, allowing surrounding spaces) and returns the corresponding Boolean. Raise `ValueError` for other text.

```python
def parse_bool(text: str) -> bool:
    normalized = text.strip().casefold()
    if normalized == "true":
        return True
    if normalized == "false":
        return False
    raise ValueError("Expected 'true' or 'false'.")

print(parse_bool(" TRUE "))
print(parse_bool("false"))

try:
    parse_bool("yes")
except ValueError as error:
    print(type(error).__name__ + ":", error)
```

### Exercise 3.3 — Compare floating-point results

```python
import math

result = 0.1 + 0.2
expected = 0.3

print("Raw result:", result)
print("Exactly equal?", result == expected)
print("Close enough?", math.isclose(result, expected))
```

### Exercise 5.2 — Distinguish missing and empty

Write a function that returns:
- `"missing"` when the value is `None`
- `"empty"` when the value is an empty string
- `"provided"` otherwise

```python
def describe_input(value: str | None) -> str:
    if value is None:
        return "missing"
    if value == "":
        return "empty"
    return "provided"

for item in (None, "", "hello"):
    print(repr(item), "->", describe_input(item))
```

### Exercise 6.3 — Shallow copy versus deep copy

Create a nested list, make a shallow copy and a deep copy, then mutate an inner list in each copy. Observe which changes affect the original.

```python
import copy

original = [["a"], ["b"]]
shallow = original.copy()
deep = copy.deepcopy(original)

shallow[0].append("from shallow")
deep[1].append("from deep")

print("original:", original)
print("shallow:", shallow)
print("deep:", deep)
```

## Part 8 — Mixed challenge

### Exercise 8.1 — Student result summary

Write a function that accepts a student's name (`str`), marks (`list[float]`), and an optional grade label (`str | None`). It should:

1. Reject an empty marks list with `ValueError`.
2. Calculate the average.
3. Use `"Not assigned"` if the grade label is `None`.
4. Return a formatted string containing the student's name, average to two decimal places, and grade label.

Try writing the function yourself before looking at the solution.

```python
def student_summary(
    name: str,
    marks: list[float],
    grade_label: str | None = None,
) -> str:
    if not marks:
        raise ValueError("marks must not be empty")

    average = sum(marks) / len(marks)
    label = "Not assigned" if grade_label is None else grade_label

    return f"{name}: average={average:.2f}, grade={label}"

print(student_summary("Asha", [82.0, 91.5, 88.0]))
print(student_summary("Rohan", [70.0, 75.0], "B"))
```

```python
# Check the empty-list behavior.
try:
    student_summary("Mira", [])
except ValueError as error:
    print(type(error).__name__ + ":", error)
```
