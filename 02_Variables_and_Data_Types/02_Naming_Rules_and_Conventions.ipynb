## 1. What is an identifier?

An **identifier** is a name used for a variable, function, class, module, or another object in a program.

Examples include `student_name`, `calculate_total`, and `StudentRecord`.

Python has rules that determine whether a name is syntactically valid. Separately, Python programmers follow conventions to make names consistent and readable.

```python
student_name = "Asha"
student_age = 20

def calculate_total(price, quantity):
    return price * quantity

print(student_name)
print(student_age)
print(calculate_total(50, 3))
```

## 2. Identifier rules

For ordinary identifiers in Python:

1. A name may contain letters, digits, and underscores (`_`).
2. A name cannot begin with a digit.
3. Names cannot be Python keywords such as `if`, `for`, or `class`.
4. Identifiers are case-sensitive.
5. Spaces and punctuation such as hyphens are not allowed in an ordinary identifier.

Python 3 supports Unicode identifiers, but for most projects, clear English names are the most portable and familiar choice.

```python
# Valid identifiers
student1 = "Riya"
_student_score = 91
total_marks = 450
name2 = "Kabir"

print(student1, _student_score, total_marks, name2)
```

### Valid or invalid?

The next cell stores examples as strings so even invalid examples can be displayed safely.

```python
examples = {
    "student_name": True,   # valid
    "_count": True,         # valid
    "score2": True,         # valid
    "2score": False,        # invalid: begins with a digit
    "first-name": False,    # invalid as an identifier: contains a hyphen
    "student name": False,  # invalid: contains a space
    "class": False,         # invalid: keyword
}

for identifier, is_valid in examples.items():
    print(f"{identifier!r:16} -> {'valid' if is_valid else 'invalid'}")
```

## 3. Names are case-sensitive

`score`, `Score`, and `SCORE` are three different identifiers. Avoid using names that differ only by capitalization because they are easy to confuse.

```python
score = 75
Score = 82
SCORE = 91

print("score:", score)
print("Score:", Score)
print("SCORE:", SCORE)
print("Are score and Score equal?", score == Score)
```

## 4. Naming conventions

Conventions are recommended styles, not usually syntax requirements. Following them makes code easier for other people—and your future self—to read.

| Style | Common use | Example |
|---|---|---|
| `snake_case` | Variables, functions, and methods | `student_name`, `calculate_total()` |
| `PascalCase` | Classes | `StudentRecord` |
| `UPPER_SNAKE_CASE` | Constants by convention | `MAX_RETRIES` |
| `_leading_underscore` | Internal-use hint | `_cache` |
| `name_` suffix | Avoid a keyword/name collision | `class_` |

These are common Python conventions described by the Python community's style guidance. The language generally does not enforce them.

```python
# Variables and functions: snake_case
student_name = "Mina"

def calculate_average(total_marks, subject_count):
    return total_marks / subject_count

# Class names: PascalCase
class StudentRecord:
    pass

# Constants by convention: UPPER_SNAKE_CASE
MAX_LOGIN_ATTEMPTS = 5

print(student_name)
print(calculate_average(360, 6))
print(StudentRecord.__name__)
print(MAX_LOGIN_ATTEMPTS)
```

## 5. Choose descriptive names

Good names communicate purpose. Prefer names that explain the value or action instead of names that make readers guess.

Avoid names that are too short unless the context is conventional (for example, `i` in a short loop). Avoid unnecessarily long names that repeat information already obvious from the context.

```python
# Harder to understand
x = 480
y = 6
z = x / y

# Clearer intent
total_marks = 480
number_of_subjects = 6
average_marks = total_marks / number_of_subjects

print("Average marks:", average_marks)
```

### Naming Boolean variables

Boolean names often read naturally as questions or states. Prefixes such as `is_`, `has_`, and `can_` can help.

```python
is_logged_in = True
has_permission = False
can_submit_assignment = True

print("Logged in:", is_logged_in)
print("Has permission:", has_permission)
print("Can submit:", can_submit_assignment)
```

## 6. Avoid shadowing built-in names

Python provides built-in functions and types such as `list`, `str`, `sum`, `max`, and `id`. Avoid assigning your own variables those names because doing so can hide the built-in in the current scope.

The next example demonstrates the problem using a separate variable name and does not overwrite any built-ins.

```python
# Good: use a descriptive name instead of 'list'
student_names = ["Asha", "Rohan", "Mina"]
name_count = len(student_names)

print(student_names)
print(name_count)
print(sum([10, 20, 30]))
```

Here is what shadowing can look like. The example is kept in a function so the built-in name is not overwritten globally.

```python
def demonstrate_shadowing():
    list = ["a", "b"]  # Avoid this naming choice in real code.
    print("Local variable:", list)
    # Inside this function, list(...) would no longer call the built-in list.
    print("Use len() normally:", len(list))

demonstrate_shadowing()
print("Built-in list still works outside the function:", list((1, 2, 3)))
```

## 7. Keywords and reserved words

Keywords have special meaning in Python syntax and cannot be used as ordinary variable or function names.

Use the `keyword` module to inspect the keywords supported by the Python interpreter running your notebook.

```python
import keyword

print("Python keywords:")
print(", ".join(keyword.kwlist))
print()
print("Is 'for' a keyword?", keyword.iskeyword("for"))
print("Is 'student_name' a keyword?", keyword.iskeyword("student_name"))
```

If you need a name similar to a keyword, use a suffix such as an underscore: `class_`, `for_`, or `from_`. Do not change the spelling of a keyword and expect it to become a valid keyword-based variable name.

```python
class_ = "CSE"
for_ = "loop example"

print(class_)
print(for_)
```

## 8. Leading and trailing underscores

Underscores can communicate intent:

- `_name`: conventionally indicates internal use; it does not make the name truly private.
- `name_`: useful when a natural name conflicts with a keyword or another name.
- `__name` inside a class: triggers name mangling in certain contexts.
- `__name__`: often used for special methods and attributes defined by Python, such as `__init__`.

Do not invent your own `__special__` names casually; those names are commonly reserved for Python's data model and related conventions.

```python
_internal_cache = {"ready": True}
class_ = "not a keyword"

print(_internal_cache)
print(class_)
```

## 9. Naming modules, packages, classes, and functions

Common conventions:

- **Module filenames:** short, lowercase names; underscores are acceptable, such as `data_loader.py`.
- **Package names:** usually short, lowercase names.
- **Classes:** `PascalCase`, such as `DatasetLoader`.
- **Functions and methods:** `snake_case`, such as `load_dataset()`.
- **Constants:** `UPPER_SNAKE_CASE`, such as `DEFAULT_TIMEOUT`.

Consistency matters more than inventing a unique style for every file.

```python
DEFAULT_TIMEOUT = 30

class DatasetLoader:
    def load_dataset(self, file_path):
        return f"Loading data from {file_path}"

loader = DatasetLoader()
print(loader.load_dataset("students.csv"))
print("Timeout:", DEFAULT_TIMEOUT)
```

## 10. Check identifiers programmatically

Python's `str.isidentifier()` checks whether a string is a valid identifier. It does not check whether the identifier is a reserved keyword, so combine it with `keyword.iskeyword()` when needed.

```python
import keyword

def explain_identifier(candidate):
    syntactically_valid = candidate.isidentifier()
    is_keyword = keyword.iskeyword(candidate)

    if syntactically_valid and not is_keyword:
        result = "valid ordinary identifier"
    elif is_keyword:
        result = "keyword; cannot be used as an ordinary identifier"
    else:
        result = "invalid identifier"

    print(f"{candidate!r:18} -> {result}")

for candidate in ["student_name", "_score", "2score", "class", "first-name"]:
    explain_identifier(candidate)
```

## 11. Common naming mistakes

### Mistake A: using spaces or hyphens

Use underscores between words: `student_score`, not `student score` or `student-score`.

### Mistake B: inconsistent capitalization

Choose a style and apply it consistently.

### Mistake C: vague names

Names such as `data`, `temp`, and `value` can be fine locally, but use more descriptive names when their meaning is not obvious.

### Mistake D: shadowing built-ins

Avoid names like `str`, `list`, `sum`, and `max` for your own variables.

### Mistake E: treating conventions as syntax

`studentName` is a valid identifier, even though `student_name` is the usual Python style for a variable.

```python
# Compare styles
student_name = "Preferred Python variable style"
studentName = "Valid, but not the usual Python variable convention"

print(student_name)
print(studentName)
```
