## 1. What is a data type?

A **data type** describes the kind of value an object represents and the operations that make sense for it.

For example:

- `25` is an integer.
- `25.5` is a floating-point number.
- `"hello"` is a string.
- `True` is a Boolean value.

Python is dynamically typed, so you do not usually declare a variable's type in advance. The object you assign to a name has a type.

```python
age = 20
temperature = 26.5
greeting = "Hello, Python!"
is_learning = True

print(age, type(age))
print(temperature, type(temperature))
print(greeting, type(greeting))
print(is_learning, type(is_learning))
```

## 2. Integers — `int`

Integers represent whole numbers, including negative numbers and zero. Python integers can grow beyond the fixed-width limits common in many lower-level numeric types, subject to available memory.

```python
student_count = 45
temperature_change = -3
zero_value = 0
large_integer = 123456789012345678901234567890

print(student_count, type(student_count))
print(temperature_change, type(temperature_change))
print(zero_value, type(zero_value))
print(large_integer, type(large_integer))
```

### Integer arithmetic

The operators `+`, `-`, and `*` perform addition, subtraction, and multiplication. The `/` operator produces a floating-point result, while `//` performs floor division. The `%` operator returns a remainder, and `**` raises a value to a power.

```python
a = 17
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
```

## 3. Floating-point numbers — `float`

A `float` represents a real-number approximation, commonly used for values with fractional parts. Most decimal fractions cannot be represented exactly in binary floating-point, so tiny rounding differences can appear.

```python
height = 1.75
temperature = -2.5
scientific_notation = 1.2e3

print(height, type(height))
print(temperature, type(temperature))
print(scientific_notation, type(scientific_notation))
```

### Floating-point precision

Do not assume that every decimal calculation produces an exact decimal representation. For many ordinary comparisons, compare with a suitable tolerance rather than relying on exact equality.

```python
result = 0.1 + 0.2

print("0.1 + 0.2 =", result)
print("Exactly equal to 0.3?", result == 0.3)
print("Close enough for this example?", abs(result - 0.3) < 1e-9)
```

## 4. Complex numbers — `complex`

A complex number has a real part and an imaginary part. Python uses `j` for the imaginary unit. Complex numbers are useful in some scientific, engineering, and mathematical applications.

```python
z = 3 + 4j

print("Value:", z)
print("Type:", type(z))
print("Real part:", z.real)
print("Imaginary part:", z.imag)
print("Magnitude:", abs(z))
print("Conjugate:", z.conjugate())
```

## 5. Booleans — `bool`

A Boolean has one of two values: `True` or `False`. Booleans commonly represent conditions and are produced by comparison operations.

In Python, `bool` is a subclass of `int`, but beginners should usually treat Boolean values as logical values rather than numbers.

```python
is_adult = True
has_passed = False

print(is_adult, type(is_adult))
print(has_passed, type(has_passed))
print("10 > 3:", 10 > 3)
print("10 == 3:", 10 == 3)
print("True + True:", True + True)  # Python evaluates this as 2
```

### Logical operators

`and`, `or`, and `not` combine or invert Boolean conditions. The expressions below are useful examples of Boolean logic.

```python
age = 20
has_id = True

print("Adult:", age >= 18)
print("Can enter:", age >= 18 and has_id)
print("Needs supervision:", not (age >= 18))
print("Eligible for student discount:", age < 25 or has_id)
```

## 6. Strings — `str`

A string is a sequence of text characters. Strings can be written with single quotes, double quotes, or triple quotes. Strings are immutable: operations that appear to modify text produce a new string rather than changing the original string object in place.

```python
single_quoted = 'Hello'
double_quoted = "Python"
multiline = """This string
spans multiple lines."""

print(single_quoted, type(single_quoted))
print(double_quoted, type(double_quoted))
print(multiline)
```

### Common string operations

Strings support concatenation with `+`, repetition with `*`, indexing, slicing, and useful methods such as `.lower()`, `.upper()`, and `.strip()`. Indexing and slicing will be studied more deeply in the data structures section.

```python
first_name = "Ada"
last_name = "Lovelace"
full_name = first_name + " " + last_name

print("Full name:", full_name)
print("Repeated:", "ha" * 3)
print("First character:", full_name[0])
print("First three characters:", full_name[:3])
print("Uppercase:", full_name.upper())
print("Lowercase:", full_name.lower())
print("Trimmed:", "   Python   ".strip())
```

### String formatting with f-strings

An f-string lets you embed expressions inside a string using braces. It is a common, readable way to construct messages.

```python
name = "Riya"
score = 92
print(f"{name} scored {score} marks.")
print(f"Half the score is {score / 2:.1f}.")
```

## 7. The special value `None`

`None` represents the absence of a value or a missing result. Its type is `NoneType`. It is different from `0`, `False`, and an empty string.

Use `is None` when checking whether a value is `None`.

```python
result = None

print(result)
print(type(result))
print("Is None?", result is None)
print("Is zero?", result == 0)
print("Is False?", result == False)
```

Be careful: Python considers `False == 0` to be `True`, because Boolean is a subclass of integer. But `None` is distinct from both. Use `is None` for a clear `None` check.

```python
print("False == 0:", False == 0)
print("False is 0:", False is 0)
print("None == False:", None == False)
print("None is None:", None is None)
```

## 8. Inspecting types: `type()` and `isinstance()`

Use `type(value)` to inspect the exact type of an object. Use `isinstance(value, SomeType)` to check whether an object belongs to a type or its subclass hierarchy. For most type checks in application code, `isinstance()` is often the more flexible choice.

```python
values = [10, 3.14, 2 + 3j, True, "text", None]

for value in values:
    print(
        f"value={value!r:10} "
        f"type={type(value).__name__:8} "
        f"is int? {isinstance(value, int)}"
    )
```

Notice that `isinstance(True, int)` returns `True`. This is a consequence of Python's type hierarchy, not a mistake in `isinstance()`.

```python
print("isinstance(True, int):", isinstance(True, int))
print("type(True) is int:", type(True) is int)
print("isinstance(10, (int, float)):", isinstance(10, (int, float)))
print("isinstance(10.5, (int, float)):", isinstance(10.5, (int, float)))
```

## 9. Type conversion

Type conversion creates a value of another type when the conversion is supported. Common constructors include `int()`, `float()`, `complex()`, `str()`, and `bool()`.

Conversion is not always lossless. For example, converting a float to an integer truncates the fractional part toward zero.

```python
print(int("42"), type(int("42")))
print(float("3.5"), type(float("3.5")))
print(complex("2+3j"), type(complex("2+3j")))
print(str(100), type(str(100)))
print(int(8.9))  # Fractional part is discarded; result is 8
print(int(-8.9)) # Truncates toward zero; result is -8
```

### Boolean conversion and truthiness

`bool(value)` returns `False` for common empty or zero-like values, including `0`, `0.0`, `""`, `None`, and empty collections. Many other values are truthy.

Do not assume that the strings `"False"` or `"0"` convert to `False`; non-empty strings are truthy.

```python
test_values = [0, 1, 0.0, 2.5, "", "False", "0", None, [], [0]]

for value in test_values:
    print(f"{value!r:10} -> {bool(value)}")
```

### Invalid conversions

Some conversions raise `ValueError` or `TypeError`. The following examples are wrapped in `try`/`except` so the notebook continues running and shows the error clearly.

```python
examples = ["123", "12.5", "hello"]

for text_value in examples:
    try:
        converted = int(text_value)
        print(f"int({text_value!r}) = {converted}")
    except ValueError as error:
        print(f"Cannot convert {text_value!r} to int: {error}")
```

## 10. Collections: a brief preview

Python also has built-in collection types. You will study them in depth in the next parts of the repository.

- `list`: ordered, mutable collection, written with square brackets.
- `tuple`: ordered, immutable collection, often written with parentheses.
- `dict`: mapping of keys to values, written with braces and key-value pairs.
- `set`: collection of unique elements, written with braces or `set()`.

An empty set is written as `set()`, because `{}` creates an empty dictionary.

```python
numbers_list = [1, 2, 2, 3]
coordinates_tuple = (10, 20)
student_dict = {"name": "Asha", "score": 95}
unique_numbers = {1, 2, 2, 3}
empty_set = set()
empty_dict = {}

for value in [numbers_list, coordinates_tuple, student_dict, unique_numbers, empty_set, empty_dict]:
    print(repr(value), "->", type(value).__name__)
```

## 11. Mutability overview

Some built-in types are immutable, meaning their value cannot be changed in place. Examples include `int`, `float`, `bool`, `complex`, `str`, and `tuple` (though a tuple can contain mutable objects).

Lists, dictionaries, and sets are mutable. Mutability and references are important when passing values around; this will be explored more deeply in later notebooks.

```python
# String example: concatenation creates a new string.
word = "cat"
new_word = word + "s"

print("word:", word)
print("new_word:", new_word)

# List example: append mutates the existing list.
items = ["cat"]
same_items = items
items.append("dog")

print("items:", items)
print("same_items:", same_items)
print("Same list object:", items is same_items)
```
