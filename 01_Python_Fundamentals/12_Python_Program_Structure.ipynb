## 1. What is program structure?

**Program structure** means how code is arranged and organized so it is readable, understandable, and behaves as intended.

Even a short Python program can have several parts:

1. A module docstring describing the file.
2. Imports for functionality the program needs.
3. Constants used throughout the program.
4. Function or class definitions, when needed.
5. A main section that starts the program's work.

A small script does not always need every part. The goal is to use a structure that suits the program rather than adding unnecessary boilerplate.

## 3. Top-level code and code blocks

**Top-level code** is code written directly in a module, not nested inside a function, class, or other compound statement.

For example:

```python
course_name = "Python"  # top-level assignment

def display_course():
    print(course_name)  # inside a function block

display_course()        # top-level function call
```

Python uses indentation to mark the body of a function or other compound statement. The function body does not run merely because Python reads the `def` statement; it runs when the function is called.

The assignment and function call at the left margin are top-level statements.

````python
course_name = "Python"

def display_course():
    print(course_name)

display_course()
````

**Expected output**

```text
Python
```

## 4. Imports: using existing functionality

An **import** makes a module or selected names available to your code.

For example, Python's standard library includes the `math` module:

```python
import math

print(math.sqrt(81))
print(math.pi)
```

**Expected output**

```text
9.0
3.141592653589793
```

The `import math` statement makes the module available under the name `math`. The dot notation, as in `math.sqrt`, accesses an attribute from that module.

### Common import styles

```python
import math
from math import sqrt
import math as mathematics
```

These are all valid patterns, but prefer clear imports. Using `import math` makes it obvious where `sqrt` comes from when you write `math.sqrt(81)`.

````python
import math

square_root = math.sqrt(81)
print(square_root)
print(round(math.pi, 2))
````

**Expected output**

```text
9.0
3.14
```

## 5. A typical order for a small script

A readable script often follows this order:

```text
1. Module docstring
2. Imports
3. Constants
4. Function and class definitions
5. Main entry-point block
```

For example:

```python
\"\"\"Calculate and display a simple total.\"\"\"

import math

DEFAULT_RADIUS = 3

def circle_area(radius):
    return math.pi * radius ** 2

if __name__ == "__main__":
    print(circle_area(DEFAULT_RADIUS))
```

This arrangement makes the file easier to scan. Python does not require every script to follow exactly this template, but it is a useful habit as programs grow.

## 6. What is `__name__`?

Python provides a special variable named `__name__` in each module.

Its value depends on how the module is being used:

- When a file is run directly as the main program, `__name__` is set to `"__main__"`.
- When a module is imported under its module name, `__name__` is usually set to that module's name.

This allows a file to provide reusable definitions while also supporting direct execution.

````python
print(__name__)
````

**Expected output in a typical notebook kernel**

```text
__main__
```

Notebook environments commonly execute cells in a context where `__name__` is `"__main__"`. In an imported `.py` module, the value is normally the module's name instead.

## 7. Understanding `if __name__ == "__main__":`

This pattern checks whether the current module is being run as the main program.

```python
def main():
    print("Program started")

if __name__ == "__main__":
    main()
```

When you run the file directly, the condition is true and `main()` is called.

When another file imports this module, the condition is normally false, so the main program does not automatically start during the import.

### Why is this useful?

Without this guard, top-level instructions such as printing messages, asking for input, or starting a long process may run whenever the file is imported. The guard helps separate reusable definitions from the actions that should happen when the script is run directly.

````python
def main():
    print("Program started")

if __name__ == "__main__":
    main()
````

**Expected output in this notebook**

```text
Program started
```

The notebook kernel usually has `__name__ == "__main__"`, so the guarded block runs. In a standalone `.py` file run directly, it behaves the same way.

## 8. Direct execution versus importing

Imagine a file named `greetings.py`:

```python
def greet(name):
    return f"Hello, {name}!"

def main():
    print("Running greetings.py directly")
    print(greet("Asha"))

if __name__ == "__main__":
    main()
```

If you run:

```bash
python greetings.py
```

the output is:

```text
Running greetings.py directly
Hello, Asha!
```

If another file imports `greetings`, the `greet` and `main` functions become available, but the guarded call to `main()` does not run automatically during import.

This is the key benefit of the entry-point guard.

## 9. A runnable import demonstration

The following cell creates two temporary Python files and runs them in separate processes:

- `greetings.py` defines a function and has a main guard.
- `run_import.py` imports the module and calls the function itself.

Using temporary files keeps the exercise self-contained and avoids adding files to your repository.

````python
from pathlib import Path
from tempfile import TemporaryDirectory
import subprocess
import sys

with TemporaryDirectory() as temp_dir:
    folder = Path(temp_dir)

    (folder / "greetings.py").write_text(
        'def greet(name):\n'
        '    return f"Hello, {name}!"\n'
        '\n'
        'def main():\n'
        '    print("Running greetings.py directly")\n'
        '    print(greet("Asha"))\n'
        '\n'
        'if __name__ == "__main__":\n'
        '    main()\n',
        encoding="utf-8",
    )

    (folder / "run_import.py").write_text(
        'import greetings\n'
        'print("Imported greetings.py")\n'
        'print(greetings.greet("Ravi"))\n',
        encoding="utf-8",
    )

    direct = subprocess.run(
        [sys.executable, str(folder / "greetings.py")],
        capture_output=True, text=True, check=True
    )
    imported = subprocess.run(
        [sys.executable, str(folder / "run_import.py")],
        cwd=folder, capture_output=True, text=True, check=True
    )

    print("--- Run greetings.py directly ---")
    print(direct.stdout.strip())
    print("--- Run run_import.py ---")
    print(imported.stdout.strip())
````

**Expected output**

```text
--- Run greetings.py directly ---
Running greetings.py directly
Hello, Asha!
--- Run run_import.py ---
Imported greetings.py
Hello, Ravi!
```

Notice that importing `greetings` does not print `Running greetings.py directly`. The main guard prevents that part from running on import. The importing script can still call `greetings.greet("Ravi")`.

## 10. Keep top-level code predictable

When a module is imported, Python executes its top-level statements and definitions. This means top-level code should generally be kept simple.

Usually reasonable at top level:

- Imports.
- Constants.
- Function and class definitions.
- Small, intentional initialization.

Often best placed inside `main()` or another function:

- Starting a program's main workflow.
- Printing large reports automatically.
- Asking a user for input.
- Running long computations.
- Performing actions that should not happen just because a module is imported.

This is a guideline, not an absolute rule. The right organization depends on the module's purpose.

## 11. Comments, docstrings, and structure

Comments explain particular parts of code to readers. A **docstring** is a string literal placed at the beginning of a module, function, or class to document its purpose.

Example:

```python
\"\"\"Display a greeting.\"\"\"

def greet(name):
    \"\"\"Return a greeting for the supplied name.\"\"\"
    return f"Hello, {name}!"
```

The first string is the module docstring; the second is the function docstring.

Use names and structure that make the code self-explanatory. Add comments when they clarify *why* something is done, rather than restating an obvious line.

## 12. Build a complete small program

The following is a complete example you can save as `simple_program.py`.


This example demonstrates a module docstring, a constant, function definitions, a main function, and the entry-point guard. It does not require any external packages.

````python
"""A tiny program demonstrating basic organization."""

APP_NAME = "Python Fundamentals"

def make_greeting(name):
    """Return a greeting for the given name."""
    return f"Welcome to {APP_NAME}, {name}!"

def main():
    """Run the program."""
    user_name = "Asha"
    print(make_greeting(user_name))

if __name__ == "__main__":
    main()
````

**Expected output**

```text
Welcome to Python Fundamentals, Asha!
```

## 13. Common mistakes

1. **Putting every action at top level.** Importing the module may then trigger actions unexpectedly.
2. **Thinking `if __name__ == "__main__":` is mandatory.** It is a common and useful pattern, not a requirement for every file.
3. **Confusing a definition with a call.** Defining a function does not run its body; calling it does.
4. **Assuming imports only make names available.** Importing a module also executes its top-level code.
5. **Using unclear file organization.** Group related definitions and keep the entry point easy to find.
6. **Assuming the notebook always behaves exactly like a `.py` file.** Notebook kernels have their own execution context, though the main-guard condition commonly evaluates to true.
