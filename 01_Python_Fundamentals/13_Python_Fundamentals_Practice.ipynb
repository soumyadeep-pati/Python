## Part 4 — Comments, docstrings, identifiers, and naming

- A **comment** starts with `#` and explains code to a reader.
- A **docstring** is a string placed at the beginning of a module, class, or function to document it.
- An **identifier** is a name used for something such as a variable or function.
- Common conventions: `snake_case` for variables and functions, `UPPER_SNAKE_CASE` for constants, and `PascalCase` for class names.

Python keywords such as `if`, `for`, and `class` are reserved syntax words and cannot be used as ordinary identifiers.

````python
"""A tiny example of documentation and naming."""

# A constant is conventionally written in uppercase.
MAX_ATTEMPTS = 3

# Variable names usually use snake_case.
student_name = "Asha"
attempt_count = 1

print("Student:", student_name)
print("Attempt:", attempt_count, "of", MAX_ATTEMPTS)
````

### Exercise 4.1 — Identifier check

The next cell stores possible names as strings, so it is safe to include names that would be invalid if used directly in Python code. `str.isidentifier()` checks identifier shape; `keyword.iskeyword()` checks whether a name is a Python keyword.

````python
import keyword

candidate_names = [
    "student_name",
    "student-name",
    "2nd_student",
    "_score",
    "class",
    "total2",
]

for name in candidate_names:
    valid_shape = name.isidentifier()
    reserved_word = keyword.iskeyword(name)
    can_be_used_as_identifier = valid_shape and not reserved_word
    print(
        f"{name!r}: shape={valid_shape}, "
        f"keyword={reserved_word}, usable={can_be_used_as_identifier}"
    )
````

## Part 5 — Dynamic typing and object identity

Python names refer to objects. A name can later refer to an object of a different type; this is called **dynamic typing**. `type(value)` reports an object's type. `==` compares values, while `is` checks whether two references point to the very same object.

````python
item = 42
print("First value:", item)
print("First type:", type(item).__name__)

item = "forty-two"
print("Second value:", item)
print("Second type:", type(item).__name__)
````

````python
first = [1, 2, 3]
second = [1, 2, 3]
third = first

print("first == second:", first == second)
print("first is second:", first is second)
print("first is third:", first is third)
````

**Important:** Do not use `is` as a replacement for `==` when comparing ordinary values. Use `is` when identity is what matters, commonly for checks such as `value is None`.

## Part 6 — Interactive mode, scripts, and bytecode

**Interactive mode** runs commands and shows results immediately, which is useful for experimentation. A **script** is a `.py` file that can be run as a complete program. Python implementations commonly compile source code to bytecode before executing it; CPython bytecode can be inspected with the `dis` module.

The cell below demonstrates bytecode inspection without requiring a separate file.

````python
import dis

def add_numbers(left, right):
    return left + right

print("Function result:", add_numbers(4, 6))
print("\nBytecode instructions:")
dis.dis(add_numbers)
````

**What to notice:** `dis.dis()` displays implementation-level instructions. The exact instructions can vary by Python version, so focus on the idea rather than memorizing instruction names.

## Part 7 — Program structure and the main guard

A small script commonly has this organization:

1. A module docstring.
2. Imports.
3. Constants.
4. Function or class definitions.
5. A main entry point.

The `if __name__ == "__main__":` guard runs its block when the file is executed directly, but not when it is imported as a module. In a notebook, `__name__` is usually `"__main__"`, so the next example is illustrative of the guard's condition.

````python
"""Demonstrate a simple main entry point."""

APP_NAME = "Fundamentals Practice"


def show_welcome():
    """Display a short welcome message."""
    print(f"Welcome to {APP_NAME}!")


def main():
    """Run the program's main actions."""
    show_welcome()


if __name__ == "__main__":
    main()
````
