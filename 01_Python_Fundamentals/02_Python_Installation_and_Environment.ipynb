## 1. What is a Python Installation?

A Python installation provides the software required to execute Python programs.

A typical Python installation contains:

- Python interpreter
- Python standard library
- Package management tools
- Python executable
- Supporting files and configuration

After Python is installed, the operating system can use the Python executable to run Python programs.

A simplified view is:

```text
Operating System
       │
       ▼
Python Installation
       │
       ├── Python Interpreter
       ├── Standard Library
       ├── pip
       └── Supporting Components

## 2. Python Interpreter

The Python interpreter is the program that executes Python code.

For example:


````python
print("Hello, Python!")
````

When this code is executed, the Python interpreter processes the code and produces the output.

The interpreter can be accessed from a terminal.

A common command is:
```
python
```
On some Linux systems, the command used to start Python 3 is:
```
python3
```
The exact command depends on the operating system and how Python was installed.


## 3. Checking Whether Python is Installed

The first step after installing Python is to verify that it is available from the terminal.

Common commands are:

python --version

or:
```
python3 --version
```
A successful command displays the installed Python version.
For example:
```
Python 3.x.x

The exact version depends on the Python installation.

Checking the version is useful because Python programs and packages can depend on specific Python versions.

````python
import sys

print(sys.version)
````

## 4. Finding the Python Executable

A Python installation contains a Python executable.

The executable is the program that starts the Python interpreter.

Inside Python, the location of the currently running interpreter can be checked using:


````python
import sys

print(sys.executable)
````

This is especially important when working with multiple Python installations or virtual environments.

For example, a system Python installation and a project's virtual environment may use different Python executables.

Therefore, checking sys.executable helps confirm which Python environment is actually running the code.

## 5. Python Version

Python is released in different versions.

A Python version is commonly represented using:

```text
MAJOR.MINOR.PATCH

For example:

3.14.4

Here:
```text
- 3 is the major version.
- 14 is the minor version.
- 4 is the patch version.
```
The version matters because:
```text
- Language features can change between versions.
- Libraries may support specific Python versions.
- Projects may require a particular Python version.
- Different projects may use different Python versions.
```
For professional development, always know which Python version your project is using.

## 6. Python Package Manager — pip

`pip` is a package management tool commonly used with Python.

It allows Python packages to be installed, upgraded, and removed.

For example:

```bash
pip install requests

This installs the requests package into the currently active Python environment.

A package manager is useful because Python projects often depend on libraries that are not part of the Python standard library.

Common pip operations include:

```
pip install package_name
pip uninstall package_name
pip list
pip show package_name

The exact behavior of these commands depends on the active Python environment.


## 7. Python Packages

A Python package is a collection of Python code that can be installed and used by other Python programs.

For example, a project might use packages for:

- HTTP requests
- Data processing
- Machine Learning
- Web development
- Testing
- Database access

Examples of commonly used Python packages include:

- NumPy
- Pandas
- Matplotlib
- Requests
- scikit-learn

Packages extend Python's capabilities without requiring programmers to implement every feature from scratch.

## 8. Standard Library vs External Packages

Python provides a large **standard library**.

The standard library contains modules that are available as part of the Python installation.

Examples include:

```python
import math
import os
import sys
import json

External packages are installed separately.

For example:
```
pip install requests

A useful distinction is:

```text
Python Installation
│
├── Standard Library
│
└── External Packages
       ├── NumPy
       ├── Pandas
       ├── Requests
       └── Other Packages

Not every library used by a Python project is part of the standard library.


## 9. Why Virtual Environments are Needed

Different Python projects can require different package versions.

For example:

```text
Project A
    requests 2.x
    numpy 1.x

Project B
    requests 3.x
    numpy 2.x

If all packages are installed into one global Python environment, different projects can interfere with each other.

A virtual environment provides an isolated Python environment for a project.

The idea is:

```text
System Python
│
├── Project A
│      └── Virtual Environment
│
├── Project B
│      └── Virtual Environment
│
└── Project C
       └── Virtual Environment

Each project can therefore manage its own dependencies.

### Important Principle
A Python project should generally use its own virtual environment instead of relying on a shared global environment.


## 10. Creating a Virtual Environment

Python provides the `venv` module for creating virtual environments.

A common command is:

```bash
python -m venv env

Here:
```text
- python starts the Python interpreter.
- -m tells Python to run a module.
- venv is Python's built-in virtual environment module.
- .venv is the directory where the environment will be created.
```
The directory name env is a common convention.



## 11. Activating a Virtual Environment

After creating a virtual environment, it must be activated before using it as the project's active environment.

### Linux / WSL

```bash
source env/bin/activate

#### Windows PowerShell

.venv\Scripts\Activate.ps1

#### Windows Command Prompt

.venv\Scripts\activate.bat

When the environment is active, the terminal commonly displays its name before the command prompt.

For example:

(.venv) user@computer:~/project$

The exact prompt depends on the shell and operating system.


## 12. Verifying the Active Environment

After activating a virtual environment, we should verify that Python is coming from the expected environment.

Inside Python:



````python
import sys

print(sys.executable)
````

The resulting path should point to the project's virtual environment.

## 13. Installing Packages in a Virtual Environment

After activating the virtual environment, packages can be installed using `pip`.

For example:

```bash
pip install requests

The package is installed into the active environment.

This means that the package belongs to the project environment rather than being installed globally.

We can view installed packages using:

```
pip list

We can inspect a specific package using:

```
pip show requests


## 14. Requirements Files

Python projects often maintain a list of required packages.

A common file name is:

```text
requirements.txt

For example:
requests
numpy
pandas

Packages listed in the file can be installed using:
pip install -r requirements.txt

A requirements file makes it easier to recreate a project's dependencies in another environment.
The general workflow is:

Project
   ↓
requirements.txt
   ↓
Create Virtual Environment
   ↓
Activate Virtual Environment
   ↓
Install Dependencies
   ↓
Run Project


## 15. Deactivating a Virtual Environment

When finished working with a virtual environment, it can be deactivated.

Use:

```bash
deactivate

After deactivation, the terminal returns to the previously active Python environment.

Deactivation does not delete the virtual environment.

It only removes the virtual environment from the current shell session.


## 16. Python Environment Workflow

A typical Python project workflow is:

```text
Create Project
     ↓
Install / Select Python
     ↓
Create Virtual Environment
     ↓
Activate Virtual Environment
     ↓
Install Dependencies
     ↓
Write Python Code
     ↓
Run and Test Code
     ↓
Save Dependency Information


## 17. Python Environment and Jupyter

Jupyter Notebook also needs to use the correct Python environment.

If Jupyter is connected to a different Python environment than the project, code may execute using a different Python interpreter than expected.

This can cause problems such as:

- A package appears to be installed but cannot be imported.
- The notebook uses a different Python version.
- Packages are installed into one environment while the notebook uses another.
- Project dependencies appear to be missing.

Therefore, always verify the Python interpreter being used by the notebook.

The following code displays the interpreter used by the current notebook:

```python


````python
import sys

print(sys.executable)
````

## 18. Environment Variables

An environment variable is a named value provided by the operating system environment.

Environment variables are commonly used to configure applications.

Examples include:

- Paths
- Configuration values
- Runtime settings
- API configuration
- Environment-specific settings

Python can access environment variables using the `os` module.

Example:

```python


````python
import os

print(os.environ.get("PATH"))
````

Environment variables are especially useful when configuration should not be hard-coded directly into source code.

Sensitive information such as passwords, private keys, and API credentials should not be committed directly to source-control repositories.


## 19. PATH and Python

The `PATH` environment variable tells the operating system where it can look for executable programs.

When Python is correctly available through `PATH`, commands such as:

```bash
python --version

can be executed from the terminal without specifying the complete path to the Python executable.

The exact configuration of PATH depends on the operating system.

Understanding PATH becomes important when:
```text
- Multiple Python installations exist.
- A command cannot be found.
- A virtual environment is activated.
- Different shells behave differently.



## 20. Global Python vs Project Python

There are two important concepts to distinguish.

### Global / System Python

This is a Python installation available to the operating system or user.

It may be used by multiple applications or projects.

### Project Python

This is the Python interpreter provided by a project's virtual environment.

For example:

```text
Project
│
├── env/
│    └── Python Interpreter
│
└── Python Source Code



## 21. Environment Verification Checklist

Before starting development, verify:

### Python

```bash
python --version

Python Executable
```bash
python -c "import sys; print(sys.executable)"

pip
```bash
python -m pip --version

Installed Packages
```bash
python -m pip list

Virtual Environment

Confirm that the Python executable points to the project's env directory.

Using:
```bash

python -m pip

is often preferable to calling pip directly because it explicitly associates pip with the Python interpreter being used.
