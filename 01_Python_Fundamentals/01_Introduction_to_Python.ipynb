## 3. Characteristics of Python

Python has several important characteristics.

### High-Level

Python provides abstractions over many low-level computer operations.

### General-Purpose

Python can be used to build different types of programs and solve problems in different domains.

### Interpreted

Python programs are executed through the Python runtime rather than being directly executed by the computer's CPU as raw source code.

The Python implementation handles the process of translating and executing Python code.

### Dynamically Typed

Python variables do not require the programmer to explicitly declare their data type when assigning a value.

For example:


````python
x = 10
````

The variable x refers to an integer object.
Later, the same variable name can refer to another type of object:

````python
x = "Python"
````

### Object-Oriented

It provides concepts such as:
```text

- Inheritance
- Encapsulation
- Polymorphism
```
Python also supports other programming styles.

### Cross-Platform
Python programs can generally be executed on different operating systems when the required Python environment and dependencies are available.
Common platforms include:
```text

- Linux
- macOS
```

Python is open-source software and its source code is publicly available.

## 7. How Python Code Runs

A Python program begins as source code written by the programmer.

For example:


````python
print("Hello, Python!")
````

When using CPython, the source code is processed by the Python implementation.

A simplified view is:
```text
Python Source Code
        ↓
Python Interpreter
        ↓
Bytecode
        ↓
Python Virtual Machine
        ↓
Program Execution
```
This is a simplified conceptual model.
The important idea is that Python source code is processed by the Python runtime, which handles the execution of the program.

### Source Code
Source code is the code written by the programmer.


Example:

````python
print("Hello")
````

### Bytecode
Python source code is compiled into an intermediate representation called bytecode.
Bytecode is not the same as machine code executed directly by the CPU.

### Python Virtual Machine
The Python Virtual Machine executes Python bytecode as part of the Python runtime.
