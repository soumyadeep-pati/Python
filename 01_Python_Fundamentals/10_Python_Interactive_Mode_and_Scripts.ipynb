## 1. Two common ways to run Python code

Python code is commonly run in two ways:

1. **Interactive mode:** enter code at the Python prompt and see results as you work.
2. **Script mode:** save code in a `.py` file and ask Python to run that file.

For example, both approaches can calculate a total. Interactive mode is useful for trying something quickly; a script is useful when you want to save, reuse, test, or share a program.

## 2. What is the Python interpreter?

The **Python interpreter** is the software that processes Python code and runs it.

When you type a command such as `python` or `python3` in a terminal, your operating system attempts to start a Python interpreter using that command. The exact command depends on your operating system and installation.

Check the installed version from a terminal:

```bash
python --version
```

On some systems, use:

```bash
python3 --version
```

Example output:

```text
Python 3.x.y
```

The exact version number will depend on your installation.

**Important:** A terminal is not itself Python. The terminal runs commands; one of those commands can start Python.

## 3. What is interactive mode?

In **interactive mode**, you enter Python code directly into the interpreter. Python evaluates the input and displays results when appropriate.

Start it from a terminal:

```bash
python
```

or, where appropriate:

```bash
python3
```

A prompt similar to this appears:

```text
Python 3.x.y ...
>>>
```

The `>>>` symbol is the primary Python prompt. It means Python is ready for input.

To leave interactive mode, you can type:

```python
exit()
```

You can also use `quit()`. On many terminals, `Ctrl+D` exits Python on Linux/macOS and `Ctrl+Z` followed by Enter exits on Windows.

## 4. The REPL

REPL stands for:

- **Read** — read the code you enter.
- **Evaluate/Execute** — process it.
- **Print** — display a result when appropriate.
- **Loop** — return to the prompt for more input.

For example, in the Python prompt:

```text
>>> 2 + 3
5
>>> 10 * 4
40
>>> "Hello, Python!"
'Hello, Python!'
```

The REPL is convenient because you can try one expression, inspect its result, and immediately try another.

## 5. Expressions and statements in the REPL

An **expression** produces a value. In the standard Python REPL, the value of an expression entered by itself is displayed.

```text
>>> 7 * 6
42
```

An **assignment statement** associates a name with a value. Assignment by itself does not normally display the assigned value:

```text
>>> answer = 42
>>> answer
42
```

The first line assigns a value to `answer`. The second line evaluates the expression `answer`, so the REPL displays its value.

Try these lines one at a time in your own Python prompt. The `>>>` symbols are prompts; do not type them as part of a normal command.

## 6. Interactive mode versus a script

A Python script is a text file that contains Python code, usually saved with the `.py` extension.

Example script contents:

```python
print("Hello from a script!")
total = 10 + 20
print(total)
```

Save this as `hello.py`. Then run it from the directory containing the file:

```bash
python hello.py
```

On systems where the executable is named `python3`, use:

```bash
python3 hello.py
```

Expected output:

```text
Hello from a script!
30
```

Unlike the REPL, running a script does not automatically display every expression's value. Use `print()` when you want output to appear.

## 7. Why does an expression display in the REPL but not in a script?

In the REPL:

```text
>>> 10 + 20
30
```

The REPL displays the value of a standalone expression.

In a script file:

```python
10 + 20
```

Python evaluates the expression, but the result is not automatically printed. The script normally finishes without visible output.

To display the result, write:

```python
print(10 + 20)
```

Expected output:

```text
30
```

This is one of the most important practical differences between interactive execution and script execution.

## 8. Creating and running a script from the terminal

You can create a script using a text editor such as VS Code. You can also create a small file directly from a shell.

For example, on Linux/Ubuntu, this command writes a file named `hello.py`:

```bash
cat > hello.py <<'PY'
print("Hello from Python")
print(5 + 7)
PY
```

Then run it:

```bash
python hello.py
```

Expected output:

```text
Hello from Python
12
```

The `cat` command and `<<'PY'` syntax belong to the shell, not to Python. They are just one way to write a file from a Linux terminal. In VS Code, create a new file, enter the Python code, and save it as `hello.py` instead.

## 9. Running scripts in VS Code

A simple workflow in VS Code is:

1. Open your Python learning folder.
2. Create a new file named `hello.py`.
3. Write Python code in the file.
4. Save the file.
5. Open the integrated terminal in the correct folder.
6. Run `python hello.py` or `python3 hello.py`, depending on your environment.

Example file:

```python
message = "Learning Python"
print(message)

first_number = 12
second_number = 8
print(first_number + second_number)
```

Expected output:

```text
Learning Python
20
```

**Tip:** Make sure VS Code is using the Python interpreter or virtual environment you intend to use. A different interpreter can have different installed packages.

## 10. Multiple lines in interactive mode

The REPL accepts single-line input and can also accept multi-line compound statements, such as a function or an `if` block.

For example, type the following interactively:

```python
for number in range(3):
    print(number)
```

After the indented body, press Enter on a blank line if the prompt requires it to finish the block.

Expected output:

```text
0
1
2
```

The prompt may change from `>>>` to `...` while Python is waiting for the rest of a multi-line statement. The `...` prompt indicates continuation input.

## 11. Errors in interactive mode and scripts

Both interactive code and scripts can produce errors.

For example:

```python
print(10 / 0)
```

This raises a `ZeroDivisionError` because division by zero is not allowed.

In interactive mode, you see the traceback and then return to the prompt, where you can try another command. When a script encounters an unhandled exception, execution of that script normally stops and Python displays a traceback.

The same underlying Python error can occur in either mode; the surrounding workflow differs.

## 12. When should you use each mode?

| Situation | Recommended approach | Why |
|---|---|---|
| Quickly test a calculation | REPL | Immediate feedback |
| Check how an expression behaves | REPL | Easy to experiment |
| Save a solution for later | Script | Code is stored in a file |
| Build a small reusable program | Script | Easy to rerun and share |
| Run a sequence of steps repeatedly | Script | The same file can be executed again |
| Explore code one step at a time | REPL or notebook | Results are easy to inspect |

Jupyter Notebook is another interactive environment. It organizes code and explanatory text into cells, so it is especially useful for learning, data analysis, and mathematical experiments.

