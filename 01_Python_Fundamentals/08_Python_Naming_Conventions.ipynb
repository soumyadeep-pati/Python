## 1. What is a naming convention?

A **naming convention** is an agreed way of writing names in a program so that code is consistent and easy to understand.

Consider these names:

- `student_name`
- `StudentName`
- `studentname`

Python can use all three as identifiers in suitable contexts, but Python programmers generally use different styles for different purposes.

For example, `student_name` is the usual style for a variable or function, while `StudentRecord` is the usual style for a class.

### Rules versus conventions

- **Naming rules** determine whether Python accepts a name at all.
- **Naming conventions** are recommendations that help programmers write consistent, readable code.

A name can be valid Python but still be a poor choice.

```python
x = 85
student_marks = 85
```

Both assignments are valid. However, `student_marks` communicates the meaning more clearly than `x`.

````python
x = 85
student_marks = 85

print(x)
print(student_marks)
````

**Expected output**

```text
85
85
```

The output is identical, but the second name tells a reader more about the purpose of the value.

## 2. The main Python naming styles

Python commonly uses these styles:

| Style | Example | Typical use |
|---|---|---|
| `snake_case` | `student_name` | Variables, functions, and modules |
| `PascalCase` | `StudentRecord` | Classes |
| `UPPER_SNAKE_CASE` | `MAX_RETRIES` | Constants |
| `mixedCase` | `studentName` | Not the usual style for ordinary Python names |
| `_single_leading_underscore` | `_internal_value` | Indicates a non-public or internal name by convention |
| `__double_leading_underscore` | `__private_attribute` | Has a special name-mangling effect in class definitions |
| `trailing_underscore_` | `class_` | Can avoid a clash with a keyword |

For most beginner code, concentrate first on `snake_case`, `PascalCase`, and `UPPER_SNAKE_CASE`.

## 3. Variable names: use `snake_case`

Use lowercase words separated by underscores for ordinary variable names.

**Recommended:**

```python
student_name = "Asha"
total_marks = 450
is_logged_in = True
average_temperature = 28.5
```

**Avoid inconsistent styles in ordinary Python code:**

```python
studentName = "Asha"
TotalMarks = 450
averagetemperature = 28.5
```

These names may be accepted by Python, but they do not follow the usual variable-naming style recommended by PEP 8.

### Choose meaningful names

A name should help explain what the value represents.

| Unclear or overly short | Clearer alternative |
|---|---|
| `n` | `student_count` |
| `a` | `first_number` |
| `t` | `total_price` |
| `tmp` | `temporary_result` |
| `data` | `student_records` (when that is what it contains) |

Short names can be reasonable for small, familiar contexts, such as `i` for a loop index. Avoid making every name unnecessarily long, though: clarity is the goal, not maximum length.

````python
student_name = "Asha"
math_marks = 92
science_marks = 88
total_marks = math_marks + science_marks

print(student_name)
print(total_marks)
````

**Expected output**

```text
Asha
180
```

The names describe the information, so the code is easier to read without extra explanation.

## 4. Function names: use `snake_case`

Function names should normally be lowercase, with words separated by underscores. A good function name often describes an action.

Examples of recommended names:

- `calculate_total()`
- `get_student_details()`
- `print_report()`
- `is_valid_email()`

Names such as `CalculateTotal()` or `getStudentDetails()` are valid identifiers in many contexts, but they are not the standard Python function style.

The following is a small example. Function definitions are introduced fully in a later section.

```python
def calculate_total(price, quantity):
    return price * quantity

print(calculate_total(50, 3))
```

**Expected output**

```text
150
```

`calculate_total` describes what the function does. The parameters `price` and `quantity` also communicate the inputs' meanings.

## 5. Class names: use `PascalCase`

A **class name** normally starts each word with a capital letter and has no underscores between words. This style is called `PascalCase` (also commonly called CapWords in Python guidance).

Examples:

- `Student`
- `BankAccount`
- `MachineLearningModel`
- `WeatherReport`

Avoid using the ordinary variable style for class names:

- `student_record`
- `bank_account`

Example:

```python
class StudentRecord:
    pass
```

This defines a class named `StudentRecord`. The `pass` statement is a placeholder that does nothing; it allows the class body to be syntactically complete.

**Expected result:** the cell runs without producing output.

## 6. Constant names: use `UPPER_SNAKE_CASE`

A **constant** is a value that a program intends to keep unchanged. Python does not generally enforce constants just because of their names. Instead, uppercase naming communicates the programmer's intention.

Examples:

```python
PI = 3.14159
MAX_LOGIN_ATTEMPTS = 5
DEFAULT_TIMEOUT_SECONDS = 30
COMPANY_NAME = "Example Ltd."
```

Avoid naming an intended constant in the same way as an ordinary variable if doing so would make its role unclear:

```python
max_login_attempts = 5
```

That lowercase name is still valid Python; the uppercase form is simply the conventional signal for a constant.

**Important:** `MAX_LOGIN_ATTEMPTS = 5` does not make reassignment impossible. Python allows code to assign another value to that name. The convention relies on programmers respecting the intention.

````python
PI = 3.14159
MAX_LOGIN_ATTEMPTS = 5

print(PI)
print(MAX_LOGIN_ATTEMPTS)
````

**Expected output**

```text
3.14159
5
```

## 7. Module and package names

A **module** is commonly a Python `.py` file. A **package** is a way of organizing related modules.

Use short, lowercase module names. Underscores may be used when they improve readability.

Examples:

- `math_helpers.py`
- `student_records.py`
- `data_loader.py`

Avoid spaces and prefer not to use names that shadow important standard-library modules or commonly used third-party packages. For example, naming your own file `random.py` can interfere with importing Python's `random` module when running code from that directory.

Package directories are also usually lowercase, with underscores where helpful:

```text
student_app/
    __init__.py
    student_records.py
    report_helpers.py
```

A package does not always need an `__init__.py` file in modern Python, but the file is still common and can be useful.

## 8. Avoid built-in names and confusing names

Python has built-in names such as `list`, `str`, `sum`, `type`, and `max`. Reusing one of these as a variable name can hide the built-in in the relevant scope.

**Avoid:**

```python
list = [10, 20, 30]
sum = 100
```

Prefer:

```python
numbers = [10, 20, 30]
total = 100
```

Why? After assigning a value to `list`, a later call such as `list("abc")` may fail in that scope because `list` no longer refers to the built-in list constructor.

Also avoid names that are too vague, misleading, or needlessly abbreviated. A name should match what it represents.

````python
numbers = [10, 20, 30]
total = sum(numbers)

print(total)
print(list("abc"))
````

**Expected output**

```text
60
['a', 'b', 'c']
```

Here, `numbers` and `total` do not hide the built-in names `sum` or `list`.

## 9. Leading and trailing underscores

Underscores can communicate special intent.

### One leading underscore: `_name`

A single leading underscore conventionally means “internal use” or “not part of the public interface.”

```python
_internal_cache = {}
```

This is a convention, not a security boundary. It does not prevent access.

### Two leading underscores in a class: `__name`

Inside a class, a name beginning with two underscores and not ending with two underscores generally triggers **name mangling**. Python changes the attribute's stored name to help avoid accidental clashes with subclasses.

This is an advanced detail; for now, know that `__name` is not simply a stronger version of privacy.

### One trailing underscore: `name_`

A trailing underscore is useful when a natural name conflicts with a Python keyword.

```python
class_ = "Computer Science"
```

`class` is a keyword, so it cannot be used as an ordinary identifier. `class_` is valid.

### Double underscores at both ends: `__name__`

Names such as `__init__` and `__name__` are special names defined by Python or its conventions. They are often called *dunder* names (double underscore). Do not invent them for ordinary variables; use them only when following the relevant Python feature's documentation.

## 10. Keep naming consistent

Consistent names reduce the mental effort needed to understand code.

**Inconsistent:**

```python
student_name = "Asha"
studentAge = 20
StudentCity = "Kolkata"
```

**Consistent:**

```python
student_name = "Asha"
student_age = 20
student_city = "Kolkata"
```

All three names describe related pieces of information and follow the same `snake_case` convention.

Consistency is particularly important in larger projects, where many files and contributors may be involved.

## 11. Naming conventions at a glance

| Program element | Recommended style | Example |
|---|---|---|
| Variable | `snake_case` | `total_marks` |
| Function | `snake_case` | `calculate_average()` |
| Class | `PascalCase` | `StudentRecord` |
| Constant | `UPPER_SNAKE_CASE` | `MAX_ATTEMPTS` |
| Module | lowercase, underscores if needed | `data_loader.py` |
| Package | lowercase, underscores if needed | `student_tools` |
| Internal-use name | one leading underscore | `_helper_value` |
| Keyword-conflicting name | trailing underscore | `class_` |

These are the common conventions described by PEP 8, Python's style guide. Project-specific guidelines can add further requirements, so follow the conventions used by the project you are contributing to.

## 12. Common mistakes

1. **Using `PascalCase` for every name.** Reserve it mainly for class names.
2. **Using spaces in identifiers.** Write `student_name`, not `student name`.
3. **Starting a name with a digit.** `2nd_place` is invalid; `place_2` is valid.
4. **Using a keyword as an identifier.** `for = 5` is invalid; `for_count = 5` is valid.
5. **Using names that hide built-ins.** Prefer `numbers` over `list` and `total` over `sum`.
6. **Using cryptic abbreviations everywhere.** Prefer `total_price` over `tp` when the meaning is not obvious.
7. **Thinking uppercase makes a value immutable.** `MAX_RETRIES` is a convention, not an enforcement mechanism.
8. **Changing style from line to line.** Choose a convention and apply it consistently.
