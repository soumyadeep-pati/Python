## 1. What is a keyword?

A **keyword** is a word that Python reserves for a special purpose in the language. For example, `if` starts a conditional statement, `for` starts a loop, and `def` defines a function.

Because Python gives these words a special meaning, you cannot normally use them as variable names, function names, or other identifiers.

Examples of Python keywords include:

- `if`, `elif`, `else`
- `for`, `while`, `break`, `continue`
- `def`, `return`, `class`
- `True`, `False`, `None`
- `and`, `or`, `not`
- `try`, `except`, `finally`, `raise`

````python
age = 20

if age >= 18:
    print("Adult")
else:
    print("Under 18")
````

## 2. Why keywords are reserved

Python needs to recognize the structure and meaning of your program. If `if` could freely be used as an ordinary variable name, Python could not reliably tell whether you meant a condition or a variable.

This is invalid:

```python
if = 10
```

Python raises a `SyntaxError` because `if` is a keyword. Use a different identifier, such as `if_count` or `condition`.

````python
condition = True
if_count = 10

print(condition)
print(if_count)
````

Python provides a standard-library module named `keyword`. Its `kwlist` attribute contains the keywords recognized by the running interpreter.

The exact list can vary by Python version. Some words are **soft keywords**: they have special meaning only in certain language contexts and can be used as identifiers in other contexts.

````python
import keyword

print("Python version:", sys.version.split()[0] if "sys" in globals() else __import__("sys").version.split()[0])
print("Number of hard keywords:", len(keyword.kwlist))
print(keyword.kwlist)
````

## 4. Hard keywords and soft keywords

A hard keyword is reserved in the language wherever Python parses an identifier. A soft keyword is special only in a specific syntactic context.

In supported Python versions, `keyword.softkwlist` lists the soft keywords. This attribute may be empty or differ depending on the Python version.

````python
import keyword

print("Hard keywords:")
print(keyword.kwlist)

print("\nSoft keywords in this interpreter:")
print(getattr(keyword, "softkwlist", []))
````

## 5. Check whether a word is a keyword

Use `keyword.iskeyword(word)` to test whether a string is a hard keyword. Use `keyword.issoftkeyword(word)` when available to check for a soft keyword.

These checks inspect Python's language keywords; they do not check whether a name is already defined in your program.

````python
import keyword

words = ["if", "class", "score", "return", "match"]

for word in words:
    print(f"{word:10} hard keyword? {keyword.iskeyword(word)}")
````

````python
import keyword

for word in ["match", "case", "type", "score"]:
    is_soft = getattr(keyword, "issoftkeyword", lambda value: False)(word)
    print(f"{word:10} soft keyword? {is_soft}")
````

## 6. What is an identifier?

An **identifier** is a name you create to identify something in a program, such as:

- a variable: `student_name`
- a function: `calculate_total`
- a class: `Student`
- a module: `data_loader`

Identifiers make your code understandable. A meaningful name is usually better than a name like `x1` or `temp2`, except where short names are conventional and clear.

````python
student_name = "Riya"
total_marks = 450

def calculate_average(total, subjects):
    """Return the average mark per subject."""
    return total / subjects

print(student_name)
print(calculate_average(total_marks, 5))
````

## 7. Identifier rule 1: allowed characters

In ordinary beginner code, identifiers can contain:

- letters (`A`–`Z`, `a`–`z`);
- digits (`0`–`9`), but not as the first character;
- underscores (`_`).

Valid examples:

```python
name = "Riya"
student2 = "Aman"
total_marks = 450
_private_value = 10
```

Invalid examples:

```python
2student = "Aman"  # Starts with a digit
total-marks = 450  # Hyphen is interpreted as subtraction
user name = "Riya" # Contains a space
```

````python
name = "Riya"
student2 = "Aman"
total_marks = 450
_private_value = 10

print(name, student2, total_marks, _private_value)
````

## 8. Identifier rule 2: identifiers cannot start with a digit

A name may contain digits after its first character, but it cannot begin with a digit.

Valid:
```python
student1 = "Asha"
mark2 = 95
```

Invalid:
```python
1student = "Asha"
```

A leading digit makes the expression ambiguous to the parser, so Python rejects it as a syntax error.

````python
student1 = "Asha"
mark2 = 95

print(student1)
print(mark2)

# To test an invalid identifier without breaking this notebook,
# run it as text using compile() inside try/except.
invalid_code = "1student = 'Asha'"
try:
    compile(invalid_code, "<example>", "exec")
except SyntaxError as error:
    print("Invalid identifier example:", error.msg)
````

## 9. Identifier rule 3: no spaces or hyphens

Spaces separate tokens in Python code. Hyphens mean subtraction, so they cannot be used inside an identifier.

Use underscores instead:

- `first name` → invalid
- `first-name` → parsed as subtraction, not a single name
- `first_name` → valid

````python
first_name = "Asha"
second_name = "Rao"

full_name = first_name + " " + second_name
print(full_name)

for invalid_code in ["first name = 'Asha'", "first-name = 'Asha'"]:
    try:
        compile(invalid_code, "<example>", "exec")
    except SyntaxError as error:
        print(repr(invalid_code), "->", error.msg)
````

## 10. Identifier rule 4: keywords cannot be identifiers

Do not use hard keywords as variable, function, or class names.

Bad names include `if`, `for`, `return`, `class`, and `True`.

Choose names that describe the value or purpose instead, such as `if_condition`, `loop_count`, `result`, or `is_valid`.

````python
result = 42
loop_count = 3
is_valid = True

print(result, loop_count, is_valid)

import keyword
for word in ["if", "for", "return", "class", "True"]:
    print(word, "is a keyword:", keyword.iskeyword(word))
````

## 11. Identifier rule 5: names are case-sensitive

Python treats uppercase and lowercase letters as different characters. Therefore, `score`, `Score`, and `SCORE` are three different names.

Case sensitivity is a syntax/runtime fact, not just a style convention.

````python
score = 10
Score = 20
SCORE = 30

print("score =", score)
print("Score =", Score)
print("SCORE =", SCORE)
````

## 12. Unicode identifiers

Python 3 supports many Unicode letters in identifiers, so names are not limited to English A–Z. However, for shared projects and beginner code, simple English names are often easier for teams to read, type, search, and maintain.

You may encounter Unicode identifiers in existing code, but use them deliberately rather than just because they are allowed.

````python
# Python 3 permits Unicode letters in identifiers.
café = "coffee"
print(café)

# A more portable team convention is often:
cafe_name = "coffee"
print(cafe_name)
````

## 13. Underscores and special naming patterns

Underscores have useful conventions:

- `total_marks`: common style for variables and functions.
- `_internal_value`: a convention that signals “internal use”; it does not make the name truly private.
- `__name` inside a class: triggers name mangling in many contexts.
- `__name__`: commonly used for Python-defined special names, such as `__init__`.

Avoid inventing your own double-underscore names unless you understand the special behavior involved.

````python
ordinary_name = "visible"
_internal_name = "conventionally internal"

class Example:
    def __init__(self, value):
        self.__value = value

    def get_value(self):
        return self.__value

obj = Example(7)
print(ordinary_name)
print(_internal_name)
print(obj.get_value())
print([name for name in vars(obj) if "value" in name])
````

## 14. The special name `__name__`

Python sets the special variable `__name__` for each module. When a Python file is run directly, its `__name__` is usually `"__main__"`. When imported, it usually has the module's name.

This pattern lets a file provide reusable functions while also having a section that runs only when the file is executed directly.

````python
print("Current module name:", __name__)

def main():
    """Run the program's main demonstration."""
    print("Hello from main()")

if __name__ == "__main__":
    main()
````

## 15. Naming variables: `snake_case`

PEP 8, Python's widely used style guide, recommends `snake_case` for variables and functions. Words are lowercase and separated by underscores.

Examples:

- `student_name`
- `total_price`
- `calculate_average`
- `is_authenticated`

This is a style recommendation rather than a syntax rule. Python will accept many other names, but consistent conventions make code easier to read.

````python
student_name = "Riya"
total_price = 799.50

def calculate_average(total, count):
    """Calculate the average of a total divided by a count."""
    return total / count

print(student_name)
print(calculate_average(90, 3))
````

## 16. Naming classes: `CapWords` / `PascalCase`

PEP 8 recommends `CapWords` for class names: capitalize the first letter of each word and do not use underscores between words.

Examples: `StudentRecord`, `BankAccount`, `DecisionTree`.

By convention, classes often represent things or concepts, while functions describe actions.

````python
class StudentRecord:
    """Store a student's name and score."""

    def __init__(self, student_name, score):
        self.student_name = student_name
        self.score = score

student = StudentRecord("Riya", 92)
print(student.student_name, student.score)
````

## 17. Naming constants: `UPPER_SNAKE_CASE`

For module-level constants—values intended not to be reassigned—PEP 8 recommends uppercase words separated by underscores.

Examples: `MAX_RETRIES`, `SECONDS_PER_HOUR`, `DEFAULT_TIMEOUT`.

Python does not make a variable constant just because its name is uppercase. This is a convention that communicates intent.

````python
SECONDS_PER_MINUTE = 60
MINUTES_PER_HOUR = 60
SECONDS_PER_HOUR = SECONDS_PER_MINUTE * MINUTES_PER_HOUR

print(SECONDS_PER_HOUR)
````

## 18. Naming booleans clearly

Boolean variables hold `True` or `False`. Names beginning with `is_`, `has_`, `can_`, or `should_` often make conditions easier to read.

Compare:
```python
active = True
```
with:
```python
is_active = True
```

Both are valid. The second makes the expected boolean meaning more obvious in many contexts.

````python
is_logged_in = True
has_permission = False
can_edit = True

if is_logged_in and can_edit:
    print("Editing is allowed for this example.")
else:
    print("Editing is not allowed.")
````

## 19. Avoid built-in names

Names such as `list`, `str`, `sum`, `max`, and `id` are built-ins. Reusing them as variables hides the built-in function or type within the current scope.

This is legal in many cases but usually a poor idea:

```python
list = [1, 2, 3]
```

Prefer `numbers`, `items`, or `values`. If you accidentally overwrite a built-in in a notebook, restarting the kernel may be the simplest way to restore a clean environment.

````python
numbers = [3, 1, 4]
print("List:", numbers)
print("Length:", len(numbers))
print("Sum:", sum(numbers))
print("Largest:", max(numbers))
````

## 20. Naming conventions at a glance

| What you are naming | Common convention | Example |
|---|---|---|
| Variable | `snake_case` | `total_score` |
| Function | `snake_case` | `calculate_total()` |
| Class | `CapWords` | `StudentRecord` |
| Constant | `UPPER_SNAKE_CASE` | `MAX_RETRIES` |
| Internal-use name | Leading underscore | `_cache` |
| Special Python name | Double underscores at both ends | `__init__` |

These are widely used conventions, not all mandatory language rules.

## 21. `match` and `case`: soft keyword example

In modern Python, `match` and `case` are soft keywords in pattern-matching syntax. They have special meaning in that syntax, but they are not hard keywords in every context.

This is version-dependent: pattern matching was introduced in Python 3.10. If you use an older interpreter, the code below may not run.

````python
import keyword

print("match is a hard keyword:", keyword.iskeyword("match"))
print("match is a soft keyword:", getattr(keyword, "issoftkeyword", lambda x: False)("match"))

# In Python 3.10+, match/case can be used for pattern matching:
value = 2
match value:
    case 1:
        print("One")
    case 2:
        print("Two")
    case _:
        print("Something else")
````

## 22. Identifier validity with `str.isidentifier()`

Strings have an `isidentifier()` method that checks whether the string is a syntactically valid identifier. It does **not** check whether the identifier is a keyword.

That means `"if".isidentifier()` is `True` even though `if` cannot be used as an ordinary identifier because it is a keyword. Combine it with `keyword.iskeyword()` when checking hard keywords.

````python
import keyword

names = ["student_name", "2students", "first-name", "if", "_private", "café"]

for name in names:
    valid_identifier = name.isidentifier()
    is_keyword = keyword.iskeyword(name)
    usable_as_ordinary_name = valid_identifier and not is_keyword
    print(
        f"{name:15} identifier={valid_identifier!s:5} "
        f"keyword={is_keyword!s:5} usable={usable_as_ordinary_name}"
    )
````

## 23. Why `isidentifier()` does not check scope

A name may be syntactically valid but still not exist in your program. For example, `customer_total` is a valid identifier even before you assign anything to it.

Identifier validity asks, “Could this string be a name?” Scope and assignment ask, “Does this name refer to a value here?”

````python
name_text = "customer_total"
print("Valid identifier?", name_text.isidentifier())

# The name does not exist as a Python variable merely because the string is valid.
try:
    print(customer_total)
except NameError as error:
    print(type(error).__name__, "-", error)

customer_total = 250
print(customer_total)
````

## 24. Syntax errors versus naming-style issues

A syntax error prevents Python from parsing the code. A style issue may still run correctly but be harder to maintain.

- `2name = "A"` → syntax error.
- `student-name = "A"` → parsed as subtraction and then usually fails at runtime or with a syntax problem depending on context.
- `StudentName = "A"` → valid, but `student_name` is the common variable style.
- `x = 10` → valid, but a meaningful name may be clearer.

Do not confuse “Python allows it” with “it is a good naming choice.”

````python
# All of these assignments are valid Python syntax, but not all are equally clear.
x = 10
student_count = 10
StudentCount = 10

print(x, student_count, StudentCount)
````

## 25. Name collisions and scope

A name in a local scope can hide a name from an outer scope. This is one reason to choose descriptive names and avoid reusing built-in names.

The following example shows a local variable named `message`. It does not change the separate `message` variable outside the function.

````python
message = "outside"

def show_message():
    message = "inside"
    print("Inside function:", message)

show_message()
print("Outside function:", message)
````

## 27. Common mistakes to avoid

1. Using a keyword as a variable name.
2. Starting an identifier with a digit.
3. Including spaces or hyphens in a name.
4. Assuming `score` and `Score` are the same variable.
5. Overwriting built-ins such as `list` or `sum`.
6. Using unclear names when a descriptive name would help.
7. Treating naming conventions as syntax rules—or syntax rules as mere style.
8. Assuming `isidentifier()` detects keywords or checks whether a variable exists.

## 30. Mini-project: a simple identifier checker

Build a helper function that receives a string and reports whether it can be used as an ordinary identifier in the current Python interpreter.

Requirements:

- return `False` for non-identifiers;
- return `False` for hard keywords;
- return `True` for syntactically valid, non-keyword identifiers;
- print the result for several examples.

Note: this is a beginner-friendly check. Soft keywords are context-dependent, so this simple function does not reject them automatically.

````python
import keyword

def is_ordinary_identifier(name):
    """Return whether name is a valid non-keyword identifier."""
    return isinstance(name, str) and name.isidentifier() and not keyword.iskeyword(name)

examples = ["student_name", "if", "2students", "_cache", "first-name", "match"]

for example in examples:
    print(f"{example:15} -> {is_ordinary_identifier(example)}")
````

## 32. Summary

- **Keywords** are reserved words that form Python syntax.
- **Identifiers** are names for variables, functions, classes, and other objects.
- Identifiers cannot start with a digit, contain spaces, or use hyphens as part of a name.
- Python identifiers are case-sensitive.
- Use `keyword.kwlist`, `keyword.iskeyword()`, and `str.isidentifier()` to explore names.
- Follow conventions: `snake_case` for variables/functions, `CapWords` for classes, and `UPPER_SNAKE_CASE` for constants.
- Avoid shadowing built-ins and prefer meaningful names.
- A valid identifier is not necessarily a defined variable, and a style convention is not necessarily a syntax rule.

### Recommended practice
