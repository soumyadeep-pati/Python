## 1. Why comments and documentation matter

Code tells Python what to do. Comments and documentation tell people how to understand, use, and maintain that code.

Consider these two versions:

```python
total = price * quantity
```

The expression is already readable. A comment like `# Multiply price by quantity` adds little.

But this version may need context:

```python
# The supplier's discount applies before sales tax.
subtotal = unit_price * quantity
discounted_total = subtotal * (1 - discount_rate)
```

The comment explains a business rule that may not be obvious from the code alone.

**Core principle:** document intent, assumptions, constraints, and public behavior—not every keystroke.

````python
unit_price = 12.50
quantity = 4
discount_rate = 0.10

subtotal = unit_price * quantity
discounted_total = subtotal * (1 - discount_rate)

print(f"Subtotal: ₹{subtotal:.2f}")
print(f"After discount: ₹{discounted_total:.2f}")
````

## 2. Single-line comments

A comment begins with `#`. Python ignores the comment text when executing the program.


Comments can appear on a line by themselves or after a statement. Prefer a separate comment line when the explanation is more than a few words.

````python
# Store the number of attempts allowed.
max_attempts = 3

attempts_used = 1  # One attempt has already been used.

print("Attempts remaining:", max_attempts - attempts_used)
````

## 3. Comments should explain the reason

A redundant comment merely translates the code into English. A useful comment records the reason behind a decision.

**Redundant**
```python
# Set retries to 3
retries = 3
```

**More useful**
```python
# Keep retries low so a temporary outage does not delay the whole job.
retries = 3
```

The second comment is useful only if that explanation is actually true for the program. Never invent a rationale just to make a comment sound impressive.

````python
# Use a small batch size because this example simulates a memory-limited process.
batch_size = 16

# Keep the output deterministic so examples and tests are repeatable.
random_seed = 42

print(batch_size, random_seed)
````

## 9. What is a docstring?

A **docstring** is a string literal placed first in a module, function, class, or method body. Python stores it as documentation, accessible through the object's `.__doc__` attribute.

A docstring commonly uses triple quotes (`"""..."""`) because it can span multiple lines. Triple quotes alone do not make a string a docstring; its position matters.

````python
def greet(name):
    """Return a friendly greeting for a person."""
    return f"Hello, {name}!"

print(greet("Asha"))
print(greet.__doc__)
````

## 10. A string is not automatically a docstring

The first statement in a function body is special. A string elsewhere in the function is just an expression whose value is not used.

```python
def example():
    """This is the docstring."""
    "This is not the function's docstring."
```

Check `example.__doc__` to see what Python recorded.

````python
def example():
    """This is the function's docstring."""
    "This string is not the docstring."
    return "done"

print(example.__doc__)
print(example())
````

## 11. Writing a useful one-line docstring

For a short function, use one concise sentence that describes its behavior. Start with a verb where natural.

Good:
```python
def square(number):
    """Return the square of a number."""
```

Avoid vague descriptions such as `"""Does stuff."""` or descriptions that simply repeat the function name.

````python
def square(number):
    """Return the square of a number."""
    return number ** 2

print(square(5))
````

## 12. Multi-line docstrings

For more complex behavior, use a summary line, a blank line, and then details. The summary should remain useful when shown alone.

````python
def calculate_discount(price, rate):
    """Return a price after applying a discount.

    The discount rate is expressed as a decimal: 0.20 means 20%.
    This function does not modify the original inputs.
    """
    return price * (1 - rate)

print(calculate_discount(100, 0.20))
print(calculate_discount.__doc__)
````

## 13. Document parameters and return values

A reader should be able to answer these questions:

1. What arguments does the function expect?
2. What does it return?
3. What assumptions or limits apply?
4. Can it raise an exception or change external state?

The exact docstring format varies by project. Common styles include Google-style, NumPy-style, and reStructuredText/Sphinx-style documentation. Pick one style and use it consistently.

````python
def convert_minutes_to_seconds(minutes):
    """Convert a duration from minutes to seconds.

    Args:
        minutes: Duration in minutes; must be non-negative.

    Returns:
        The duration expressed in seconds.

    Raises:
        ValueError: If minutes is negative.
    """
    if minutes < 0:
        raise ValueError("minutes must be non-negative")
    return minutes * 60

print(convert_minutes_to_seconds(2.5))
````

## 14. Validate behavior, not just the documentation

A docstring may claim that an input must be non-negative. The implementation should enforce that rule if the rule is essential.

Documentation is not validation. Python will not automatically enforce rules written in a docstring.

````python
def safe_divide(numerator, denominator):
    """Divide numerator by denominator.

    Raises:
        ZeroDivisionError: If denominator is zero.
    """
    return numerator / denominator

print(safe_divide(10, 2))
try:
    safe_divide(10, 0)
except ZeroDivisionError as error:
    print(type(error).__name__, "-", error)
````

## 15. Document exceptions accurately

Mention exceptions that callers should reasonably know about, especially exceptions deliberately raised by your function. Do not list every theoretical exception if it adds no useful guidance.

For example, a function that calls `int(text)` may raise `ValueError` when `text` is not a valid integer.

````python
def parse_age(text):
    """Convert text to an integer age.

    Raises:
        ValueError: If text cannot be converted to an integer.
    """
    age = int(text)
    if age < 0:
        raise ValueError("age cannot be negative")
    return age

for candidate in ["21", "not a number", "-2"]:
    try:
        print(candidate, "->", parse_age(candidate))
    except ValueError as error:
        print(candidate, "->", error)
````

## 16. Document side effects

A side effect is an effect beyond returning a value—for example, writing a file, printing output, updating a database, or modifying a mutable object.

If a function changes something outside itself, make that behavior clear.

````python
def add_tag(tags, tag):
    """Append a tag to the provided list.

    Side effect:
        Mutates the original `tags` list.
    """
    tags.append(tag)
    return tags

my_tags = ["python", "learning"]
add_tag(my_tags, "documentation")
print(my_tags)
````

## 17. Modules also have docstrings

A module-level docstring goes at the very top of a Python file, before imports and other statements. It describes the file's purpose and important usage details.

For example, a reusable `temperature.py` module might begin with:

```python
"""Utilities for converting temperatures between common scales."""

def celsius_to_fahrenheit(celsius):
    """Convert Celsius to Fahrenheit."""
    return celsius * 9 / 5 + 32
```

A notebook cell cannot retroactively become the module docstring of a separate `.py` file. The example above is intended for a Python module.

## 18. Class and method docstrings

A class docstring describes what objects represent and how they are intended to be used. Method docstrings explain each method's behavior.

````python
class BankAccount:
    """Represent a simple bank account with a current balance."""

    def __init__(self, owner, opening_balance=0):
        """Create an account with an owner and opening balance."""
        self.owner = owner
        self.balance = opening_balance

    def deposit(self, amount):
        """Add a positive amount to the account balance."""
        if amount <= 0:
            raise ValueError("amount must be positive")
        self.balance += amount
        return self.balance

account = BankAccount("Asha", 100)
print(account.deposit(25))
print(BankAccount.__doc__)
print(BankAccount.deposit.__doc__)
````

## 20. Inspect `.__doc__`

The `.__doc__` attribute provides the docstring as a string. It is useful in scripts, tests, and simple introspection.

## 21. The `inspect` module

The standard-library `inspect` module can retrieve source code and documentation for many Python objects. Source retrieval may not work for every object, especially built-ins or code whose source is unavailable.

````python
import inspect

print("Signature:", inspect.signature(calculate_discount))
print("Docstring:", inspect.getdoc(calculate_discount))

print("\nSource (may be unavailable in some notebook environments):")
try:
    print(inspect.getsource(calculate_discount))
except (OSError, TypeError) as error:
    print("Source retrieval unavailable:", error)
````

## 22. Type hints are not comments

Type hints communicate the intended types of parameters and return values in a machine-readable form. Tools such as editors, linters, and type checkers can use them.

Python generally does **not** enforce type hints automatically at runtime.

```python
def repeat_text(text: str, times: int) -> str:
    ...
```

The docstring explains behavior; the type hints communicate intended types. They complement each other.

````python
def repeat_text(text: str, times: int) -> str:
    """Repeat text a specified number of times.

    Args:
        text: The text to repeat.
        times: Number of repetitions; expected to be non-negative.

    Returns:
        The repeated text.
    """
    if times < 0:
        raise ValueError("times must be non-negative")
    return text * times

print(repeat_text("ha", 3))
print(repeat_text.__annotations__)
````

## 23. Comments, docstrings, and type hints compared

| Tool | Main purpose | Example |
|---|---|---|
| Comment | Explain reasoning or a non-obvious decision | `# Cache this result to avoid repeated work.` |
| Docstring | Explain how to use a module, class, or function | `"""Return the median value."""` |
| Type hint | Communicate intended types | `def median(values: list[float]) -> float:` |

Use the tool that fits the information. These are complementary, not competing techniques.

## 24. Document assumptions and edge cases

A good API description makes important boundaries explicit: empty input, missing values, allowed ranges, rounding, units, and whether input is modified.

Consider a mean function. What should happen for an empty list? Make that decision deliberate and document it.

````python
def mean(values):
    """Return the arithmetic mean of a non-empty sequence of numbers.

    Raises:
        ValueError: If values is empty.
    """
    if not values:
        raise ValueError("values must not be empty")
    return sum(values) / len(values)

print(mean([2, 4, 6]))
try:
    mean([])
except ValueError as error:
    print(error)
````

## 25. Examples inside docstrings

Short examples can make a function easier to understand. Keep examples accurate and simple. If your project uses doctest, examples formatted as interactive Python sessions can also be checked automatically.

````python
def is_even(number):
    """Return whether an integer is even.

    Examples:
        >>> is_even(4)
        True
        >>> is_even(7)
        False
    """
    return number % 2 == 0

print(is_even(4))
print(is_even(7))
````

## 26. A brief introduction to `doctest`

`doctest` checks examples written in docstrings against actual output. This is useful when examples are simple and stable. Not every docstring needs doctest examples.

````python
import doctest

results = doctest.run_docstring_examples(
    is_even,
    globals(),
    name="is_even",
    verbose=True
)
````

## 27. Documentation and naming work together

A good name reduces the amount of explanation required. Compare `x = 86400` with `seconds_per_day = 86400`. A comment may explain the first, but the second is clearer immediately.

Do not use a comment as an excuse for cryptic variable names.

````python
seconds_per_day = 24 * 60 * 60
days = 3
duration_seconds = days * seconds_per_day

print(f"{days} days = {duration_seconds} seconds")
````

## 28. Avoid over-commenting

Too many comments create noise and become maintenance work. Prefer this order:

1. Choose clear names.
2. Break complex logic into well-named functions.
3. Add a comment where the *reason* or an unusual constraint is not obvious.
4. Add a docstring for reusable or public interfaces.

Comments are not a substitute for simplifying confusing code.

````python
def is_eligible_for_discount(total, minimum_total):
    """Return whether an order reaches the discount threshold."""
    return total >= minimum_total

print(is_eligible_for_discount(1200, 1000))
````

## 29. Document units and conventions

Numbers without units can be ambiguous. A value of `30` might mean seconds, minutes, Celsius, or percent. Use clear names and document units where they matter.

````python
def estimate_travel_time(distance_km, speed_kmh):
    """Estimate travel time in hours.

    Args:
        distance_km: Distance in kilometers.
        speed_kmh: Average speed in kilometers per hour; must be positive.

    Returns:
        Estimated travel time in hours.
    """
    if speed_kmh <= 0:
        raise ValueError("speed_kmh must be positive")
    return distance_km / speed_kmh

print(f"{estimate_travel_time(150, 60):.2f} hours")
````

## 30. Public API documentation

A public API is the set of functions, classes, and methods other code is expected to use. Its documentation should make normal usage, valid inputs, return values, errors, and side effects easy to discover.

Private implementation details usually need less user-facing documentation, but they still need clear names and comments where the reasoning is not obvious.

## 31. A practical documentation checklist

Before sharing a function or module, ask:

- Does the name describe its purpose?
- Does the docstring state what it does?
- Are important arguments and return values clear?
- Are units, assumptions, and edge cases explained?
- Are expected exceptions documented?
- Are side effects mentioned?
- Do examples actually run?
- Do comments agree with the current code?
- Could a confusing comment be replaced by clearer code?

## 32. Worked example: improve an undocumented function

The following function works for ordinary input but gives the reader little guidance. We will improve its name, validation, and documentation.

````python
def f(a, b):
    return a / b

print(f(10, 2))
````

### Improved version

Notice that the docstring explains the operation and the exception that callers may encounter. The implementation is also explicit about the zero denominator.

````python
def divide(numerator: float, denominator: float) -> float:
    """Divide one number by another.

    Args:
        numerator: The number to divide.
        denominator: The number to divide by; must not be zero.

    Returns:
        The quotient as a float.

    Raises:
        ZeroDivisionError: If denominator is zero.
    """
    if denominator == 0:
        raise ZeroDivisionError("denominator must not be zero")
    return numerator / denominator

print(divide(10, 2))
print(divide.__doc__)
````

## 33. Common mistakes

1. **Writing comments that repeat the code.** Explain why, not merely what.
2. **Treating any triple-quoted string as a docstring.** Position matters.
3. **Assuming docstrings enforce rules.** Add validation when needed.
4. **Leaving outdated comments behind.** Review them when behavior changes.
5. **Documenting a different behavior from the implementation.** Keep both aligned.
6. **Overloading docstrings with irrelevant details.** Prioritize information users need.
7. **Assuming type hints validate input.** They generally do not at runtime.

## 36. Mini-project: document a small utility module

Imagine creating a file named `study_utils.py` for a learning repository. It should contain a module docstring and two functions:

- `percentage(part, whole)`: calculate `part / whole * 100`, rejecting a zero `whole`;
- `format_percentage(value)`: return the value formatted with one decimal place and a `%` sign.

Before coding, decide what each function accepts, returns, and may raise. Then implement and document them.

````python
"""Small utilities for calculating and displaying percentages."""

def percentage(part: float, whole: float) -> float:
    """Calculate part as a percentage of whole.

    Raises:
        ZeroDivisionError: If whole is zero.
    """
    if whole == 0:
        raise ZeroDivisionError("whole must not be zero")
    return part / whole * 100


def format_percentage(value: float) -> str:
    """Format a numeric percentage with one decimal place and a percent sign."""
    return f"{value:.1f}%"

print(percentage(25, 80))
print(format_percentage(percentage(25, 80)))
````
