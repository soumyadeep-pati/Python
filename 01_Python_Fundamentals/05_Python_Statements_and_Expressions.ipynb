## 1. What is a statement?

A **statement** is an instruction that Python can execute. Statements are the building blocks of a program.

Examples:

```python
age = 20                 # assignment statement
print("Hello")           # expression statement: a function call
if age >= 18:            # if statement
    print("Adult")
```

The first statement binds the name `age` to the value `20`. The second calls `print`. The `if` statement chooses which indented block to execute.

A Python program is generally a sequence of statements. Python executes them in order, except when a statement changes the flow—for example, an `if` statement selects a branch or a loop repeats a block.

## 2. What is an expression?

An **expression** is a piece of code that Python evaluates to produce a value.

Examples:

- `2 + 3` produces `5`.
- `price * quantity` produces a calculated amount.
- `name.upper()` produces an uppercase string.
- `age >= 18` produces either `True` or `False`.

Expressions can contain smaller expressions. For example, in `2 + 3 * 4`, `3 * 4` is a smaller expression inside the larger expression.

````python
2 + 3
````

````python
price = 12.5
quantity = 4

price * quantity
````

In a notebook, the value of the final expression in a cell is usually displayed automatically. In a normal Python script, writing `2 + 3` by itself does not normally print `5`; use `print(2 + 3)` if you want the result displayed.

This is an important distinction between **evaluating a value** and **displaying a value**.

````python
# In a script, print() explicitly displays the result.
print(2 + 3)

# A variable can store the result of an expression.
total = price * quantity
print(total)
````

## 3. Statement versus expression

Compare the two:

```python
7 * 6
```

This is an expression. Python evaluates it and gets `42`.

```python
answer = 7 * 6
```

This is an assignment statement. Python evaluates the expression `7 * 6`, then binds the name `answer` to the resulting value.

| Code | Category | What happens? |
|---|---|---|
| `7 * 6` | Expression | Produces `42` |
| `answer = 7 * 6` | Assignment statement containing an expression | Computes `42` and assigns it to `answer` |
| `print(answer)` | Expression statement | Calls `print` and displays the value |
| `answer > 40` | Expression | Produces `True` or `False` |
| `if answer > 40:` | Compound statement | Tests a condition and selects a block |

A statement can contain one or more expressions. So “statement” and “expression” are related, but they are not interchangeable terms.

## 4. How Python evaluates an expression

For a simple arithmetic expression, Python follows operator precedence and associativity rules.

For example:

```python
result = 2 + 3 * 4
```

Python performs multiplication before addition:

1. `3 * 4` becomes `12`.
2. `2 + 12` becomes `14`.
3. The assignment statement binds `result` to `14`.

Parentheses can make the intended order explicit.

````python
result_a = 2 + 3 * 4
result_b = (2 + 3) * 4

print(result_a)  # 14
print(result_b)  # 20
````

### A few useful precedence rules

For the operators used most often at this stage:

1. Parentheses: `( ... )`
2. Exponentiation: `**`
3. Unary operators: `+x`, `-x`, `not x`
4. Multiplication and division: `*`, `/`, `//`, `%`
5. Addition and subtraction: `+`, `-`
6. Comparisons: `<`, `<=`, `>`, `>=`, `==`, `!=`, `in`, `is`
7. Boolean `not`
8. Boolean `and`
9. Boolean `or`

This is a simplified guide, not the complete Python precedence table. When an expression is hard to read, add parentheses rather than relying on memory.

````python
print(2 + 3 * 4)       # multiplication first: 14
print((2 + 3) * 4)     # parentheses first: 20
print(10 > 3 and 2 < 5)  # both comparisons are True, so result is True
````

## 5. Assignment statements

The assignment operator `=` assigns a value to a name. It does **not** mean “is equal to” in the mathematical comparison sense.

```python
score = 95
```

Read this as: “Evaluate the right-hand side, then assign the resulting value to the name `score`.”

The comparison operator `==` asks whether two values are equal:

```python
score == 95
```

That expression evaluates to `True` or `False`.

````python
score = 95
print(score)

print(score == 95)  # equality comparison
print(score == 80)
````

### Assignment can use an existing value

```python
count = 3
count = count + 1
```

Python evaluates `count + 1` using the current value (`3`), gets `4`, then reassigns `count` to `4`.

The right-hand side is evaluated before the assignment takes effect.

````python
count = 3
count = count + 1
print(count)  # 4
````

### Multiple assignment and unpacking

Python supports assigning several names in one statement:

```python
x, y = 10, 20
```

The values are unpacked and assigned to the corresponding names. This can also swap two values without a temporary variable:

```python
x, y = y, x
```

The number of names and values must normally match during unpacking.

````python
x, y = 10, 20
print("Before swap:", x, y)

x, y = y, x
print("After swap:", x, y)
````

## 6. Augmented assignment statements

Augmented assignment combines an operation with reassignment.

| Operator | Roughly equivalent to |
|---|---|
| `x += 2` | `x = x + 2` |
| `x -= 2` | `x = x - 2` |
| `x *= 2` | `x = x * 2` |
| `x /= 2` | `x = x / 2` |
| `x //= 2` | `x = x // 2` |
| `x %= 2` | `x = x % 2` |
| `x **= 2` | `x = x ** 2` |

These are convenient for counters, totals, and repeated updates. For mutable objects such as lists, augmented assignment can behave differently from a simple reassignment in some cases; we will revisit object behavior in a later topic.

````python
points = 10
points += 5
print(points)  # 15

points -= 3
print(points)  # 12

points *= 2
print(points)  # 24
````

## 7. Expression statements

An **expression statement** is an expression used as a statement. Function calls are a common example:

```python
print("Hello")
numbers.append(4)
```

The expression is evaluated for its effect. A function call may display text, change an object, write to a file, or perform another action.

A bare arithmetic expression such as `10 + 5` is legal, but in a script its result is discarded unless it is used, assigned, returned, or displayed.

````python
# This computes a value, but the script does not automatically display it.
10 + 5

# These make the result useful or visible.
saved_value = 10 + 5
print(saved_value)
````

## 8. Comparison expressions

Comparison operators produce Boolean values: `True` or `False`.

| Operator | Meaning | Example |
|---|---|---|
| `==` | equal to | `5 == 5` → `True` |
| `!=` | not equal to | `5 != 3` → `True` |
| `>` | greater than | `8 > 4` → `True` |
| `<` | less than | `2 < 1` → `False` |
| `>=` | greater than or equal to | `5 >= 5` → `True` |
| `<=` | less than or equal to | `3 <= 5` → `True` |

Use `==` to compare values. A single `=` is used for assignment.

````python
temperature = 28

print(temperature > 25)
print(temperature == 28)
print(temperature != 20)
print(temperature <= 30)
````

### Chained comparisons

Python lets you write a range test naturally:

```python
0 <= score <= 100
```

This means `score` is at least `0` and at most `100`. It is similar to:

```python
0 <= score and score <= 100
```

It does not mean that Python compares the two ends to each other; the middle value is checked against both bounds.

````python
score = 87
print(0 <= score <= 100)  # True

score = 120
print(0 <= score <= 100)  # False
````

## 9. Logical expressions: `and`, `or`, and `not`

Logical operators combine or reverse truth values.

- `A and B` is truthy only when both operands are truthy.
- `A or B` is truthy when at least one operand is truthy.
- `not A` reverses the truth value of `A`.

For example, a person can enter if they are at least 18 **and** have a ticket:

```python
age >= 18 and has_ticket
```

Python's `and` and `or` use **short-circuit evaluation**: the second operand may not be evaluated if the first operand already determines the result. Also, `and` and `or` can return one of their operands rather than always returning `True` or `False`. In conditions, their truth-value behavior is usually what matters.

````python
age = 21
has_ticket = True

can_enter = age >= 18 and has_ticket
print(can_enter)

is_weekend = False
is_holiday = True
day_off = is_weekend or is_holiday
print(day_off)

print(not has_ticket)
````

## 10. Membership and identity expressions

Two other expression types are useful to recognize:

- `in` checks membership, such as whether an item is in a list or substring is in a string.
- `is` checks object identity—whether two references point to the same object.

Use `==` for ordinary value equality. Do not use `is` as a general replacement for `==`. A common appropriate use is checking a value against `None` with `is None`.

````python
languages = ["Python", "Java", "C++"]

print("Python" in languages)
print("Ruby" not in languages)

result = None
print(result is None)
print(result == None)  # Works, but `is None` is the recommended style.
````

## 11. Truthiness: values in conditions

Python conditions do not require an expression to literally be `True` or `False`. Python checks whether the value is **truthy** or **falsy**.

Common falsy values include:

- `False`
- `None`
- Numeric zero: `0`, `0.0`
- Empty strings: `""`
- Empty containers: `[]`, `()`, `{}`, `set()`

Most other values are truthy. This is useful, but be precise: an empty list and a list containing zero are not the same thing.

````python
values = [0, 1, "", "Python", [], [0], None]

for value in values:
    print(repr(value), "->", bool(value))
````

## 12. Simple statements and compound statements

A **simple statement** fits on one logical line in ordinary code. Examples include assignments, `pass`, `break`, `continue`, `return`, `import`, and expression statements.

A **compound statement** contains a header and a suite (a block of statements). Examples include `if`, `for`, `while`, function definitions (`def`), and class definitions (`class`).

Compound statements commonly use a colon `:` after the header and indentation to mark the block.

````python
# Simple statements
message = "Python"
print(message)

# Compound statement
if len(message) > 0:
    print("The message is not empty.")
````

## 13. The `if` statement

An `if` statement runs a block only when its condition is truthy.

```python
if condition:
    # statements in this indented block
    ...
```

Use `elif` to test another condition when earlier conditions were false, and `else` for the fallback branch. Only one branch in an `if`/`elif`/`else` chain is selected.

````python
marks = 76

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "Needs improvement"

print("Grade:", grade)
````

### Trace the decision

For `marks = 76`:

1. Is `76 >= 90`? No.
2. Is `76 >= 75`? Yes.
3. Python assigns `"B"` to `grade`.
4. The remaining branches are skipped.

Order matters. If a broad condition appears before a more specific one, the broad branch may catch values that you intended for a later branch.

## 14. The `for` statement

A `for` statement takes items from an iterable one at a time and executes a block for each item.

```python
for item in iterable:
    # use item
```

A common use is iterating over a list. `range(start, stop)` generates integers starting at `start` and stopping **before** `stop`.

````python
for language in ["Python", "Java", "C++"]:
    print(language)

print("Counting:")
for number in range(1, 5):
    print(number)
````

## 15. The `while` statement

A `while` statement repeats its block as long as its condition is truthy.

```python
while condition:
    # statements that should eventually make the condition false
```

Be careful: if the condition never becomes false and the loop is not exited another way, the loop can run indefinitely. Ensure that the loop makes progress toward stopping.

````python
count = 1

while count <= 3:
    print("count =", count)
    count += 1
````

## 16. `break`, `continue`, and `pass`

These statements have different purposes:

- `break` exits the nearest enclosing loop immediately.
- `continue` skips the rest of the current loop iteration and proceeds to the next iteration.
- `pass` does nothing; it is a placeholder where Python syntax requires a statement.

They are not interchangeable.

````python
# break: stop when the target is found
for number in range(1, 8):
    if number == 4:
        break
    print(number)
````

````python
# continue: skip the current iteration
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
````

````python
# pass: a valid placeholder for code you will write later
for _ in range(2):
    pass

print("The loop completed.")
````

## 17. `del`, `assert`, and `import`

These statements serve different purposes:

- `del name` removes a name binding. It does not necessarily destroy the underlying object if other references still exist.
- `assert condition` raises `AssertionError` when the condition is false. It is useful for debugging and checking assumptions, not for validating untrusted user input in production.
- `import module` makes a module's names available through the imported module name.

Examples:

````python
temporary_value = 42
del temporary_value

# Uncommenting the next line would raise NameError:
# print(temporary_value)

assert 2 + 2 == 4

import math
print(math.sqrt(25))
````

## 18. A note about `return`

`return` is a statement used inside a function. It ends that function call and optionally sends a value back to the caller. It cannot be used as a normal top-level statement.

We will study function definitions in detail later; for now, recognize that `return` controls how a function gives a result back.

````python
def calculate_area(width, height):
    area = width * height
    return area

rectangle_area = calculate_area(5, 3)
print(rectangle_area)
````

## 19. Common mistakes

### Mistake 1: using `=` in a condition

Incorrect:

```python
# if score = 10:
#     print("Ten")
```

`=` assigns a value; it is not a comparison operator. Use `==` when asking whether values are equal.

### Mistake 2: forgetting the colon

Compound statement headers such as `if`, `for`, and `while` need a colon.

### Mistake 3: incorrect indentation

The body of a compound statement must be indented consistently. Indentation is part of Python's syntax, not merely visual formatting.

### Mistake 4: confusing a result with displayed output

`3 * 4` evaluates to `12`, but a regular script will not display that value unless you use it in a way that displays or preserves it, such as `print(3 * 4)` or `answer = 3 * 4`.

### Mistake 5: accidental infinite loops

A `while` loop must have a reliable way to stop, unless it is intentionally designed to run continuously.

## 20. Worked example: classify a number

**Problem:** Given an integer, report whether it is positive, negative, or zero, and whether it is even or odd.

**Plan:**

1. Store the integer in `number`.
2. Use an `if`/`elif`/`else` statement to classify its sign.
3. Use the remainder operator `%` to test evenness.
4. Display the results.

The expression `number % 2 == 0` is `True` when the number is divisible by two with no remainder.

````python
number = -8

if number > 0:
    sign = "positive"
elif number < 0:
    sign = "negative"
else:
    sign = "zero"

if number % 2 == 0:
    parity = "even"
else:
    parity = "odd"

print("Number:", number)
print("Sign:", sign)
print("Parity:", parity)
````

## 21. Worked example: total and average

This example combines expressions, assignment statements, a loop, and output statements.

**Task:** Calculate the total and average of a list of marks.

````python
marks = [78, 85, 92, 69]

total = 0
for mark in marks:
    total += mark

average = total / len(marks)

print("Marks:", marks)
print("Total:", total)
print("Average:", average)
````

### Trace the loop

| Iteration | `mark` | `total` after `total += mark` |
|---:|---:|---:|
| Start | — | 0 |
| 1 | 78 | 78 |
| 2 | 85 | 163 |
| 3 | 92 | 255 |
| 4 | 69 | 324 |

Then `average = total / len(marks)` calculates `324 / 4`, which is `81.0`.

Notice how a compound `for` statement controls repetition, while the expression `total + mark` computes each updated total.

## 26. Summary — key ideas

- An **expression** is evaluated to produce a value.
- A **statement** is an instruction Python executes.
- Statements often contain expressions; assignment evaluates its right-hand side before binding the result to a name.
- `=` assigns; `==` compares values for equality.
- Operator precedence determines evaluation order; parentheses improve clarity.
- Augmented assignment (`+=`, `-=`, and others) updates a variable concisely.
- Comparisons produce Boolean values; `and`, `or`, and `not` combine or reverse truth values.
- `if` chooses a branch; `for` iterates over items; `while` repeats while a condition is truthy.
- `break` exits a loop, `continue` skips to the next iteration, and `pass` does nothing.
- Indentation and colons are part of Python syntax.
- In scripts, use `print()` when you want to display a value explicitly.

### Quick reference

| Need | Common construct |
|---|---|
| Store a value | `x = expression` |
| Update a value | `x += amount` |
| Compare values | `x == y`, `x >= y` |
| Combine conditions | `condition_a and condition_b` |
| Choose a branch | `if` / `elif` / `else` |
| Repeat over items | `for item in iterable` |
| Repeat while true | `while condition` |
| Exit a loop | `break` |
| Skip one iteration | `continue` |
| Add a placeholder | `pass` |
| Check a development assumption | `assert condition` |
