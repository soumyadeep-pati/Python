## 6. Indentation

**Indentation** means the spaces at the beginning of a line.

Python uses indentation to define blocks of code.

For example:

```python
if True:
    print("This belongs to the if block")

The indented statement belongs to the if block.

Indentation is therefore part of Python's syntax.

This is different from languages that commonly use braces such as:

```text
{
    ...
}

Python uses indentation instead.

````python
if True:
    print("Inside the if block")

print("Outside the if block")
````

## 7. Indentation Defines Code Blocks

A **block** is a group of statements that belong together.

For example:

```python
if condition:
    statement_1
    statement_2
    statement_3

The indented statements form the body of the if statement.

The general structure is:

```text
header:
    indented statement
    indented statement

The colon : indicates that a block is beginning.

The indentation determines which statements belong to that block.

````python
number = 10

if number > 0:
    print("Positive number")
    print("The number is greater than zero")

print("Program continues")
````

## 8. Standard Indentation

The conventional indentation level in Python is **4 spaces**.

Example:

```python
if True:
    print("Hello")

The body is indented by four spaces.

Nested blocks increase the indentation level:

````python
if True:
    if True:
        print("Nested block")
````

Conceptually:

```text
Level 0
└── if
    Level 1
    └── if
        Level 2

Using a consistent indentation level is essential for readable Python code.

````python
if True:
    print("Level 1")

    if True:
        print("Level 2")
````

## 9. Indentation Must Be Consistent

Python requires indentation to be structurally consistent.

For example, this is valid:

```python
if True:
    print("First")
    print("Second")

Both statements belong to the same block.

Inconsistent indentation can produce an IndentationError or TabError.

Example of problematic structure:

````python
if True:
    print("First")
      print("Second")
````

The indentation levels do not match correctly.

The Python interpreter uses indentation to determine the structure of the program, so indentation errors can prevent the program from being parsed successfully.

## 10. Spaces vs Tabs

Python code can technically contain tabs, but mixing tabs and spaces for indentation is a bad practice.

The recommended approach is:

**Use spaces for indentation.**

The standard convention is:

**4 spaces per indentation level.**

Modern Python editors such as VS Code can automatically insert spaces when you press `Tab`.

A consistent indentation policy prevents structural errors and makes code easier to read across different editors.

````python
if True:
    print("Use consistent indentation")
    print("Four spaces per level")
````

## 11. The Colon `:`

Many Python statements that introduce a block end with a colon.

Examples:

```python

while condition:

def function():

class MyClass:

    

The colon tells Python that a new indented block follows.

Example:

````python
if True:
    print("Block begins here")
````

````python
if True:
    print("The colon introduces the block")
````

## 12. Nested Blocks

Python allows blocks to exist inside other blocks.

Example:

```python
if condition:
    for item in items:
        print(item)

Here there are two levels of indentation.

Conceptually:

```text
if block
│
└── for block
    │
    └── print statement

Each nested block increases indentation by one level.

When a block ends, the indentation returns to the previous level.

````python
numbers = [1, 2, 3]

if numbers:
    for number in numbers:
        print(number)

print("Finished")
````

## 13. Dedentation

**Dedentation** means returning to a previous indentation level.

Example:

```python
if True:
    print("Inside")

print("Outside")

The second print() is dedented, so it is outside the if block.

Python does not require a special closing keyword such as:

```text
end
```
or closing brace:

}

Instead, the indentation level tells Python when a block has ended.

````python
if True:
    print("Inside the block")

print("Outside the block")
````

## 14. Indentation in Functions

Function definitions also use indentation.

Example:

```python

Both print() statements belong to the function body.

The function call:

```python

This demonstrates an important rule:

Indentation determines which statements belong to a function.

````python
def greet():
    print("Hello")
    print("Welcome")

greet()
````

## 15. Indentation in Loops

Loops also use indentation to define their bodies.

Example:

```python

Multiple statements can belong to the same loop:

````python
for number in range(3):
    print("Number:", number)
    print("Processing...")
````

## 19. Line Continuation

A long Python expression can sometimes be split across multiple lines.

Parentheses, brackets, and braces allow implicit line continuation.

Example:

```python
numbers = [
    1,
    2,
    3,
    4
]

Another example:

```text
result = (
    10
    + 20
    + 30
)

This approach is generally preferred over using explicit backslash continuation when possible because the structure is clearer.

````python
numbers = [
    1,
    2,
    3,
    4
]

result = (
    10
    + 20
    + 30
)

print(numbers)
print(result)
````

## 20. Explicit Line Continuation

Python also supports explicit line continuation using a backslash `\`.

Example:

```python
result = 10 + \
         20 + \
         30

However, explicit backslash continuation should generally be avoided when parentheses, brackets, or braces can provide a cleaner solution.

Preferred:

```text
result = (
    10
    + 20
    + 30
)

The second form is usually easier to maintain and less error-prone.

````python
result = (
    10
    + 20
    + 30
)

print(result)
````

## 28. Syntax Errors

A **SyntaxError** occurs when Python cannot understand the structure of the source code according to Python's syntax rules.

Example:

```python
if True
    print("Hello")

The colon is missing after True.

Correct version:

````python
if True:
    print("Hello")
````

Syntax errors are normally detected before the affected code can execute.

Learning to read Python's error messages is an important programming skill.

````python
if True:
    print("Correct syntax")
````

## 29. IndentationError

An `IndentationError` occurs when indentation does not follow Python's structural rules.

For example, a statement expected to be indented may not be indented.

Correct:

```python
if True:
    print("Hello")

Incorrect:

````python
if True:
print("Hello")
````

The body of the if statement must be indented.

Indentation errors are especially common when learning Python, so understanding blocks and indentation is essential.



## 30. TabError

A `TabError` can occur when tabs and spaces are used inconsistently for indentation.

For example, one line may use a tab while another line at the same logical indentation level uses spaces.

To avoid this:

- Use spaces for indentation.
- Use four spaces per indentation level.
- Configure your editor to insert spaces when using the Tab key.
- Avoid manually mixing tabs and spaces.

## 32. Syntax vs Semantics

Syntax and semantics are different concepts.

### Syntax

Syntax describes whether code follows the structural rules of Python.

Example:

```python
x = 10

Semantics

Semantics describes what valid code means and what it does.

For example:

````python
x = 10
y = 0

result = x / y
````

The syntax is valid, but the operation causes a ZeroDivisionError during execution.

Therefore:

```text
Syntax
   ↓
Is the program structurally valid?

Semantics
   ↓
What does the valid program mean/do?

Understanding this distinction becomes increasingly important as programming knowledge grows.
