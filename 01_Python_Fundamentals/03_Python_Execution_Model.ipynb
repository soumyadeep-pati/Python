# Python Execution Model

## Overview

Understanding how Python executes a program is an important part of learning Python properly.

When we write Python code, several stages are involved before the computer produces the final result.

A simplified view is:

**Python Source Code → Parsing → Compilation → Bytecode → Python Virtual Machine → Execution**

## 1. What Happens When Python Code Runs?

Consider a simple Python statement:


````python
print("Hello, Python!")
````

When this code is executed, Python does not simply send the text directly to the CPU.

A simplified execution process is:
```text
1. Python reads the source code.
2. The source code is tokenized and parsed.
3. Python checks the structure of the program.
4. The source code is compiled into Python bytecode.
5. The bytecode is executed by the Python Virtual Machine (PVM).
6. The required operations are performed.
7. The output is produced.

Therefore:
```
Source Code → Parsing → Compilation → Bytecode → PVM → Result

The exact internal implementation depends on the Python implementation being used. The standard implementation, CPython, follows this general model.

````python
print("Hello, Python!")
````

## 2. Source Code

Python programs are normally written as source code files with the `.py` extension.

Example:



````python
name = "Python"
print(name)
````

The file might be:
```
hello.py

The .py file contains human-readable Python source code.

Source code is therefore the starting point of the Python execution process.


## 3. Parsing

Parsing is the process of analyzing the structure of Python source code.

Python checks whether the code follows the grammatical and structural rules of the language.

For example:


````python
x = 10
print(x)
````

has a valid structure.

However:
```python
x = 10
print(
```

Parsing allows Python to understand how different parts of the program are related.
Important concepts involved in this stage include:
``` text
- Tokens
- Expressions
- Statements
- Indentation
- Syntax
- Program structure
```
If the source code contains invalid Python syntax, execution normally stops before the program can run successfully.

## 4. Compilation

After Python successfully understands the source code, CPython compiles it into an intermediate representation called **bytecode**.

Bytecode is not the same as machine code.

### Machine Code

Machine code is instructions designed for a specific processor architecture.

### Python Bytecode

Python bytecode is a set of instructions designed to be executed by the Python interpreter's execution machinery.

A simplified pipeline is:

```text
Python Source Code
        ↓
      Parser
        ↓
    Compiler
        ↓
  Python Bytecode
        ↓
 Python Virtual Machine
        ↓
     Execution

Compilation happens automatically when Python runs normal source code. Developers usually do not need to manually perform this compilation step.

Bytecode is an intermediate form of Python code.

```python
x = 10

## 7. The Python Virtual Machine (PVM)

The **Python Virtual Machine (PVM)** is the execution environment responsible for executing Python bytecode.

A simplified model is:

```text
Python Source Code
        ↓
      Parsing
        ↓
    Compilation
        ↓
     Bytecode
        ↓
       PVM
        ↓
     Execution

The PVM repeatedly processes bytecode instructions and performs the required operations.

The PVM is a conceptual term used to describe the bytecode execution machinery of Python implementations such as CPython.

It is not a physical virtual machine like a virtual computer running an entire operating system.



## 8. Is Python an Interpreted Language or a Compiled Language?

Python is commonly described as an **interpreted language**.

However, saying that Python is "only interpreted" is an oversimplification.

In CPython, Python source code is generally:

1. Parsed
2. Compiled into bytecode
3. Executed by the Python runtime

Therefore, a more accurate description is:

> CPython compiles Python source code to bytecode and then executes that bytecode using its runtime execution machinery.

The terms "compiled" and "interpreted" describe different stages of execution and are not always mutually exclusive categories.

### Important distinction

```text
Python source code
        ↓
     Bytecode
        ↓
  Python runtime

This is why Python can be both described as having a compilation step and as being interpreted in the traditional sense.

Consider:

```python
x = 10

A useful conceptual model is:
```text 
x ───────→ 10
          object

Here:
```text
- 10 is an object.
- x is a name.
- The assignment connects the name x to the object.
```
Python variables are therefore better understood as names bound to objects rather than fixed memory boxes containing values.
This distinction becomes especially important when studying:
```text
- Assignment
- Mutability

- Memory management

## 10. Object Identity

Every Python object has an identity during its lifetime.

The built-in `id()` function can be used to obtain an identity value for an object.

Example:

```python
x = 10
print(id(x))

The exact value returned by id() should not be treated as a permanent memory address.

The important concept is that object identity allows Python to distinguish one object from another during execution.

````python
x = 100
y = x

print(x)
print(y)

print(id(x))
print(id(y))

print(x is y)
````

## 11. Assignment During Execution

Consider:

```python
x = 10
y = x

Conceptually:

        ┌─────────┐
x ─────→│   10    │
        └─────────┘
             ↑
y ───────────┘

Both names can refer to the same object.

The statement:

y = x

does not necessarily create a new integer object containing the same value.

Instead, it binds the name y to the object currently referenced by x.

This model becomes particularly important when working with mutable objects such as lists and dictionaries.

````python
numbers = [1, 2, 3]

other = numbers

other.append(4)

print(numbers)
print(other)

print(numbers is other)
````

- `bool`

### Mutable Objects

A mutable object can be modified after creation.

Common examples include:

Example:

The list object is modified.

Understanding mutability is important because names refer to objects, and multiple names can refer to the same object.


## 13. Execution Frames

When Python executes a function, it creates an execution frame containing information needed for that function call.

A frame can conceptually contain information such as:

- Local variables
- References to global variables
- Function arguments
- The current execution position
- Other execution state

Example:

```python
def greet(name):
    message = "Hello " + name
    return message

greet("Python")

When greet() is called, Python creates an execution context for that function call.

When the function finishes, its execution frame is no longer active.

Frames are an important part of understanding:
```text
- Function calls
- Local variables
- Call stacks
- Recursion
- Exceptions

````python
def greet(name):
    message = "Hello " + name
    return message

result = greet("Python")

print(result)
````

## 14. The Call Stack

Function calls are managed using a call stack.

Suppose:

```python
def first():
    second()

def second():
    third()

def third():
    print("Inside third")

first()

Conceptually, execution develops like this:
first()
  ↓
second()
  ↓
third()

The most recently called function is currently executing.

When third() finishes, execution returns to second().

Then second() finishes and execution returns to first().

This stack-based execution model is especially important for understanding recursion and runtime errors such as:
```
RecursionError

````python
def first():
    second()

def second():
    third()

def third():
    print("Inside third")

first()
````

Another file can import it:

When a module is imported, Python performs work to locate, load, and initialize the module.

Python also maintains information about imported modules so that repeated imports can reuse the already loaded module rather than unnecessarily executing its top-level code again.

## 16. The `__name__` Variable

Every Python module has a special variable called:

```python
__name__

Its value depends on how the module is being used.

When a file is executed directly, its __name__ is normally:

"__main__"

When the file is imported as a module, __name__ normally contains the module's name.

This leads to a common Python pattern:

````python
if __name__ == "__main__":
    ...
````

This allows code to run when the file is executed directly while preventing that section from running merely because the file was imported.

````python
print(__name__)
````

## 17. The `if __name__ == "__main__"` Pattern

A common Python program structure is:

```python
def main():
    print("Program started")

if __name__ == "__main__":
    main()

This pattern separates program logic from the code used to start the program.

It is useful because the same file can be:
```text
- Executed directly
- Imported by another Python program
```
When executed directly:

```python
__name__ == "__main__"

When imported:
```python
__name__ != "__main__"
```

This is one of the most important execution concepts for Python modules.

````python
def main():
    print("Program started")

if __name__ == "__main__":
    main()
````

## 18. `.pyc` Files and Bytecode Cache

CPython can store compiled bytecode in files ending with:

.pyc

These files are commonly stored inside a directory named:
```text
__pycache__

For example:
```text
project/

    └── ...

The cached bytecode can help Python avoid recompiling unchanged modules in certain situations.

Important points:
```text

## 21. Syntax Errors vs Runtime Errors

Execution errors can occur at different stages.

### Syntax Error

A syntax error occurs when Python cannot parse the source code correctly.

Example:


````python
if True
    print("Hello")
````

The syntax is invalid.

#### Runtime Error

A runtime error occurs while valid Python code is executing.

````python
x = 10
y = 0

print(x / y)
````

The syntax is valid, but division by zero causes an exception during execution.

Therefore:

```text
Syntax Error
    ↓
Problem detected while parsing/compiling

Runtime Error
    ↓
Problem occurs during execution

````python
x = 10
y = 0

try:
    print(x / y)
except ZeroDivisionError as error:
    print(type(error).__name__)
````
