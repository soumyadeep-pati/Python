## 1. What does “type” mean?

A **type** describes what kind of object something is and what operations are meaningful for it.

For example:

- `42` is an integer (`int`).
- `3.14` is a floating-point number (`float`).
- `"hello"` is a string (`str`).
- `True` is a Boolean value (`bool`).

The type helps Python determine how an object behaves when an operation is applied to it.

```python
print(type(42))
print(type(3.14))
print(type("hello"))
print(type(True))
```

**Expected output**

```text
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

The output identifies the type associated with each object.

## 2. What is dynamic typing?

Python is a **dynamically typed** language. In simple terms, the type is associated with an object at runtime, and a name can be assigned to objects of different types at different points in a program.

Look at this example:

```python
item = 10
print(item)
print(type(item))

item = "ten"
print(item)
print(type(item))
```

**Expected output**

```text
10
<class 'int'>
ten
<class 'str'>
```

First, `item` refers to the integer object `10`. Later, the same name `item` refers to the string object `"ten"`.

Python does not require you to declare the name's type before assigning a value.

**Important:** It is more precise to say that *objects have types and names refer to objects*. It is not that a variable permanently owns a type.

## 3. A useful mental model: names and objects

Imagine a name as a label that refers to an object.

Initially:

```python
score = 95
```

Conceptually:

```text
score ─────► integer object 95
```

Then:

```python
score = "excellent"
```

Now the name refers to a string object:

```text
score ─────► string object "excellent"
```

The name `score` has not become a special “string variable.” It now refers to an object whose type is `str`.

This is a simplified mental model, but it is useful for understanding assignment in Python.

````python
score = 95
print("Value:", score)
print("Type:", type(score))

score = "excellent"
print("Value:", score)
print("Type:", type(score))
````

**Expected output**

```text
Value: 95
Type: <class 'int'>
Value: excellent
Type: <class 'str'>
```

## 4. Static typing and dynamic typing

In a **statically typed** language, type information is checked according to that language's rules, often before a program runs. In a **dynamically typed** language such as Python, object types are handled at runtime.

This is a high-level comparison; languages differ in their exact rules, and some support both static and dynamic type-checking techniques.

| Feature | Static typing (general idea) | Python's dynamic typing |
|---|---|---|
| When types are checked | Often before execution | During execution when operations are performed |
| Must every name be declared with a type? | Depends on the language | No |
| Can a name later refer to an object of another type? | Depends on the language and its type rules | Yes |
| Can type mistakes still happen? | Yes | Yes |

For example, Python allows:

```python
value = 25
value = "twenty-five"
```

Both assignments are valid. However, dynamic typing does **not** mean every operation is allowed on every object.

## 5. Dynamic typing does not remove type rules

Python still checks whether an operation is supported by the objects involved.

For example:

```python
number = 10
message = "apples"

print(number + 5)
print(message + " are ready")
```

**Expected output**

```text
15
apples are ready
```

But this expression is not supported:

```python
number = 10
message = "apples"
print(number + message)
```

It raises a `TypeError` because Python does not define ordinary addition between an integer and a string.

The error occurs at runtime. Dynamic typing means types need not be declared in advance; it does not mean Python ignores types.

## 6. Using `type()` to inspect an object

The built-in function `type()` reports the type of an object.

```python
age = 21
price = 49.99
language = "Python"
is_learning = True

print(type(age))
print(type(price))
print(type(language))
print(type(is_learning))
```

**Expected output**

```text
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

You can also store the result and compare it, but in everyday code, `isinstance()` is often preferred when checking whether an object belongs to a type or class (including relevant subclasses).

For now, remember:

- `type(value)` helps inspect the exact type of an object.
- The displayed `<class 'int'>` means the object is an instance of the built-in `int` class.

## 7. Assignment and reassignment

An assignment associates a name with an object. Reassignment makes that name refer to another object.

```python
value = 100
print(value)

value = 200
print(value)
```

**Expected output**

```text
100
200
```

The second assignment does not mean Python has changed the integer object `100` into `200`. The name `value` is now associated with the object representing `200`.

This distinction becomes especially important when learning about mutable and immutable objects later.

## 8. Type changes versus object changes

Consider the following sequence:

```python
result = 7
result = "seven"
result = 7.0
```

The name `result` refers to objects of three different types in succession: `int`, `str`, and `float`.

```python
result = 7
print(type(result))

result = "seven"
print(type(result))

result = 7.0
print(type(result))
```

**Expected output**

```text
<class 'int'>
<class 'str'>
<class 'float'>
```

Notice that `7` and `7.0` may represent numerically equal values, but they have different types: `int` and `float`.

## 9. Objects and identity

Each object has an identity that distinguishes it from other objects during its lifetime. Python provides `id()` to return an integer representing that identity. The exact value is implementation-dependent; do not expect a particular number.

```python
first = object()
second = object()

print(first is second)
print(id(first) == id(second))
```

**Expected output**

```text
False
False
```

Here, `object()` creates two distinct objects, so `first` and `second` refer to different objects.

### What does `is` mean?

- `is` checks whether two references point to the **same object**.
- `==` checks whether two objects are **equal in value**, according to their equality behavior.

These are different questions.

````python
first_list = [1, 2, 3]
second_list = [1, 2, 3]
third_list = first_list

print(first_list == second_list)
print(first_list is second_list)
print(first_list is third_list)
````

**Expected output**

```text
True
False
True
```

Both lists `first_list` and `second_list` contain equal values, but they are separate list objects. `third_list` refers to the exact same object as `first_list`.

The list example previews mutability and data structures; those topics will be studied properly later.

## 10. A caution about `id()` and memory

`id()` is useful for illustrating identity, but it is not a portable memory address.

- Its numeric value depends on the Python implementation.
- An object's identity is guaranteed to remain stable during that object's lifetime.
- Once an object is no longer alive, its identity value may be reused for another object.

Therefore, do not use `id()` as a permanent identifier or assume that a particular number means a particular type or value.

## 11. Dynamic typing and type annotations

Python also supports **type annotations**, which can communicate the expected types to readers and tools.

```python
age: int = 21
name: str = "Asha"
```

These annotations do not, by themselves, make Python enforce those types at runtime. A type checker or other tool can use annotations to find possible mistakes before execution.

For example, Python itself permits reassignment such as:

```python
age: int = 21
age = "twenty-one"
print(age)
```

The code runs and prints `twenty-one`, although a static type checker would normally flag the assignment as inconsistent with the annotation.

**Key distinction:** Python is dynamically typed even though it supports optional type annotations and external static type-checking tools.

## 12. Common misconceptions

1. **“A variable has a permanent type.”**  
   More precisely, objects have types; a name can refer to objects of different types over time.

2. **“Dynamic typing means no type checking.”**  
   Python checks whether operations are valid at runtime.

3. **“Python requires every variable to be declared.”**  
   Ordinary names can be assigned without prior type declarations.

4. **“`==` and `is` mean the same thing.”**  
   `==` checks equality; `is` checks identity.

5. **“`id()` always gives the object's memory address.”**  
   It returns an identity value whose exact meaning depends on the implementation.

6. **“Type annotations force the type at runtime.”**  
   Annotations alone do not enforce this.
