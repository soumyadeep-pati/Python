# 13 — Parameters and Hyperparameters

## Introduction

A Machine Learning model contains values that determine how it learns from data and how it makes predictions.

Some values are **learned from the training data**.

Other values are **set before the training process begins**.

These two types of values are called:

- **Parameters**
- **Hyperparameters**

Understanding the difference between parameters and hyperparameters is an important part of understanding how Machine Learning models are trained.

A simplified training process can be represented as:

$$
\text{Training Data}
\rightarrow
\text{Training Process}
\rightarrow
\text{Trained Model}
$$

During this process, the model learns its parameters while the hyperparameters control how the training process is performed.

## 1. Parameters

A **parameter** is a value that a Machine Learning model learns from the training data during the training process.

Parameters are part of the model's learned state.

The training algorithm adjusts these values so that the model can learn useful patterns from the training data.

Therefore:

$$
\boxed{
\text{Parameters} = \text{Values learned from training data}
}
$$

For example, consider a simple linear model:

$$
\hat{y} = wx + b
$$

where:

- $x$ is the input
- $\hat{y}$ is the predicted output
- $w$ is the weight
- $b$ is the bias

The values $w$ and $b$ are parameters of the model.

Their values are learned during training.

```python
# Example of model parameters

w = 2.5
b = 1.0

print("Weight:", w)
print("Bias:", b)
```

## 2. Parameters in a Linear Model

Consider the linear model:

$$
\hat{y} = wx + b
$$

Suppose the model has learned:

$$
w = 2
$$

and:

$$
b = 3
$$

For an input:

$$
x = 5
$$

the prediction is:

$$
\hat{y} = 2(5) + 3
$$

$$
\hat{y} = 13
$$

Here:

- $w = 2$ is a learned parameter.
- $b = 3$ is a learned parameter.
- $x = 5$ is an input.
- $\hat{y} = 13$ is the prediction.

The values of $w$ and $b$ are determined during the training process.

## 4. How Parameters Are Learned

Parameters are learned through the training process.

A simplified learning process can be represented as:

$$
\text{Input Data}
\rightarrow
\text{Prediction}
\rightarrow
\text{Loss}
\rightarrow
\text{Parameter Update}
$$

The model first makes predictions using its current parameter values.

The predictions are then compared with the actual target values.

A loss function measures how well the model's predictions match the actual values.

The training algorithm uses this information to update the parameters.

A simplified parameter update can be represented as:

$$
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta
\frac{\partial J}{\partial \theta}
$$

where:

- $\theta$ represents a model parameter
- $J$ represents the loss function
- $\eta$ represents the learning rate

The training process continues until an appropriate set of parameter values is obtained according to the chosen learning algorithm.

## 5. Hyperparameters

A **hyperparameter** is a value that is configured before training or controlled outside the process that learns the model parameters.

Unlike parameters, hyperparameters are generally **not learned directly from the training data as part of the model's parameter-learning process**.

Hyperparameters control aspects of the model or the training process.

Therefore:

$$
\boxed{
\text{Hyperparameters}
=
\text{Values configured for the learning process}
}
$$

For example, a learning algorithm may use a learning rate to control the size of parameter updates.

The learning rate is selected as part of the training configuration.

The model parameters are then learned using that configuration.

## 7. Parameters vs Hyperparameters

Parameters and hyperparameters both affect a Machine Learning model, but they have different roles.

| Property | Parameters | Hyperparameters |
|---|---|---|
| Learned from training data | Yes | No, not directly |
| Configured before training | Usually no | Usually yes |
| Updated during training | Yes | Usually no |
| Role | Represent learned model information | Control the model or training process |
| Examples | Weights, biases | Learning rate, tree depth, number of neighbors |

The fundamental difference is:

$$
\boxed{
\text{Parameters} \rightarrow \text{Learned}
}
$$

while:

$$
\boxed{
\text{Hyperparameters} \rightarrow \text{Configured}
}
$$

## 8. Parameters and Hyperparameters Together

Parameters and hyperparameters work together during Machine Learning training.

The relationship can be represented as:

$$
\text{Training Data}
+
\text{Hyperparameters}
\rightarrow
\text{Training Algorithm}
\rightarrow
\text{Learned Parameters}
$$

The hyperparameters determine how the training process is configured.

The training algorithm then uses the training data to learn the model parameters.

The resulting parameters become part of the trained model.

Therefore:

$$
\boxed{
\text{Hyperparameters control the learning process}
}
$$

and:

$$
\boxed{
\text{Parameters are learned by the learning process}
}
$$

## 9. Why Hyperparameters Matter

Hyperparameters can strongly influence the behavior and performance of a Machine Learning model.

For example, consider the learning rate.

The learning rate controls the size of parameter updates during optimization.

If the learning rate is too large, parameter updates may be too large.

If the learning rate is too small, training may become very slow.

Conceptually:

$$
\text{Learning Rate}
\rightarrow
\text{Parameter Updates}
\rightarrow
\text{Training Behavior}
$$

Similarly, the maximum depth of a decision tree influences how complex the learned tree can become.

Therefore, choosing appropriate hyperparameters is an important part of building a Machine Learning model.

## 10. Hyperparameter Tuning

**Hyperparameter tuning** is the process of searching for suitable hyperparameter values that produce good model performance.

Suppose a model has a hyperparameter $k$.

We might test:

$$
k \in \{3,5,7,9\}
$$

Each value can be evaluated using an appropriate validation procedure.

The general process is:

$$
\text{Choose Hyperparameters}
\rightarrow
\text{Train Model}
\rightarrow
\text{Evaluate Model}
\rightarrow
\text{Compare Results}
$$

This process can be repeated for different hyperparameter configurations.

The goal is to find a configuration that provides good performance on data that was not used to directly fit the model parameters.

## 11. Common Hyperparameter Tuning Methods

Several methods can be used to search for suitable hyperparameters.

### 11.1 Manual Search

The practitioner selects hyperparameter values based on experience and experimentation.

### 11.2 Grid Search

Grid Search evaluates combinations from a predefined set of hyperparameter values.

For example:

$$
k \in \{3,5,7\}
$$

and:

$$
\text{Metric} \in
\{\text{Euclidean},\text{Manhattan}\}
$$

The combinations of these choices can then be evaluated.

### 11.3 Random Search

Random Search evaluates randomly selected hyperparameter combinations from specified ranges or distributions.

### 11.4 Advanced Methods

More advanced approaches include:

- Bayesian optimization
- Successive halving
- Hyperband
- Other automated optimization methods

The appropriate method depends on the model, search space, computational resources, and problem.

## 12. Example of a Hyperparameter

Consider the following model configuration:

`KNeighborsClassifier(n_neighbors=3)`

Here:

$$
k = 3
$$

is a hyperparameter.

It is selected when configuring the model.

The value is not learned from the training data as a model parameter.

Instead, it determines how the K-Nearest Neighbors algorithm will make predictions.

Different values of $k$ can produce different model behaviors.

For example:

$$
k = 1
$$

and:

$$
k = 5
$$

represent different model configurations.

Therefore:

$$
\boxed{
k \text{ is a hyperparameter}
}
$$

```python
from sklearn.neighbors import KNeighborsClassifier

# n_neighbors is a hyperparameter
model = KNeighborsClassifier(n_neighbors=3)

print(model)
```

## 13. Understanding the Example

In the previous example:

`n_neighbors=3`

sets the number of neighbors used by the K-Nearest Neighbors classifier.

Therefore:

$$
\boxed{
n\_neighbors = 3
\rightarrow
\text{Hyperparameter}
}
$$

The value is specified when the model is created.

It is not learned by fitting the model.

The training data is then used according to this configuration when the model is trained.

This demonstrates the important distinction between configuring a model and learning from data.

## 14. Parameters and Hyperparameters in the Complete Workflow

The relationship between parameters and hyperparameters can be placed into the complete Machine Learning workflow.

First, the training data is provided.

Then, the hyperparameters are configured.

The training algorithm uses the training data and the selected configuration to learn the model parameters.

The learned parameters form part of the trained model.

The trained model can then be used for inference.

The complete process can be represented as:

$$
\boxed{
\text{Training Data}
\rightarrow
\text{Hyperparameter Configuration}
\rightarrow
\text{Training}
\rightarrow
\text{Learned Parameters}
\rightarrow
\text{Trained Model}
}
$$

After training:

$$
\boxed{
\text{New Data}
\rightarrow
\text{Trained Model}
\rightarrow
\text{Prediction}
}
$$

## 15. Important Distinction

A common mistake is to treat every value associated with a Machine Learning model as a parameter.

This is not correct.

A **parameter** is a value learned as part of the model-fitting process.

A **hyperparameter** is a value that controls the model or training process and is not directly learned as a model parameter.

Therefore:

$$
\boxed{
\text{Parameter} \neq \text{Hyperparameter}
}
$$

Both can influence the final model, but they have different roles.

The key distinction is:

$$
\boxed{
\text{Parameter} = \text{Learned from data}
}
$$

$$
\boxed{
\text{Hyperparameter} = \text{Configured outside parameter learning}
}
$$
