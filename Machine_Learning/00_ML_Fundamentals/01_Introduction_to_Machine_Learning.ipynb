# 01 — Introduction to Machine Learning

## 1. Introduction

Machine Learning (ML) is a branch of Artificial Intelligence (AI) that focuses on building systems that can learn patterns from data and use those patterns to make predictions, decisions, or other useful outputs.

In traditional programming, a programmer explicitly defines the rules that a computer should follow.

In Machine Learning, instead of manually defining all the rules, we provide data and a learning algorithm. The algorithm uses the data to construct a model that can capture useful patterns.

The basic idea can be represented as:

$$
\text{Data} + \text{Learning Algorithm}
\longrightarrow
\text{Machine Learning Model}
$$

The trained model can then be used on new data:

$$
\text{New Data} + \text{Trained Model}
\longrightarrow
\text{Prediction or Decision}
$$

Machine Learning is useful when explicitly writing rules for every possible situation would be difficult, impractical, or impossible.

## 2. What Is Machine Learning?

Machine Learning is the process of developing computational models that learn useful patterns from data and use those learned patterns to perform a task.

A Machine Learning system generally involves:

1. **Data** — information used for learning.
2. **Features** — measurable properties of the data.
3. **Target** — the value or outcome that we may want to predict.
4. **Learning Algorithm** — the method used to learn patterns from data.
5. **Model** — the result produced by the learning process.
6. **Prediction** — the output produced by the trained model for new data.

A simplified Machine Learning workflow is:

$$
\text{Data}
\rightarrow
\text{Learning}
\rightarrow
\text{Model}
\rightarrow
\text{Prediction}
$$

## 3. Traditional Programming vs Machine Learning

### 3.1 Traditional Programming

In traditional programming, the programmer provides explicit rules and data to a computer program.

The program applies those rules to the data and produces an output.

$$
\text{Data} + \text{Rules}
\longrightarrow
\text{Program}
\longrightarrow
\text{Output}
$$

For example, suppose we want to determine whether a number is even.

We can explicitly define the rule:

$$
n \bmod 2 = 0
$$

If the remainder is zero, the number is even.

The rule is directly written by the programmer.

### 3.2 Machine Learning

In Machine Learning, the rules are not necessarily written explicitly.

Instead, the system receives examples and uses a learning algorithm to construct a model.

$$
\text{Data} + \text{Expected Outputs}
\longrightarrow
\text{Learning Algorithm}
\longrightarrow
\text{Model}
$$

The model can then be applied to new data.

$$
\text{New Data} + \text{Model}
\longrightarrow
\text{Prediction}
$$

## 4. A Simple Machine Learning Example

Consider a dataset containing information about houses.

Each house may have information such as:

- Area
- Number of bedrooms
- Age
- Location
- Price

Suppose the goal is to predict the price of a house.

The information describing a house is used as input to the Machine Learning model.

For example:

$$
\text{Area} = 1500
$$

$$
\text{Bedrooms} = 3
$$

$$
\text{Age} = 10
$$

The model uses these inputs to produce a predicted price.

Conceptually:

$$
\text{House Information}
\longrightarrow
\text{Machine Learning Model}
\longrightarrow
\text{Predicted Price}
$$

The model learns from previously observed examples and uses the learned relationship to make predictions for new houses.

## 5. Important Terminology

Before studying Machine Learning algorithms, it is important to understand the basic terminology used to describe data and models.

The most important terms are:

- Sample
- Feature
- Target
- Dataset
- Model
- Training
- Prediction

These terms form the foundation for understanding the rest of Machine Learning.

## 5.1 Sample

A **sample** is a single observation or example in a dataset.

For example, consider a house-price dataset.

One row of the dataset may represent one house:

| Area | Bedrooms | Age | Price |
|---:|---:|---:|---:|
| 1500 | 3 | 10 | 250000 |

This single row represents one **sample**.

A dataset generally contains multiple samples:

$$
\text{Dataset}
=
\{\text{Sample}_1,\text{Sample}_2,\ldots,\text{Sample}_m\}
$$

where \(m\) represents the number of samples.

## 5.2 Feature

A **feature** is an individual measurable property or characteristic of a sample that is used as input to a Machine Learning model.

For the house-price example, possible features are:

- Area
- Number of bedrooms
- Age

If a sample has \(n\) features, its feature representation can be written as:

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n)
$$

For example, a house may be represented as:

$$
\mathbf{x}
=
(1500,3,10)
$$

where each value corresponds to a feature.

## 5.3 Target

The **target** is the value that a Machine Learning model is intended to predict or estimate.

In the house-price example, the target is the house price.

For a sample, the target can be represented as:

$$
y
$$

For example:

$$
y = 250000
$$

The features are the inputs, while the target is the desired output.

Therefore:

$$
\mathbf{x}
\longrightarrow
y
$$

where:

- \(\mathbf{x}\) represents the input features.
- \(y\) represents the target.

## 5.4 Dataset

A **dataset** is a collection of samples used for analysis, learning, evaluation, or other computational tasks.

Suppose a dataset contains \(m\) samples and \(n\) features.

The input data can be represented as a matrix:

$$
X \in \mathbb{R}^{m \times n}
$$

where:

- \(m\) = number of samples
- \(n\) = number of features

The dataset can be represented as:

$$
X =
\begin{bmatrix}
x_{11} & x_{12} & \cdots & x_{1n} \\
x_{21} & x_{22} & \cdots & x_{2n} \\
\vdots & \vdots & \ddots & \vdots \\
x_{m1} & x_{m2} & \cdots & x_{mn}
\end{bmatrix}
$$

If the dataset has a target value for each sample, the targets can be represented as:

$$
\mathbf{y}
=
\begin{bmatrix}
y_1\\
y_2\\
\vdots\\
y_m
\end{bmatrix}
$$

Thus, a supervised learning dataset can be represented by:

$$
(X,\mathbf{y})
$$

## 5.5 Model

A **model** is a mathematical or computational representation learned from data.

The model attempts to capture a useful relationship between inputs and outputs.

For a prediction problem, we can represent a model as a function:

$$
\hat{y} = f(\mathbf{x})
$$

where:

- \(\mathbf{x}\) is the input feature vector.
- \(f\) is the learned model.
- \(\hat{y}\) is the predicted output.

The prediction is therefore produced by applying the learned model to an input.

## 5.6 Training

**Training** is the process of using data to learn the parameters or structure of a Machine Learning model.

During training, the learning algorithm examines the available data and adjusts the model so that it can perform the required task.

Conceptually:

$$
\text{Training Data}
\longrightarrow
\text{Learning Algorithm}
\longrightarrow
\text{Trained Model}
$$

The purpose of training is to obtain a model that can make useful predictions or decisions on data it has not previously encountered.

## 5.7 Prediction

A **prediction** is the output produced by a trained Machine Learning model when it receives input data.

If the trained model is represented by \(f\), then:

$$
\hat{y} = f(\mathbf{x})
$$

Here:

- \(\mathbf{x}\) is the input.
- \(f\) is the trained model.
- \(\hat{y}\) is the predicted output.

For example, a trained house-price model can receive information about a new house and produce an estimated price.

## 6. Basic Machine Learning Workflow

A basic Machine Learning workflow consists of several stages.

$$
\text{Data}
\rightarrow
\text{Preparation}
\rightarrow
\text{Training}
\rightarrow
\text{Evaluation}
\rightarrow
\text{Prediction}
$$

The major stages are:

1. **Collect or obtain data**
2. **Understand the data**
3. **Prepare the data**
4. **Select an appropriate learning approach**
5. **Train a model**
6. **Evaluate the model**
7. **Use the trained model for prediction**

These stages form a general workflow. Different Machine Learning problems may require different methods and additional steps.

## 7. Why Machine Learning Is Used

Machine Learning is useful when a system needs to learn patterns from data.

Common reasons for using Machine Learning include:

- Making predictions
- Classifying observations
- Finding patterns in data
- Detecting unusual behavior
- Making decisions based on learned patterns
- Discovering relationships that may be difficult to specify manually

The usefulness of Machine Learning depends on the availability and quality of data, the suitability of the learning method, and the ability of the resulting model to generalize to new data.

## 8. Types of Machine Learning

Machine Learning can be broadly divided into different learning settings.

The major categories introduced here are:

1. **Supervised Learning**
2. **Unsupervised Learning**
3. **Semi-Supervised Learning**
4. **Reinforcement Learning**

These categories differ mainly in the type of information available during learning and the way the learning problem is defined.

Each category will be studied in greater detail in later sections.
