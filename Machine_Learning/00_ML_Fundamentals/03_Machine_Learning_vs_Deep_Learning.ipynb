# 03 — Machine Learning vs Deep Learning

## 1. Introduction

Machine Learning (ML) and Deep Learning (DL) are closely related fields.

Machine Learning is a subfield of Artificial Intelligence that focuses on learning patterns from data.

Deep Learning is a subfield of Machine Learning that uses neural networks with multiple layers to learn representations from data.

The relationship is:

$$
\boxed{
\text{Artificial Intelligence}
\supset
\text{Machine Learning}
\supset
\text{Deep Learning}
}
$$

This notebook introduces the relationship and major differences between Machine Learning and Deep Learning.

## 2. What Is Machine Learning?

Machine Learning is a field in which algorithms learn patterns from data and use those patterns to perform a task.

A general Machine Learning process is:

$$
\text{Data}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Model}
$$

The trained model can then process new input:

$$
\text{New Input}
\rightarrow
\text{Trained Model}
\rightarrow
\text{Prediction}
$$

Machine Learning includes many different types of algorithms, such as:

- Linear Regression
- Logistic Regression
- K-Nearest Neighbors
- Decision Trees
- Support Vector Machines
- Naive Bayes
- Random Forest
- Clustering algorithms

These algorithms can use different mathematical and computational approaches to learn from data.

## 3. What Is Deep Learning?

Deep Learning is a subfield of Machine Learning that uses artificial neural networks with multiple layers.

A neural network consists of connected computational units that transform input data through a sequence of layers.

A simplified representation is:

$$
\text{Input}
\rightarrow
\text{Hidden Layers}
\rightarrow
\text{Output}
$$

Deep neural networks contain multiple layers between the input and output.

The term **deep** refers to the presence of multiple layers in the network.

Deep Learning models can learn useful representations of data during training.

## 4. Relationship Between Machine Learning and Deep Learning

Deep Learning belongs to Machine Learning.

Therefore:

$$
\boxed{\text{Deep Learning} \subset \text{Machine Learning}}
$$

Machine Learning itself belongs to Artificial Intelligence:

$$
\boxed{
\text{Artificial Intelligence}
\supset
\text{Machine Learning}
\supset
\text{Deep Learning}
}
$$

This means that Deep Learning is not a separate field outside Machine Learning.

It is a particular approach within Machine Learning that is based primarily on multi-layer neural networks.

## 5. Machine Learning vs Deep Learning

Machine Learning and Deep Learning can be compared across several aspects.

| Aspect | Machine Learning | Deep Learning |
|---|---|---|
| Relationship | Broad subfield of AI | Subfield of ML |
| Main models | Many types of algorithms | Mainly multi-layer neural networks |
| Feature representation | Often designed or prepared manually | Can learn representations from data |
| Data requirements | Often works well with smaller datasets | Often benefits from larger datasets |
| Computation | Usually lower computational requirements | Often higher computational requirements |
| Training time | Often shorter | Often longer |
| Hardware | CPU can often be sufficient | GPUs or other accelerators are often useful |
| Model complexity | Varies | Often high |

## 6. Feature Representation

One important difference between traditional Machine Learning and Deep Learning is how features are represented.

In many Machine Learning problems, the input data is first transformed into useful features.

The general process can be represented as:

$$
\text{Raw Data}
\rightarrow
\text{Feature Engineering}
\rightarrow
\text{ML Algorithm}
\rightarrow
\text{Prediction}
$$

Feature engineering may involve selecting, transforming, or constructing useful input variables.

In Deep Learning, neural networks can learn useful representations directly from the input data.

A simplified process is:

$$
\text{Raw Data}
\rightarrow
\text{Deep Neural Network}
\rightarrow
\text{Prediction}
$$

This does not mean that Deep Learning never requires data preprocessing or feature preparation. Rather, a major advantage is that the network can learn representations as part of the training process.

## 7. Feature Engineering in Machine Learning

Feature engineering is the process of creating or transforming input features so that they are useful for a Machine Learning model.

Suppose a dataset contains information about houses.

The raw information may include:

- Area
- Number of rooms
- Construction year
- Location

Additional useful features may be constructed from the available information.

The general idea is:

$$
\text{Raw Variables}
\rightarrow
\text{Feature Engineering}
\rightarrow
\text{Useful Features}
$$

The quality of the features can have a significant effect on the performance of many Machine Learning models.

Feature engineering is therefore an important part of many traditional Machine Learning workflows.

## 8. Representation Learning in Deep Learning

Deep Learning models can learn representations from data through their layers.

A simplified view is:

$$
\text{Input}
\rightarrow
\text{Representation}_1
\rightarrow
\text{Representation}_2
\rightarrow
\cdots
\rightarrow
\text{Prediction}
$$

Different layers can learn different levels of representation.

For example, when processing an image, earlier layers may learn simple patterns while later layers can combine those patterns into more complex representations.

The exact representations learned depend on the architecture, data, training procedure, and task.

## 9. Data Requirements

Machine Learning and Deep Learning can behave differently as the amount of available data changes.

Many traditional Machine Learning algorithms can perform well with relatively small or medium-sized datasets, depending on the problem.

Deep Learning models often contain a large number of trainable parameters and can benefit substantially from larger datasets.

Conceptually:

$$
\text{More Training Data}
\rightarrow
\text{More Information Available for Learning}
$$

However, more data does not automatically guarantee a better model.

The quality of the data, the problem formulation, the model architecture, and the training process are also important.

## 10. Computational Requirements

Traditional Machine Learning models often have relatively modest computational requirements.

Many models can be trained efficiently using a CPU.

Deep Learning models can require significantly more computation because neural networks may contain many parameters and involve large numbers of numerical operations during training.

For large Deep Learning workloads, specialized hardware such as GPUs can significantly accelerate computation.

A simplified comparison is:

$$
\text{Traditional ML}
\rightarrow
\text{Often Lower Computational Cost}
$$

$$
\text{Deep Learning}
\rightarrow
\text{Often Higher Computational Cost}
$$

The actual computational requirement depends on the dataset, algorithm, model size, and task.

## 11. Model Complexity

Machine Learning includes models with different levels of complexity.

For example, a simple linear model can represent a relatively simple relationship:

$$
\hat{y} = w_1x_1 + w_2x_2 + b
$$

where:

- \(x_1,x_2\) are input features.
- \(w_1,w_2\) are model parameters.
- \(b\) is a bias term.
- \(\hat{y}\) is the predicted output.

Deep Learning models can contain many layers and a large number of trainable parameters.

A simplified neural network can be represented as:

$$
\mathbf{x}
\rightarrow
f_1
\rightarrow
f_2
\rightarrow
f_3
\rightarrow
\hat{y}
$$

where each function represents a transformation performed by a layer or group of layers.

## 12. Training Process

Both Machine Learning and Deep Learning involve training.

In general, training attempts to find model parameters that allow the model to perform the required task effectively.

For a model with parameters represented by \(\theta\), the training process can be viewed as minimizing a loss function:

$$
\theta^*
=
\arg\min_{\theta} L(\theta)
$$

where:

- \(\theta\) represents the model parameters.
- \(L(\theta)\) represents the loss.
- \(\theta^*\) represents parameters that minimize the loss according to the training procedure.

Deep Learning commonly uses optimization methods based on gradient descent and backpropagation.

These concepts will be studied in detail in the Deep Learning section.

## 13. Interpretability

Interpretability refers to how easily humans can understand why a model produces a particular output.

Some traditional Machine Learning models can be relatively easy to interpret.

For example, a small decision tree can be inspected by following its decision rules.

Deep Learning models can be much more difficult to interpret because they may contain many layers and a very large number of parameters.

Therefore, a common generalization is:

$$
\text{Model Complexity}
\uparrow
\quad\Rightarrow\quad
\text{Interpretability May Decrease}
$$

This is not an absolute rule. Interpretability depends on the specific model, task, and interpretation method.

## 14. Typical Applications

Machine Learning is widely used for tasks such as:

- Classification
- Regression
- Clustering
- Anomaly detection
- Recommendation
- Prediction

Deep Learning is particularly effective for many tasks involving complex, high-dimensional data.

Examples include:

- Image recognition
- Speech recognition
- Natural language processing
- Computer vision
- Generative models

Deep Learning is not automatically the best choice for every problem.

The appropriate approach depends on the data, task, computational resources, and required performance.

## 15. A Conceptual Comparison

The difference between a traditional Machine Learning workflow and a Deep Learning workflow can be summarized as follows.

### Machine Learning

$$
\text{Raw Data}
\rightarrow
\text{Preprocessing}
\rightarrow
\text{Feature Engineering}
\rightarrow
\text{ML Model}
\rightarrow
\text{Prediction}
$$

### Deep Learning

$$
\text{Raw Data}
\rightarrow
\text{Preprocessing}
\rightarrow
\text{Deep Neural Network}
\rightarrow
\text{Learned Representations}
\rightarrow
\text{Prediction}
$$

These are simplified workflows.

Both approaches can require preprocessing, careful model selection, training, validation, and evaluation.

## 16. Is Deep Learning Always Better Than Machine Learning?

No.

Deep Learning is not universally better than traditional Machine Learning.

For some problems, a traditional Machine Learning algorithm may be more appropriate because:

- The dataset may be relatively small.
- The input may already be well represented by meaningful features.
- Training resources may be limited.
- Interpretability may be important.
- A simpler model may already provide sufficient performance.

Deep Learning can be particularly useful when:

- Large amounts of data are available.
- The input contains complex patterns.
- Learning representations from raw or less-processed data is valuable.
- Sufficient computational resources are available.

Therefore, the choice should be based on the problem rather than assuming that the more complex method is always better.
