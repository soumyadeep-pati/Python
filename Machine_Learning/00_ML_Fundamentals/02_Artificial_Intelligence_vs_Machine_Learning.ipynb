# 02 — Artificial Intelligence vs Machine Learning

## 1. Introduction

Artificial Intelligence (AI) and Machine Learning (ML) are closely related fields, but they are not the same thing.

Artificial Intelligence is the broader field concerned with creating systems that can perform tasks associated with intelligent behavior.

Machine Learning is a subfield of Artificial Intelligence that focuses on developing systems that learn patterns from data.

A simple relationship is:


$$
\boxed{\text{Artificial Intelligence} \supset \text{Machine Learning}}
$$

Machine Learning is therefore one approach used to build Artificial Intelligence systems.

## 2. What Is Artificial Intelligence?

**Artificial Intelligence (AI)** is the field of computing concerned with creating systems that can perform tasks that require capabilities associated with intelligence.

These capabilities may include:

- Reasoning
- Learning
- Problem solving
- Planning
- Perception
- Understanding language
- Decision making

An AI system may use different techniques to perform these tasks.

The general idea can be represented as:

$$
\text{AI System}
\rightarrow
\text{Intelligent Behavior}
$$

Artificial Intelligence is a broad field, so not every AI system necessarily learns from data using Machine Learning.

## 3. What Is Machine Learning?

**Machine Learning (ML)** is a subfield of Artificial Intelligence that focuses on algorithms and models that learn patterns from data.

Instead of explicitly specifying every rule, a Machine Learning system uses data to learn a model.

The basic process is:

$$
\text{Data}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Model}
$$

The trained model can then be used on new data:

$$
\text{New Data}
\rightarrow
\text{Trained Model}
\rightarrow
\text{Prediction}
$$

Therefore, Machine Learning provides a data-driven approach to building systems that can perform useful tasks.

## 4. Relationship Between AI and ML

Artificial Intelligence is the broader concept.

Machine Learning is one of the major approaches used within Artificial Intelligence.

The relationship can be visualized as:

$$
\boxed{
\begin{array}{c}
\text{Artificial Intelligence}\\
\downarrow\\
\text{Machine Learning}\\
\downarrow\\
\text{Learning from Data}
\end{array}
}
$$

A more general representation is:

$$
\text{Artificial Intelligence}
\supset
\text{Machine Learning}
$$

This means:

- Every Machine Learning system belongs to the broader field of Artificial Intelligence.
- Not every Artificial Intelligence system has to use Machine Learning.

## 5. AI Without Machine Learning

An intelligent system can sometimes be created using explicitly defined rules rather than learning from data.

For example, consider a simple rule-based system:

- If the temperature is below a certain value, recommend a jacket.
- If the temperature is above a certain value, recommend lighter clothing.

The rules are explicitly provided by the programmer.

Conceptually:

$$
\text{Input}
+
\text{Explicit Rules}
\rightarrow
\text{Output}
$$

The system does not need to learn those rules from a dataset.

This demonstrates why Artificial Intelligence is broader than Machine Learning.

## 6. Machine Learning as an Approach to AI

Machine Learning approaches the problem differently.

Instead of explicitly defining all the rules, we provide examples and allow an algorithm to learn a model from the data.

Conceptually:

$$
\text{Examples}
+
\text{Learning Algorithm}
\rightarrow
\text{Learned Model}
$$

The learned model can then be used to process new inputs.

$$
\text{New Input}
+
\text{Learned Model}
\rightarrow
\text{Output}
$$

The important difference is that the behavior of the system is learned from data rather than being completely specified through manually written rules.

## 7. Artificial Intelligence vs Machine Learning

The main difference can be summarized as follows:

| Aspect | Artificial Intelligence | Machine Learning |
|---|---|---|
| Scope | Broad field | Subfield of AI |
| Main goal | Build systems capable of intelligent behavior | Learn patterns from data |
| Learning from data | May or may not be used | Central concept |
| Rules | May be explicitly defined | Usually learned from data |
| Output | Intelligent behavior or decisions | Predictions, classifications, patterns, or decisions |
| Relationship | Contains ML | Belongs to AI |

Therefore:

$$
\boxed{\text{AI is broader than ML}}
$$

## 8. Traditional Rule-Based Approach vs Machine Learning

Consider a system that needs to classify an email as either spam or not spam.

### Rule-Based Approach

A programmer may manually define rules such as:

- If certain words appear, increase the spam score.
- If too many links appear, increase the spam score.
- If certain patterns appear, classify the email as spam.

The system follows explicitly defined rules.

$$
\text{Email}
+
\text{Rules}
\rightarrow
\text{Classification}
$$

### Machine Learning Approach

A Machine Learning system can be trained using examples of spam and non-spam emails.

The learning algorithm uses these examples to learn patterns.

$$
\text{Labeled Emails}
+
\text{Learning Algorithm}
\rightarrow
\text{Trained Model}
$$

The trained model can then classify a new email:

$$
\text{New Email}
+
\text{Trained Model}
\rightarrow
\text{Prediction}
$$

## 9. AI, ML, and Deep Learning

Deep Learning (DL) is another important concept related to AI and ML.

Deep Learning is a subfield of Machine Learning that uses neural networks with multiple layers to learn representations from data.

The relationship can be represented as:

$$
\boxed{
\text{Artificial Intelligence}
\supset
\text{Machine Learning}
\supset
\text{Deep Learning}
}
$$

Therefore:

- Artificial Intelligence is the broadest field.
- Machine Learning is a subfield of Artificial Intelligence.
- Deep Learning is a subfield of Machine Learning.

Deep Learning will be studied separately in the Deep Learning section.

## 10. Examples of AI and ML

Different systems can involve different approaches.

### Artificial Intelligence

Examples of AI tasks include:

- Planning
- Reasoning
- Decision making
- Game playing
- Language understanding
- Perception

### Machine Learning

Examples of Machine Learning tasks include:

- Predicting house prices
- Classifying emails
- Recognizing patterns in data
- Predicting customer behavior
- Detecting anomalies

Many modern AI systems combine multiple techniques, including Machine Learning and Deep Learning.

## 11. A Conceptual Hierarchy

The relationship between the major concepts can be summarized as:

$$
\boxed{
\text{Artificial Intelligence}
\supset
\text{Machine Learning}
\supset
\text{Deep Learning}
}
$$

This hierarchy should be understood as a relationship between fields and approaches.

Artificial Intelligence describes the broader goal and field.

Machine Learning provides methods for learning from data.

Deep Learning provides a family of Machine Learning methods based on multi-layer neural networks.

## 12. Key Differences

The most important distinction is the way the terms describe scope and methodology.

**Artificial Intelligence** focuses on building systems capable of intelligent behavior.

**Machine Learning** focuses on learning patterns from data to build models that can perform a task.

Therefore:

$$
\text{AI} = \text{Broad Field}
$$

$$
\text{ML} = \text{Data-Driven Subfield of AI}
$$

A Machine Learning system can be part of an Artificial Intelligence system, but Artificial Intelligence is not limited to Machine Learning.

## 13. Summary

Artificial Intelligence and Machine Learning are related but distinct concepts.

The key points are:

1. **Artificial Intelligence** is the broader field concerned with intelligent behavior in computational systems.
2. **Machine Learning** is a subfield of Artificial Intelligence.
3. Machine Learning focuses on learning patterns from data.
4. AI systems do not necessarily have to use Machine Learning.
5. Deep Learning is a subfield of Machine Learning.

The overall relationship is:

$$
\boxed{
\text{Artificial Intelligence}
\supset
\text{Machine Learning}
\supset
\text{Deep Learning}
}
$$

Understanding this relationship is important before studying different Machine Learning types, algorithms, and mathematical foundations.
