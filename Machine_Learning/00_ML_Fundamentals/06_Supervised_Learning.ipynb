# 06 — Supervised Learning

## 1. Introduction

**Supervised Learning** is a type of Machine Learning in which a model learns from examples that contain both input data and corresponding target values.

The training data can be represented as:

$$
\mathcal{D}
=
\{(\mathbf{x}_1,y_1),(\mathbf{x}_2,y_2),\ldots,(\mathbf{x}_m,y_m)\}
$$

where:

- \(\mathbf{x}_i\) represents the input features of sample \(i\).
- \(y_i\) represents the corresponding target.
- \(m\) represents the number of training samples.

The objective is to learn a relationship between the input and the target.

Conceptually:

$$
\boxed{
\text{Input}
+
\text{Known Target}
\rightarrow
\text{Learning}
\rightarrow
\text{Model}
}
$$

## 2. Why Is It Called Supervised Learning?

The word **supervised** refers to the presence of known target values during training.

Consider a dataset containing information about houses.

| Area | Bedrooms | Age | Price |
|---:|---:|---:|---:|
| 1000 | 2 | 15 | 180000 |
| 1500 | 3 | 10 | 250000 |
| 2000 | 4 | 5 | 350000 |

The model receives:

- Input features: Area, Bedrooms, Age
- Known target: Price

The known target provides supervision during the learning process.

Therefore:

$$
\text{Features}
\rightarrow
\text{Model}
\rightarrow
\text{Predicted Target}
$$

while the known target is used during training to guide the learning process.

## 3. Input and Target

In supervised learning, the data is generally divided into:

- **Input features**
- **Target**

The input features are represented by:

$$
X
$$

The target values are represented by:

$$
\mathbf{y}
$$

Therefore, the supervised learning dataset can be represented as:

$$
(X,\mathbf{y})
$$

For a single sample:

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n)
$$

and its corresponding target is:

$$
y
$$

The learning problem can therefore be expressed as:

$$
\mathbf{x}
\rightarrow
y
$$

## 4. Learning a Function

The purpose of supervised learning is to learn a function that maps inputs to targets.

We can represent the learned relationship as:

$$
\hat{y} = f(\mathbf{x})
$$

where:

- \(\mathbf{x}\) is the input.
- \(f\) is the learned model.
- \(\hat{y}\) is the predicted output.

During training, the model uses examples containing both \(\mathbf{x}\) and \(y\).

After training, the model can receive a new input:

$$
\mathbf{x}_{new}
$$

and produce a prediction:

$$
\hat{y}_{new}
=
f(\mathbf{x}_{new})
$$

The goal is for the predictions to be useful for previously unseen data.

## 5. Training in Supervised Learning

Training is the process of learning a model from labeled examples.

The basic workflow is:

$$
\text{Training Data}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Trained Model}
$$

The training data contains input-target pairs:

$$
(\mathbf{x}_1,y_1),
(\mathbf{x}_2,y_2),
\ldots,
(\mathbf{x}_m,y_m)
$$

The learning algorithm uses these examples to determine the parameters or structure of a model.

After training, the model can be used to make predictions on new inputs.

## 6. Prediction

After a supervised learning model has been trained, it can be used to make predictions.

Suppose a new sample has feature vector:

$$
\mathbf{x}_{new}
$$

The trained model produces:

$$
\hat{y}_{new}
=
f(\mathbf{x}_{new})
$$

The prediction depends on the type of supervised learning problem.

For example:

- A classification model may predict a class.
- A regression model may predict a numerical value.

Therefore:

$$
\boxed{
\text{New Input}
\rightarrow
\text{Trained Model}
\rightarrow
\text{Prediction}
}
$$

## 7. Main Types of Supervised Learning

The two major supervised learning tasks are:

1. **Classification**
2. **Regression**

The distinction is based mainly on the type of target being predicted.

### Classification

The target represents a category or class.

$$
\mathbf{x}
\rightarrow
\text{Class}
$$

### Regression

The target represents a numerical quantity.

$$
\mathbf{x}
\rightarrow
\text{Numerical Value}
$$

Both tasks use labeled training data.

## 8. Classification

**Classification** is a supervised learning task in which the target represents a discrete category or class.

For example, an email classification problem may have two classes:

$$
y \in \{\text{Spam},\text{Not Spam}\}
$$

The model learns from previously labeled examples.

Conceptually:

$$
\text{Labeled Examples}
\rightarrow
\text{Classification Model}
$$

Then:

$$
\text{New Input}
\rightarrow
\text{Predicted Class}
$$

Classification problems can contain two classes or multiple classes.

## 9. Binary Classification

**Binary classification** is a classification problem with two possible classes.

For example:

$$
y \in \{0,1\}
$$

The two values may represent:

- Spam / Not Spam
- Positive / Negative
- Accepted / Rejected
- Fraud / Not Fraud

The model receives an input and predicts one of the two classes.

$$
\mathbf{x}
\rightarrow
\hat{y}
$$

where:

$$
\hat{y} \in \{0,1\}
$$

## 10. Multiclass Classification

**Multiclass classification** involves more than two possible classes.

For example, an image classification problem may contain:

$$
y \in
\{\text{Cat},\text{Dog},\text{Horse}\}
$$

The model learns to distinguish between the available classes.

Conceptually:

$$
\mathbf{x}
\rightarrow
\{\text{Class}_1,\text{Class}_2,\ldots,\text{Class}_k\}
$$

where \(k\) represents the number of classes.

Multiclass classification is different from binary classification because there are more than two possible target categories.

## 11. Regression

**Regression** is a supervised learning task in which the target represents a numerical value.

For example, a model may predict the price of a house.

The target can be represented as:

$$
y \in \mathbb{R}
$$

The model learns a relationship between the input features and the numerical target.

Conceptually:

$$
\mathbf{x}
\rightarrow
\hat{y}
$$

where \(\hat{y}\) is a predicted numerical value.

Examples of regression tasks include:

- House-price prediction
- Temperature prediction
- Sales prediction
- Demand prediction

## 12. Classification vs Regression

Classification and regression can be compared using the type of target.

| Aspect | Classification | Regression |
|---|---|---|
| Target | Category | Numerical value |
| Output | Class | Number |
| Example | Spam / Not Spam | House Price |
| Main goal | Assign a class | Estimate a value |

The distinction can be summarized as:

$$
\boxed{
\text{Classification}
\rightarrow
\text{Discrete Target}
}
$$

$$
\boxed{
\text{Regression}
\rightarrow
\text{Numerical Target}
}
$$

## 13. Loss and Error

During training, a supervised learning model needs a way to measure how well its predictions match the target values.

This is done using a **loss function** or error measure.

For a sample, the prediction is:

$$
\hat{y}
$$

and the actual target is:

$$
y
$$

The difference between them can be used to measure prediction error.

A general loss function can be written as:

$$
L(y,\hat{y})
$$

The training process attempts to find a model that performs well according to an appropriate loss function.

The exact loss function depends on the supervised learning task and the model.

## 14. Model Training as Optimization

A supervised learning model generally contains parameters that are learned from the training data.

Let the model parameters be:

$$
\theta
$$

The model can be represented as:

$$
\hat{y}=f(\mathbf{x};\theta)
$$

A loss function measures the model's performance.

For a dataset containing \(m\) samples, an objective may be written as:

$$
\theta^*
=
\arg\min_{\theta}
\frac{1}{m}
\sum_{i=1}^{m}
L\left(y_i,f(\mathbf{x}_i;\theta)\right)
$$

The exact optimization procedure depends on the learning algorithm.

This mathematical formulation provides a general view of supervised learning as a problem of finding suitable model parameters.

## 15. Generalization

A supervised learning model should not simply memorize the training examples.

It should learn patterns that allow it to make useful predictions on new data.

This ability is called **generalization**.

The desired behavior is:

$$
\text{Training Examples}
\rightarrow
\text{Learn General Pattern}
\rightarrow
\text{Predict New Examples}
$$

A model that performs well on training data but poorly on unseen data may have learned the training examples too specifically.

This problem is related to **overfitting**, which will be studied separately.

## 16. Training Data and Test Data

To evaluate whether a supervised learning model can generalize, the available data is often divided into different subsets.

A common division is:

- **Training set**
- **Validation set**
- **Test set**

### Training Set

Used to learn the model.

### Validation Set

Used during model development for tasks such as model selection and hyperparameter tuning.

### Test Set

Used to provide an estimate of performance on unseen data after model development.

Conceptually:

$$
\text{Dataset}
\rightarrow
\begin{cases}
\text{Training Set}\\
\text{Validation Set}\\
\text{Test Set}
\end{cases}
$$

The exact splitting strategy depends on the problem and available data.

## 17. Common Supervised Learning Algorithms

Many different algorithms can be used for supervised learning.

Examples include:

### Classification

### Regression

- Linear Regression

## 18. Advantages of Supervised Learning

Supervised Learning has several important advantages.

### 1. Clear Learning Objective

The target values provide a defined objective for learning.

### 2. Measurable Performance

Predictions can be compared with known target values.

### 3. Wide Range of Applications

Supervised learning can be applied to:

- Classification
- Regression
- Prediction
- Detection
- Risk estimation

### 4. Many Available Algorithms

Different algorithms can be selected depending on the problem and data.

The effectiveness of supervised learning still depends strongly on the quality and representativeness of the labeled data.

## 19. Limitations of Supervised Learning

Supervised Learning also has limitations.

### 1. Requires Labeled Data

The training process generally requires known target values.

### 2. Labeling Can Be Expensive

Creating accurate labels may require human effort or domain expertise.

### 3. Poor Data Can Produce Poor Models

Incorrect, incomplete, biased, or unrepresentative training data can negatively affect the learned model.

### 4. Generalization Is Not Guaranteed

A model may perform well on training data but poorly on unseen data.

Therefore:

$$
\boxed{
\text{Good Supervised Learning}
\neq
\text{Only a Good Algorithm}
}
$$

The data, problem formulation, features, training process, evaluation method, and model all matter.

## 20. Complete Supervised Learning Workflow

A simplified supervised learning workflow is:

$$
\text{Labeled Data}
\rightarrow
\text{Data Preparation}
\rightarrow
\text{Model Selection}
\rightarrow
\text{Training}
\rightarrow
\text{Evaluation}
\rightarrow
\text{Prediction}
$$

More explicitly:

1. Obtain labeled data.
2. Understand and prepare the data.
3. Separate input features and targets.
4. Split the data when appropriate.
5. Select a suitable algorithm.
6. Train the model.
7. Evaluate its performance.
8. Tune or improve the model when necessary.
9. Use the trained model on new data.

## 21. Conceptual Example

Consider a model that predicts whether a student will pass an examination.

The input features may include:

- Study hours
- Attendance
- Previous scores

The target may be:

$$
y \in \{0,1\}
$$

where:

- \(0\) represents failure.
- \(1\) represents passing.

The training data contains examples:

$$
(\mathbf{x}_1,y_1),
(\mathbf{x}_2,y_2),
\ldots,
(\mathbf{x}_m,y_m)
$$

The model learns from these examples.

For a new student:

$$
\mathbf{x}_{new}
\rightarrow
\text{Trained Model}
\rightarrow
\hat{y}_{new}
$$

This is a supervised classification problem because the target is a category.
