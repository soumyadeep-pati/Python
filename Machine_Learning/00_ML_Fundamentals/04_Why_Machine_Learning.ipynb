# 04 — Why Machine Learning?

## 1. Introduction

Machine Learning is useful because many real-world problems are difficult to solve by writing explicit rules for every possible situation.

When the relationship between inputs and outputs is complex, a set of manually written rules may become difficult to create, maintain, and update.

Machine Learning provides another approach:

$$
\text{Data}
\rightarrow
\text{Learning}
\rightarrow
\text{Model}
\rightarrow
\text{Prediction}
$$

Instead of manually specifying every rule, we allow an algorithm to learn useful patterns from available data.

## 2. The Problem With Explicit Rules

Consider a system that determines whether an email is spam.

A rule-based system might attempt to define rules such as:

- If a particular word appears, increase the spam score.
- If the email contains many links, increase the spam score.
- If certain patterns appear, classify the email as spam.

The basic process is:

$$
\text{Input}
+
\text{Manually Defined Rules}
\rightarrow
\text{Output}
$$

The difficulty is that real-world data can contain many different patterns.

As the number of possible situations increases, the number of rules may also increase.

This can make the system difficult to design and maintain.

## 3. Learning From Examples

Machine Learning provides an alternative approach.

Instead of manually writing every rule, we can provide examples to a learning algorithm.

For example, a spam classification dataset may contain:

| Email | Target |
|---|---|
| Email 1 | Spam |
| Email 2 | Not Spam |
| Email 3 | Spam |
| Email 4 | Not Spam |

The learning algorithm uses these examples to learn a model.

$$
\text{Examples}
+
\text{Learning Algorithm}
\rightarrow
\text{Model}
$$

The trained model can then be used to process new examples.

## 4. Complex Relationships

Many real-world problems involve relationships between variables that are difficult to describe using simple rules.

Consider house-price prediction.

The price of a house may depend on:

- Area
- Number of bedrooms
- Location
- Age
- Floor
- Nearby facilities
- Other factors

The relationship between these variables and price may be complex.

A Machine Learning model can attempt to learn this relationship from existing examples.

Conceptually:

$$
\text{Features}
\rightarrow
\text{Learned Relationship}
\rightarrow
\text{Prediction}
$$

The model does not need to be given every possible combination of conditions explicitly.

## 5. Handling Large Amounts of Data

Modern systems can generate and store very large amounts of data.

Examples include:

- Images
- Text
- Audio
- Video
- Sensor measurements
- Transactions
- User interactions

Analyzing large datasets manually is difficult.

Machine Learning algorithms can process large amounts of data systematically and learn patterns from that data.

The general idea is:

$$
\text{Large Dataset}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Learned Patterns}
$$

The usefulness of the resulting model depends on the quality and relevance of the available data.

## 6. Prediction

One major reason for using Machine Learning is to make predictions about new or unknown data.

Suppose we have a dataset containing previous observations.

The model learns from those observations and can then be applied to a new input.

$$
\text{Training Data}
\rightarrow
\text{Model}
$$

Then:

$$
\text{New Data}
\rightarrow
\text{Model}
\rightarrow
\text{Prediction}
$$

For example, a model may predict:

- House prices
- Customer demand
- Whether a transaction is fraudulent
- Whether an email is spam
- Whether a customer may leave a service

## 7. Classification

Machine Learning can be used to assign observations to categories.

This task is called **classification**.

For example, an email classification system may predict:

$$
\text{Email}
\rightarrow
\begin{cases}
\text{Spam}\\
\text{Not Spam}
\end{cases}
$$

A medical classification problem may involve different categories.

A Machine Learning model learns patterns from previously observed examples and uses those patterns to classify new observations.

Classification will be studied in detail when supervised learning algorithms are introduced.

## 8. Regression

Machine Learning can also be used to predict numerical values.

This task is called **regression**.

For example, a model may predict the price of a house:

$$
\text{House Features}
\rightarrow
\text{Predicted Price}
$$

The output is a numerical value.

Another example is predicting temperature:

$$
\text{Input Variables}
\rightarrow
\text{Predicted Temperature}
$$

Regression is another important Machine Learning task that will be studied later.

## 9. Pattern Discovery

Machine Learning can also be used when the desired patterns are not explicitly known.

For example, a dataset may contain information about many customers.

A learning algorithm may discover groups of customers with similar characteristics.

Conceptually:

$$
\text{Data}
\rightarrow
\text{Pattern Discovery}
\rightarrow
\text{Groups or Structure}
$$

This type of problem is associated with **unsupervised learning**.

The goal is not necessarily to predict a predefined target.

Instead, the algorithm attempts to discover useful structure in the data.

## 10. Adaptation to New Data

A manually written rule system generally changes only when its rules are changed.

A Machine Learning system can be retrained using new data.

Conceptually:

$$
\text{Old Data}
\rightarrow
\text{Model}
$$

When additional relevant data becomes available:

$$
\text{Old Data} + \text{New Data}
\rightarrow
\text{Retraining}
\rightarrow
\text{Updated Model}
$$

This makes Machine Learning useful in situations where the underlying patterns can change over time.

However, retraining is not automatically required for every Machine Learning system. The appropriate strategy depends on the problem and how the data changes.

## 11. Automation of Data-Driven Decisions

Machine Learning can be used to automate decisions or predictions that would otherwise require repeated manual analysis.

For example, a system may process many transactions and assign each transaction a risk score.

Conceptually:

$$
\text{Transaction Data}
\rightarrow
\text{ML Model}
\rightarrow
\text{Risk Score}
$$

The resulting prediction or score can then be used as part of a larger decision-making system.

Machine Learning does not necessarily replace human decision-making. In many systems, it provides information that supports or assists a decision.

## 12. When Machine Learning Is Appropriate

Machine Learning can be particularly useful when:

1. A large amount of relevant data is available.
2. The relationship between inputs and outputs is difficult to specify manually.
3. The problem involves patterns that can be learned from examples.
4. Predictions or classifications are required for new observations.
5. The system needs to process data repeatedly.
6. The underlying problem can be represented using measurable information.

The presence of data alone does not mean that Machine Learning is the best solution.

The problem, data quality, objectives, computational resources, and required reliability must all be considered.

## 13. When Machine Learning May Not Be Necessary

Machine Learning is not the solution to every problem.

A simple rule-based program may be more appropriate when:

- The rules are clear and stable.
- The problem is simple.
- Very little data is available.
- The required behavior can be precisely specified.
- A Machine Learning model would add unnecessary complexity.

For example, if a program only needs to determine whether a number is even, a simple mathematical rule is sufficient:

$$
n \bmod 2 = 0
$$

There is no need to train a Machine Learning model for such a problem.

Therefore:

$$
\boxed{
\text{Use Machine Learning when learning from data provides value}
}
$$

## 14. Machine Learning as a Problem-Solving Approach

Machine Learning should be viewed as one possible approach to solving a computational problem.

A general problem-solving process is:

$$
\text{Problem}
\rightarrow
\text{Data}
\rightarrow
\text{Learning Method}
\rightarrow
\text{Model}
\rightarrow
\text{Evaluation}
\rightarrow
\text{Solution}
$$

The goal is not simply to use Machine Learning.

The goal is to solve the problem effectively.

Therefore, before selecting an algorithm, we should understand:

- What is the problem?
- What data is available?
- What should the system produce?
- What patterns might be present?
- How will performance be measured?

## 15. Generalization

A Machine Learning model should not simply memorize the training data.

It should learn patterns that are useful for making predictions on new data.

This ability is called **generalization**.

Conceptually:

$$
\text{Training Data}
\rightarrow
\text{Learning}
\rightarrow
\text{Generalizable Model}
$$

The model is then applied to previously unseen data:

$$
\text{Unseen Data}
\rightarrow
\text{Model}
\rightarrow
\text{Prediction}
$$

A model that performs well only on its training data but poorly on new data has not learned a useful general pattern.

Generalization is one of the central ideas in Machine Learning.

## 16. The Core Reason for Machine Learning

The fundamental reason for using Machine Learning is that we can sometimes learn useful behavior from data more effectively than manually specifying all the rules.

The central idea can be expressed as:

$$
\boxed{
\text{Examples}
\rightarrow
\text{Learn Patterns}
\rightarrow
\text{Build Model}
\rightarrow
\text{Make Predictions}
}
$$

Machine Learning becomes valuable when the learned model provides a practical advantage over manually designed rules or other approaches.

## 17. Summary

Machine Learning is useful because it provides a way to learn patterns from data and use those patterns to perform tasks.

The major reasons for using Machine Learning include:

- Handling complex relationships
- Learning from examples
- Processing large amounts of data
- Making predictions
- Performing classification
- Performing regression
- Discovering patterns and structure
- Supporting data-driven decisions
- Adapting models through retraining when appropriate
- Automating repeated data-driven tasks

However, Machine Learning is not always necessary.

A simple rule or traditional algorithm may be better when the problem is simple and the required behavior can be explicitly defined.

The central principle is:

$$
\boxed{
\text{Choose Machine Learning when learning from data provides a useful solution to the problem.}
}
$$
