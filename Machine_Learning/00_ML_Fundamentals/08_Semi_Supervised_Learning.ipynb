# 08 — Semi-Supervised Learning

## 1. Introduction

**Semi-Supervised Learning** is a type of Machine Learning that uses both labeled and unlabeled data during the learning process.

In supervised learning, the training data contains labeled examples:

$$
(X_L,\mathbf{y}_L)
$$

In unsupervised learning, the data does not contain predefined target values:

$$
X_U
$$

Semi-Supervised Learning combines both:

$$
\boxed{
\text{Labeled Data}
+
\text{Unlabeled Data}
\rightarrow
\text{Learning}
}
$$

The main idea is to use a relatively small amount of labeled data together with a larger amount of unlabeled data.

## 2. Why Semi-Supervised Learning Is Needed

Creating labeled datasets can require significant time, effort, and domain knowledge.

For example, consider an image classification problem.

Suppose we have:

$$
10{,}000
$$

images.

It may be possible to collect all of these images automatically, but assigning a correct label to every image may require human effort.

We may therefore have:

- A small set of labeled images.
- A much larger set of unlabeled images.

For example:

$$
\text{Labeled Data} = 1{,}000
$$

$$
\text{Unlabeled Data} = 9{,}000
$$

Semi-Supervised Learning provides methods for learning from both sets of data.

## 3. Labeled and Unlabeled Data

Consider a classification problem.

The labeled dataset can be represented as:

$$
\mathcal{D}_L
=
\{(\mathbf{x}_1,y_1),
(\mathbf{x}_2,y_2),
\ldots,
(\mathbf{x}_l,y_l)\}
$$

where each input has a known target.

The unlabeled dataset can be represented as:

$$
\mathcal{D}_U
=
\{\mathbf{x}_{l+1},
\mathbf{x}_{l+2},
\ldots,
\mathbf{x}_{l+u}\}
$$

where the target values are unknown.

Therefore, the complete available data consists of:

$$
\mathcal{D}
=
\mathcal{D}_L
\cup
\mathcal{D}_U
$$

The learning process attempts to make use of information from both datasets.

## 4. Supervised vs Semi-Supervised Learning

### Supervised Learning

Supervised learning uses labeled examples:

$$
(X,\mathbf{y})
\rightarrow
\text{Model}
$$

The target value is known for each training example.

### Semi-Supervised Learning

Semi-Supervised Learning uses both labeled and unlabeled examples:

$$
(X_L,\mathbf{y}_L)
+
X_U
\rightarrow
\text{Model}
$$

The important difference is that only part of the available data has known target values.

## 5. Unlabeled Data

An **unlabeled example** is an input example for which the desired target value is not provided.

For example:

$$
\mathbf{x}_1 \rightarrow y_1
$$

is labeled because the target \(y_1\) is known.

But:

$$
\mathbf{x}_2 \rightarrow ?
$$

is unlabeled because the target is unknown.

Semi-Supervised Learning attempts to extract useful information from both:

$$
\underbrace{(\mathbf{x}_1,y_1)}_{\text{Labeled}}
\qquad
\underbrace{\mathbf{x}_2}_{\text{Unlabeled}}
$$

The unlabeled examples can provide information about the structure or distribution of the input data.

## 6. Basic Idea

The central idea of Semi-Supervised Learning is that unlabeled data may contain useful information about the problem.

Suppose a small labeled dataset provides information about the relationship between inputs and targets.

A larger unlabeled dataset provides additional information about the distribution and structure of the inputs.

Conceptually:

$$
\text{Labeled Information}
+
\text{Unlabeled Structure}
\rightarrow
\text{Learning}
$$

The exact way this information is combined depends on the semi-supervised learning method.

## 7. Example: Image Classification

Consider an image classification problem.

Suppose we have images of different animals.

A small portion of the images has labels:

| Image | Label |
|---|---|
| Image 1 | Cat |
| Image 2 | Dog |
| Image 3 | Cat |

The remaining images do not have labels:

| Image | Label |
|---|---|
| Image 4 | Unknown |
| Image 5 | Unknown |
| Image 6 | Unknown |

A Semi-Supervised Learning method can use both sets.

The general process is:

$$
\text{Labeled Images}
+
\text{Unlabeled Images}
\rightarrow
\text{Learning}
\rightarrow
\text{Model}
$$

## 8. Why Unlabeled Data Can Be Useful

Unlabeled data does not directly provide target values.

However, it can contain information about the distribution and structure of the inputs.

For example, suppose observations form natural groups in the feature space.

The unlabeled data can provide information about where observations are concentrated and how they are distributed.

Conceptually:

$$
\text{Unlabeled Data}
\rightarrow
\text{Input Structure}
$$

while:

$$
\text{Labeled Data}
\rightarrow
\text{Input-Target Relationship}
$$

Semi-Supervised Learning attempts to use both types of information.

## 9. A General Semi-Supervised Learning Workflow

A simplified workflow is:

$$
\text{Labeled Data}
+
\text{Unlabeled Data}
\rightarrow
\text{Learning Method}
\rightarrow
\text{Model}
$$

A practical process may include:

1. Collect labeled and unlabeled data.
2. Prepare the input data.
3. Use the labeled examples to provide direct supervision.
4. Use the unlabeled examples according to the chosen semi-supervised method.
5. Train the model.
6. Evaluate the model using appropriate labeled evaluation data.
7. Use the trained model for predictions.

## 10. Pseudo-Labels

One common idea in Semi-Supervised Learning is **pseudo-labeling**.

A model can first be trained using labeled data.

It can then make predictions for some unlabeled examples.

If a prediction is considered sufficiently reliable, the predicted class can be treated as a temporary label.

Conceptually:

$$
\text{Labeled Data}
\rightarrow
\text{Initial Model}
$$

Then:

$$
\text{Unlabeled Data}
\rightarrow
\text{Initial Model}
\rightarrow
\text{Pseudo-Labels}
$$

The labeled data and selected pseudo-labeled examples can then be used in further training.

Pseudo-labeling is one approach to Semi-Supervised Learning, not the definition of Semi-Supervised Learning itself.

## 11. Confidence in Pseudo-Labels

Pseudo-labeling can introduce incorrect labels if the model makes unreliable predictions.

For this reason, a confidence criterion may be used.

Suppose a model produces class probabilities:

$$
P(y=k\mid\mathbf{x})
$$

for different classes.

A pseudo-label may be considered only when the model's confidence is sufficiently high.

For example:

$$
\max_k P(y=k\mid\mathbf{x}) \geq \tau
$$

where:

- \(\tau\) is a chosen confidence threshold.
- \(k\) represents a class.

Only predictions satisfying the selected criterion may be used as pseudo-labels.

The choice of threshold affects the quality and quantity of pseudo-labeled data.

## 12. Consistency-Based Learning

Another important idea in Semi-Supervised Learning is **consistency**.

The basic idea is that a model should produce similar predictions when the same unlabeled example is subjected to small, appropriate changes.

Let:

$$
\mathbf{x}
$$

be an unlabeled example.

A transformed version may be represented as:

$$
\tilde{\mathbf{x}}
$$

A consistency-based method encourages:

$$
f(\mathbf{x})
\approx
f(\tilde{\mathbf{x}})
$$

The exact transformations and consistency objective depend on the particular method.

The general idea is to use unlabeled examples by encouraging stable predictions.

## 13. Self-Training

**Self-training** is another general approach to Semi-Supervised Learning.

The process can be described as:

1. Train an initial model using labeled data.
2. Use the model to predict labels for unlabeled data.
3. Select some predictions according to a chosen criterion.
4. Add selected examples to the training data.
5. Retrain the model.
6. Repeat when appropriate.

Conceptually:

$$
\text{Labeled Data}
\rightarrow
\text{Initial Model}
\rightarrow
\text{Predictions on Unlabeled Data}
\rightarrow
\text{Selected Examples}
\rightarrow
\text{Updated Model}
$$

Self-training can be useful, but incorrect pseudo-labels can also propagate errors.

## 14. Semi-Supervised Learning and Clustering

Unsupervised clustering can sometimes provide information that is useful in a semi-supervised setting.

Suppose unlabeled observations naturally form groups.

The clustering structure can provide information about the distribution of the data.

Labeled examples can then provide information about the relationship between groups and target classes.

Conceptually:

$$
\text{Unlabeled Data}
\rightarrow
\text{Structure}
$$

and:

$$
\text{Labeled Data}
\rightarrow
\text{Target Information}
$$

These sources of information can be combined depending on the chosen method.

## 15. Advantages of Semi-Supervised Learning

Semi-Supervised Learning can provide several advantages.

### 1. Reduces Dependence on Large Labeled Datasets

A model may use a large amount of unlabeled data together with a smaller labeled dataset.

### 2. Makes Use of Available Unlabeled Data

Unlabeled data that would otherwise not provide direct supervision can contribute information to the learning process.

### 3. Useful When Labeling Is Expensive

In some domains, obtaining accurate labels requires substantial human effort or specialized knowledge.

### 4. Can Improve Learning

When the assumptions of the method are appropriate, unlabeled data can provide useful additional information.

## 16. Limitations of Semi-Supervised Learning

Semi-Supervised Learning also has important limitations.

### 1. Incorrect Assumptions Can Hurt Performance

Unlabeled data is useful only when the assumptions made by the learning method are appropriate.

### 2. Incorrect Pseudo-Labels Can Propagate Errors

If incorrect predictions are treated as labels, later training may reinforce those errors.

### 3. Evaluation Still Requires Reliable Labels

The quality of a model must be evaluated using suitable labeled evaluation data.

### 4. More Complex Training

Semi-Supervised methods can introduce additional design choices and complexity compared with straightforward supervised learning.

Therefore:

$$
\boxed{
\text{More Unlabeled Data}
\neq
\text{Automatically Better Model}
}
$$

## 17. When Semi-Supervised Learning Is Useful

Semi-Supervised Learning can be considered when:

- A small amount of labeled data is available.
- A large amount of unlabeled data is available.
- Obtaining additional labels is expensive or time-consuming.
- The unlabeled data is relevant to the same problem.
- The assumptions of the chosen semi-supervised method are reasonable.

A common situation is:

$$
|\mathcal{D}_L|
\ll
|\mathcal{D}_U|
$$

where:

- \(|\mathcal{D}_L|\) is the amount of labeled data.
- \(|\mathcal{D}_U|\) is the amount of unlabeled data.

## 18. Supervised vs Unsupervised vs Semi-Supervised

| Aspect | Supervised | Unsupervised | Semi-Supervised |
|---|---|---|---|
| Labeled data | Required | Not required | Available |
| Unlabeled data | Not necessary | Available | Available |
| Main information | Targets | Input structure | Targets + input structure |
| Common goal | Prediction | Structure discovery | Prediction using both sources |
| Example | Image classification with all labels | Customer clustering | Image classification with few labels |

The key distinction is:

$$
\boxed{
\text{Supervised}
\rightarrow
\text{Labeled Data}
}
$$

$$
\boxed{
\text{Unsupervised}
\rightarrow
\text{No Predefined Targets}
}
$$

$$
\boxed{
\text{Semi-Supervised}
\rightarrow
\text{Labeled + Unlabeled Data}
}
$$

## 19. Mathematical View

Let the labeled dataset be:

$$
\mathcal{D}_L
=
\{(\mathbf{x}_i,y_i)\}_{i=1}^{l}
$$

and the unlabeled dataset be:

$$
\mathcal{D}_U
=
\{\mathbf{x}_i\}_{i=l+1}^{l+u}
$$

where:

- \(l\) is the number of labeled examples.
- \(u\) is the number of unlabeled examples.

A semi-supervised method attempts to learn a model using information from both:

$$
\mathcal{D}_L
\cup
\mathcal{D}_U
$$

The exact objective depends on the method.

A general conceptual objective can be written as:

$$
\text{Learning Objective}
=
\text{Supervised Information}
+
\text{Unsupervised Information}
$$

The two components are defined differently by different algorithms.

## 20. Important Point About Semi-Supervised Learning

Semi-Supervised Learning is not simply a mixture of supervised and unsupervised algorithms.

It is a learning setting in which both labeled and unlabeled examples are available and a learning method is designed to use them.

The central idea is:

$$
\boxed{
\text{Small Labeled Dataset}
+
\text{Large Unlabeled Dataset}
\rightarrow
\text{Learning}
}
$$

Whether the unlabeled data improves performance depends on:

- The quality of the data.
- The relationship between labeled and unlabeled examples.
- The assumptions of the learning method.
- The quality of the labels.
- The model and training procedure.

## 21. Summary

Semi-Supervised Learning uses both labeled and unlabeled data.

The basic structure is:

$$
\boxed{
(X_L,\mathbf{y}_L)
+
X_U
\rightarrow
\text{Learning}
\rightarrow
\text{Model}
}
$$

Important concepts introduced in this notebook include:

- Labeled data
- Unlabeled data
- Pseudo-labeling
- Confidence-based selection
- Self-training
- Consistency-based learning
- The relationship between supervised and unsupervised information

The central motivation is:

$$
\boxed{
\text{Use Available Labeled Information}
+
\text{Extract Useful Information From Unlabeled Data}
}
$$

Semi-Supervised Learning is particularly relevant when labeled data is limited but unlabeled data is abundant.
