# 05 — Types of Machine Learning

## 1. Introduction

Machine Learning problems can be categorized according to the type of information available during the learning process and the way the system learns from data.

The major types of Machine Learning introduced in this notebook are:

1. **Supervised Learning**
2. **Unsupervised Learning**
3. **Semi-Supervised Learning**
4. **Reinforcement Learning**

Each type represents a different learning setting.

A simplified view is:

$$
\boxed{
\text{Machine Learning}
\rightarrow
\begin{cases}
\text{Supervised Learning}\\
\text{Unsupervised Learning}\\
\text{Semi-Supervised Learning}\\
\text{Reinforcement Learning}
\end{cases}
}
$$

## 2. Learning From Data

The different types of Machine Learning mainly differ in what information is available to the learning system.

For example, a dataset may contain both input features and known target values:

$$
(X,\mathbf{y})
$$

In another problem, only the input data may be available:

$$
X
$$

In reinforcement learning, an agent interacts with an environment and receives feedback in the form of rewards or penalties.

Therefore, the learning setting can be described by the information available to the learning system.

The four major settings introduced in this notebook are:

- Supervised Learning
- Unsupervised Learning
- Semi-Supervised Learning
- Reinforcement Learning

## 3. Supervised Learning

**Supervised Learning** is a type of Machine Learning in which the learning algorithm is provided with examples containing inputs and corresponding target values.

The training data can be represented as:

$$
\mathcal{D}
=
\{(\mathbf{x}_1,y_1),(\mathbf{x}_2,y_2),\ldots,(\mathbf{x}_m,y_m)\}
$$

where:

- \(\mathbf{x}_i\) represents the input features of sample \(i\).
- \(y_i\) represents the corresponding target.
- \(m\) represents the number of training samples.

The goal is to learn a relationship between the inputs and targets.

Conceptually:

$$
\boxed{
\text{Input}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Target Prediction}
}
$$

The target values provide supervision during the learning process.

## 4. Supervised Learning Example

Consider a house-price dataset.

| Area | Bedrooms | Age | Price |
|---:|---:|---:|---:|
| 1000 | 2 | 15 | 180000 |
| 1500 | 3 | 10 | 250000 |
| 2000 | 4 | 5 | 350000 |

Here:

- Area, bedrooms, and age are input features.
- Price is the target.

The model learns from the known examples.

After training, a new house can be provided to the model:

$$
\mathbf{x}_{new}
\rightarrow
\text{Trained Model}
\rightarrow
\hat{y}_{new}
$$

where \(\hat{y}_{new}\) is the predicted price.

Because the training examples contain known target values, this is a supervised learning problem.

## 5. Main Tasks in Supervised Learning

Two major Machine Learning tasks are commonly studied under supervised learning:

1. **Classification**
2. **Regression**

### Classification

The target represents a category.

$$
\mathbf{x}
\rightarrow
\text{Class}
$$

Examples:

- Spam or not spam
- Disease or no disease
- Cat or dog

### Regression

The target represents a numerical value.

$$
\mathbf{x}
\rightarrow
\text{Numerical Value}
$$

Examples:

- House price
- Temperature
- Sales amount

Classification and regression will be studied in detail in later notebooks.

## 6. Unsupervised Learning

**Unsupervised Learning** is a type of Machine Learning in which the learning algorithm receives input data without predefined target values.

The available data can be represented as:

$$
X
$$

There is no corresponding target vector provided during learning.

The goal is to discover useful patterns, structures, or relationships within the data.

Conceptually:

$$
\boxed{
\text{Input Data}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Discovered Structure}
}
$$

The algorithm attempts to find structure within the available data without being given the correct output for each sample.

## 7. Unsupervised Learning Example

Suppose a dataset contains information about customers:

| Age | Income | Spending |
|---:|---:|---:|
| 22 | 30000 | 25000 |
| 25 | 35000 | 28000 |
| 45 | 90000 | 40000 |
| 48 | 95000 | 42000 |

Suppose no predefined customer categories are provided.

An unsupervised learning algorithm may attempt to identify groups of customers with similar characteristics.

Conceptually:

$$
\text{Customer Data}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Customer Groups}
$$

The groups are discovered from the structure of the data rather than being provided as predefined targets.

## 8. Main Tasks in Unsupervised Learning

Important unsupervised learning tasks include:

### Clustering

Clustering attempts to group similar observations.

$$
\text{Data}
\rightarrow
\text{Groups of Similar Observations}
$$

### Dimensionality Reduction

Dimensionality reduction attempts to represent data using fewer dimensions while preserving useful information.

$$
\text{High-Dimensional Data}
\rightarrow
\text{Lower-Dimensional Representation}
$$

### Representation or Structure Discovery

Unsupervised methods can also be used to discover useful structure or relationships within data.

These tasks will be studied in greater detail in the Unsupervised Learning section.

## 9. Semi-Supervised Learning

**Semi-Supervised Learning** combines labeled and unlabeled data.

In many real-world situations, obtaining labeled data can be expensive or time-consuming, while collecting unlabeled data may be easier.

A semi-supervised dataset may contain:

$$
\mathcal{D}_L
=
\{(\mathbf{x}_1,y_1),\ldots,(\mathbf{x}_l,y_l)\}
$$

for labeled data, and:

$$
\mathcal{D}_U
=
\{\mathbf{x}_{l+1},\ldots,\mathbf{x}_{l+u}\}
$$

for unlabeled data.

The learning process uses both types of information.

Conceptually:

$$
\boxed{
\text{Labeled Data}
+
\text{Unlabeled Data}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Model}
}
$$

## 10. Semi-Supervised Learning Example

Consider an image classification problem.

Suppose we have:

- A small number of images with known labels.
- A large number of images without labels.

For example:

$$
\text{Labeled Images} = 100
$$

$$
\text{Unlabeled Images} = 10{,}000
$$

Obtaining labels for all images may require substantial human effort.

A semi-supervised learning approach can use both the labeled and unlabeled examples.

The general idea is:

$$
\text{Small Labeled Dataset}
+
\text{Large Unlabeled Dataset}
\rightarrow
\text{Learning}
$$

The exact method used depends on the problem and the chosen algorithm.

## 11. Reinforcement Learning

**Reinforcement Learning (RL)** is a type of Machine Learning in which an agent learns by interacting with an environment.

The main components are:

- **Agent** — the learner or decision-making system.
- **Environment** — the system with which the agent interacts.
- **State** — the current situation of the environment.
- **Action** — an operation selected by the agent.
- **Reward** — feedback received after an action.

A simplified interaction is:

$$
\text{State}
\rightarrow
\text{Agent}
\rightarrow
\text{Action}
\rightarrow
\text{Environment}
\rightarrow
\text{Reward}
$$

The agent uses the feedback to improve its future decisions.

## 12. Reinforcement Learning Example

Consider an agent learning to navigate through an environment.

At a particular moment, the agent observes a state:

$$
s_t
$$

The agent selects an action:

$$
a_t
$$

The environment responds with a new state and a reward:

$$
s_{t+1}, r_{t+1}
$$

The interaction can be represented as:

$$
s_t
\rightarrow
a_t
\rightarrow
(s_{t+1},r_{t+1})
$$

Through repeated interaction, the agent can learn a strategy for selecting actions.

The objective is generally related to maximizing cumulative reward over time.

## 13. Comparing the Four Types

The four learning settings can be compared as follows:

| Type | Data / Feedback | Main Goal |
|---|---|---|
| Supervised Learning | Labeled data | Learn to predict targets |
| Unsupervised Learning | Unlabeled data | Discover structure or patterns |
| Semi-Supervised Learning | Labeled + unlabeled data | Learn using both sources |
| Reinforcement Learning | Rewards from interaction | Learn effective actions |

The central difference is the kind of information available to the learning system.

## 15. Important Distinction

The types of Machine Learning should not be understood as completely isolated categories.

Different techniques can sometimes be combined.

For example:

- A system may use supervised learning together with unsupervised methods.
- Semi-supervised methods explicitly combine labeled and unlabeled data.
- Reinforcement learning can use models and techniques that are also found in other areas of Machine Learning.

The classification into types is mainly useful for understanding the learning setting and the type of feedback available to the system.
