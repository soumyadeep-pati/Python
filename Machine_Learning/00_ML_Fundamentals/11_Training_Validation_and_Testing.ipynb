# 11 — Training, Validation, and Testing

## Introduction

A Machine Learning model should not be evaluated only on the data used to train it.

A dataset is commonly divided into separate subsets:

1. Training Set
2. Validation Set
3. Test Set

Each subset has a different purpose.

The basic idea is:

$$
\boxed{
\text{Training} \rightarrow \text{Learning}
}
$$

$$
\boxed{
\text{Validation} \rightarrow \text{Model Selection}
}
$$

$$
\boxed{
\text{Testing} \rightarrow \text{Final Evaluation}
}
$$

This separation helps us determine whether a Machine Learning model can generalize to previously unseen data.

## 1. Why Do We Split the Dataset?

Suppose we have a dataset containing information about houses.

Each sample contains features such as:

- Area
- Number of bedrooms
- Age
- Location

The target is:

- House price

We want to train a Machine Learning model that predicts the price of a new house.

If we train and evaluate the model using exactly the same samples, the evaluation may not represent how the model performs on new data.

A model may perform very well on samples it has already seen but perform poorly on unseen samples.

Therefore, the available dataset is divided into different subsets.

Conceptually:

$$
\text{Dataset}
\rightarrow
\begin{cases}
\text{Training Set} \\
\text{Validation Set} \\
\text{Test Set}
\end{cases}
$$

## 2. Training Set

The **training set** is the portion of the dataset used to train the Machine Learning model.

For supervised learning, the training dataset can be represented as:

$$
D_{\text{train}}
=
\{(x_i,y_i)\}_{i=1}^{m_{\text{train}}}
$$

where:

- $x_i$ is the feature vector of the $i$-th sample.
- $y_i$ is the corresponding target.
- $m_{\text{train}}$ is the number of training samples.

The learning algorithm uses the training data to learn the model parameters.

Conceptually:

$$
\text{Training Data}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Trained Model}
$$

The training set is therefore used primarily for **learning**.

## 3. Validation Set

The **validation set** is a separate portion of the dataset used during model development.

It is used to make decisions about the model, such as:

- Comparing different models.
- Selecting hyperparameters.
- Comparing different model configurations.
- Detecting possible overfitting during development.
- Selecting the final model configuration.

Conceptually:

$$
\text{Training Set}
\rightarrow
\text{Train Candidate Models}
$$

$$
\text{Validation Set}
\rightarrow
\text{Compare Candidate Models}
$$

The validation set therefore supports **model development and selection**.

## 4. Test Set

The **test set** is a separate portion of the dataset reserved for final evaluation.

It should not be used to train the model or repeatedly make development decisions.

After the model and its configuration have been selected, the test set is used to estimate how the final model performs on unseen data.

Conceptually:

$$
\text{Training Set}
\rightarrow
\text{Learning}
$$

$$
\text{Validation Set}
\rightarrow
\text{Model Selection}
$$

$$
\text{Test Set}
\rightarrow
\text{Final Evaluation}
$$

The test set should remain as independent from the development process as possible.

## 5. Training, Validation, and Test Sets Together

The three subsets have different responsibilities.

| Dataset | Main Purpose |
|---|---|
| Training Set | Learn model parameters |
| Validation Set | Select and improve the model |
| Test Set | Perform final evaluation |

The complete development process can be represented as:

$$
\text{Dataset}
\rightarrow
\text{Train / Validation / Test Split}
$$

Then:

$$
\text{Training Set}
\rightarrow
\text{Model Training}
$$

$$
\text{Validation Set}
\rightarrow
\text{Model Selection}
$$

$$
\text{Test Set}
\rightarrow
\text{Final Evaluation}
$$

The most important distinction is:

> The test set should not influence model development.

## 6. Dataset Split

Suppose a dataset contains $m$ total samples.

The dataset can be divided into:

$$
m
=
m_{\text{train}}
+
m_{\text{validation}}
+
m_{\text{test}}
$$

where:

- $m_{\text{train}}$ = number of training samples.
- $m_{\text{validation}}$ = number of validation samples.
- $m_{\text{test}}$ = number of test samples.

Therefore:

$$
D
=
D_{\text{train}}
\cup
D_{\text{validation}}
\cup
D_{\text{test}}
$$

The subsets should contain different samples.

The purpose of the split is to create separate data for learning, development, and final evaluation.

## 7. Common Dataset Split Ratios

There is no universal split ratio that is correct for every Machine Learning problem.

Common examples include:

### 80 / 20

$$
80\% \rightarrow \text{Training}
$$

$$
20\% \rightarrow \text{Testing}
$$

This is useful when a separate validation set is not required.

### 70 / 15 / 15

$$
70\% \rightarrow \text{Training}
$$

$$
15\% \rightarrow \text{Validation}
$$

$$
15\% \rightarrow \text{Testing}
$$

### 60 / 20 / 20

$$
60\% \rightarrow \text{Training}
$$

$$
20\% \rightarrow \text{Validation}
$$

$$
20\% \rightarrow \text{Testing}
$$

These are examples rather than strict rules.

The appropriate division depends on:

- Dataset size
- Problem type
- Amount of available data
- Model complexity
- Evaluation requirements

## 8. Model Parameters

A **model parameter** is a value learned by the Machine Learning algorithm from the training data.

Consider a simple linear model:

$$
\hat{y}=w_1x_1+w_2x_2+b
$$

The parameters are:

$$
w_1,\;w_2,\;b
$$

These values are learned during training.

The training process attempts to find parameter values that produce good predictions for the training data.

In general:

$$
\theta^*
=
\arg\min_{\theta}
\frac{1}{m}
\sum_{i=1}^{m}
L(y_i,f(x_i;\theta))
$$

where:

- $\theta$ represents the model parameters.
- $L$ is a loss function.
- $f(x_i;\theta)$ is the model prediction.
- $y_i$ is the actual target.

## 9. Hyperparameters

A **hyperparameter** is a configuration value that controls aspects of the learning process or model structure.

Examples include:

- Learning rate
- Number of neighbors in KNN
- Maximum tree depth
- Number of trees
- Regularization strength

Hyperparameters are different from model parameters.

### Parameters

Learned from training data.

### Hyperparameters

Selected as part of the model-development process.

The validation set can be used to compare different hyperparameter configurations.

For example:

$$
\lambda_1,\lambda_2,\lambda_3
$$

may represent three different regularization values.

We can train models using these configurations and compare their validation performance.

## 10. Model Selection Using the Validation Set

Suppose we train three candidate models:

$$
M_1,\;M_2,\;M_3
$$

Their validation performance is:

| Model | Validation Performance |
|---|---:|
| $M_1$ | 82% |
| $M_2$ | 89% |
| $M_3$ | 85% |

Based on the validation results, $M_2$ would be selected during model development.

The test set remains untouched.

After selecting the final model, we evaluate $M_2$ on the test set.

The workflow is therefore:

$$
\text{Training}
\rightarrow
\text{Validation}
\rightarrow
\text{Model Selection}
\rightarrow
\text{Testing}
$$

## 11. Training Error

**Training error** is the error measured on the training dataset.

For example, Mean Squared Error is:

$$
MSE
=
\frac{1}{m}
\sum_{i=1}^{m}
(y_i-\hat{y}_i)^2
$$

where:

- $y_i$ is the actual target.
- $\hat{y}_i$ is the model prediction.
- $m$ is the number of samples.

A low training error means that the model fits the training data well.

However:

> Low training error does not necessarily mean that the model will perform well on unseen data.

A model may learn the training data too closely and fail to generalize.

## 12. Validation Error

**Validation error** is the error measured on the validation dataset.

The validation set contains samples that were not used as the primary training data.

Validation performance provides information about how the current model configuration performs on unseen development data.

For example:

| Model | Training Error | Validation Error |
|---|---:|---:|
| Model A | Low | Low |
| Model B | Very Low | High |
| Model C | High | High |

Model B may have learned the training data extremely well but perform poorly on unseen validation data.

This can indicate possible **overfitting**.

Overfitting and underfitting will be studied in detail in a later notebook.

## 13. Test Error

**Test error** is the error measured on the test dataset after the final model has been selected.

The test set is intended to represent unseen data.

The process is:

$$
\text{Training Data}
\rightarrow
\text{Model Development}
$$

$$
\text{Validation Data}
\rightarrow
\text{Model Selection}
$$

$$
\text{Test Data}
\rightarrow
\text{Final Evaluation}
$$

The test result provides an estimate of how well the final model may perform on new data.

The test set should therefore be kept separate from model-development decisions.

## 14. Generalization

**Generalization** is the ability of a Machine Learning model to perform well on previously unseen data.

A model should not simply memorize the training samples.

Instead, it should learn patterns that are useful for new samples.

Conceptually:

$$
\text{Training Data}
\rightarrow
\text{Learn General Patterns}
\rightarrow
\text{Unseen Data}
\rightarrow
\text{Good Predictions}
$$

A model that performs well on training data but poorly on unseen data has poor generalization.

Therefore, evaluation using data that was not used for training is fundamental to Machine Learning.

## 15. Why the Test Set Should Not Be Used for Model Selection

Suppose we train several models and repeatedly evaluate all of them on the test set.

Eventually, we select the model that performs best on the test set.

The problem is that the test results have now influenced our model-development decisions.

The test set is no longer truly independent of the development process.

This can result in an overly optimistic estimate of performance.

The preferred workflow is:

$$
\text{Training Set}
\rightarrow
\text{Train Models}
$$

$$
\text{Validation Set}
\rightarrow
\text{Select Model}
$$

$$
\text{Test Set}
\rightarrow
\text{Final Evaluation}
$$

The test set should be treated as a final evaluation resource.

## 16. Data Leakage

**Data leakage** occurs when information that should not be available during model training or development becomes available to the model.

Data leakage can produce unrealistically good evaluation results.

Examples include:

- Using test-set information during training.
- Using future information to predict the past.
- Performing data-dependent preprocessing on the entire dataset before splitting.
- Allowing target-related information to enter the input features.

The fundamental principle is:

> Information from evaluation data should not improperly influence the training process.

## 17. Splitting Before Data-Dependent Preprocessing

Some preprocessing operations learn information from data.

Examples include:

- Calculating means.
- Calculating standard deviations.
- Learning category mappings.
- Selecting features using statistical information.

Suppose a mean is calculated from the training data:

$$
\mu_{\text{train}}
=
\frac{1}{m_{\text{train}}}
\sum_{i=1}^{m_{\text{train}}}x_i
$$

This training-derived value can then be used to transform other datasets.

The conceptual workflow is:

$$
\text{Original Dataset}
\rightarrow
\text{Split}
$$

Then:

$$
\text{Training Data}
\rightarrow
\text{Learn Preprocessing Parameters}
$$

Finally:

$$
\text{Learned Transformation}
\rightarrow
\text{Validation Data and Test Data}
$$

This helps prevent information from the validation or test sets from leaking into the training process.

## 18. Random Splitting

A common approach is to randomly assign samples to different subsets.

Conceptually:

$$
D
\rightarrow
D_{\text{train}}
\cup
D_{\text{validation}}
\cup
D_{\text{test}}
$$

Random splitting can help prevent the original ordering of the dataset from determining the subsets.

However, random splitting is not appropriate for every problem.

The splitting strategy should reflect the structure of the data.

For independent samples, random splitting is often appropriate.

For structured or time-dependent data, another strategy may be required.

## 19. Stratified Splitting

For classification problems, the distribution of classes should be considered when splitting the dataset.

Suppose a dataset contains:

- 900 samples of Class A
- 100 samples of Class B

Therefore:

$$
P(y=\text{Class A})=0.9
$$

and:

$$
P(y=\text{Class B})=0.1
$$

A poor split could produce training, validation, and test sets with very different class distributions.

**Stratified splitting** attempts to preserve the class proportions across the subsets.

Conceptually:

$$
P_{\text{train}}(y)
\approx
P_{\text{validation}}(y)
\approx
P_{\text{test}}(y)
$$

when stratification is appropriate.

This can make evaluation more representative of the original dataset.

## 20. Time-Dependent Data

For time-dependent data, randomly mixing past and future observations can cause information from the future to influence evaluation.

Suppose we want to predict future values.

A chronological split can be represented as:

$$
\text{Past Data}
\rightarrow
\text{Training}
$$

$$
\text{Later Data}
\rightarrow
\text{Validation}
$$

$$
\text{Future Data}
\rightarrow
\text{Testing}
$$

If:

$$
t_1<t_2<t_3<\cdots<t_n
$$

then the training data should generally occur earlier in time than the validation and test data.

This makes the evaluation more representative of the real-world prediction scenario.

## 21. Training, Validation, and Test Sets — Example

Suppose we have a student-performance dataset.

Each sample contains:

- Study hours
- Attendance
- Previous score
- Assignment score

The target is:

- Final examination score

Suppose there are:

$$
1,000
$$

samples.

We divide them into:

$$
700 \rightarrow \text{Training}
$$

$$
150 \rightarrow \text{Validation}
$$

$$
150 \rightarrow \text{Testing}
$$

### Training

The model learns relationships between the features and target using the 700 training samples.

### Validation

Different model configurations are compared using the 150 validation samples.

### Testing

After selecting the final model, its performance is evaluated using the 150 test samples.

The test result provides a final estimate of performance on unseen data.

## 22. Validation Set Is Not Always Necessary

A separate validation set is useful, but it is not always required.

For small datasets, creating separate training, validation, and test sets may leave too few samples for training.

In such situations, **cross-validation** can be used to make more efficient use of the available training data.

Cross-validation divides the available training data into multiple parts and repeatedly trains and evaluates the model using different partitions.

The detailed study of cross-validation will be covered later.

Therefore:

> The exact evaluation strategy depends on the dataset, problem, and model-development requirements.

## 23. Small and Large Datasets

The appropriate dataset-splitting strategy depends partly on the amount of available data.

### Large Dataset

When many samples are available, separate training, validation, and test sets can often be created.

For example:

$$
80\% \rightarrow \text{Training}
$$

$$
10\% \rightarrow \text{Validation}
$$

$$
10\% \rightarrow \text{Testing}
$$

### Small Dataset

When fewer samples are available, allocating a large amount of data to validation and testing may reduce the amount available for training.

Cross-validation can be useful in such situations.

The objective is not to follow a fixed percentage.

The objective is to obtain a reliable estimate of how the model will perform on unseen data.

## 24. Common Mistakes

### Mistake 1 — Training and Testing on the Same Data

This does not properly measure generalization.

### Mistake 2 — Using the Test Set for Hyperparameter Tuning

This allows the test set to influence model development.

### Mistake 3 — Preprocessing Before Splitting

If preprocessing learns information from the complete dataset, information from evaluation data can leak into training.

### Mistake 4 — Ignoring Class Distribution

For classification problems, an inappropriate split can create unrepresentative subsets.

### Mistake 5 — Randomly Splitting Time-Dependent Data

Random splitting can allow future information to influence training.

### Mistake 6 — Repeatedly Checking the Test Set

Repeatedly using the test set during development reduces its value as an independent final evaluation.

## 25. Training vs Validation vs Testing

The three stages answer different questions.

### Training

> How can the model learn from the available training examples?

### Validation

> Which model or configuration should we choose?

### Testing

> How well does the final selected model perform on unseen data?

Therefore:

$$
\boxed{
\begin{aligned}
\text{Training} &\rightarrow \text{Learn} \\
\text{Validation} &\rightarrow \text{Select} \\
\text{Testing} &\rightarrow \text{Evaluate}
\end{aligned}
}
$$

Keeping these responsibilities separate is fundamental to reliable Machine Learning evaluation.

## 26. Key Concepts

The important concepts from this notebook are:

- **Training Set** — used to learn model parameters.
- **Validation Set** — used during model development and model selection.
- **Test Set** — reserved for final evaluation.
- **Model Parameters** — learned from training data.
- **Hyperparameters** — selected as part of model development.
- **Training Error** — error measured on training data.
- **Validation Error** — error measured on validation data.
- **Test Error** — error measured on test data.
- **Generalization** — ability to perform well on unseen data.
- **Data Leakage** — improper transfer of information into training or development.
- **Random Splitting** — commonly used when samples can be treated independently.
- **Stratified Splitting** — helps preserve class proportions.
- **Chronological Splitting** — useful for time-dependent data.
- **Cross-Validation** — an alternative evaluation strategy that will be studied later.

## 27. Final Summary

Machine Learning models should be evaluated on data that was not used to directly train them.

A dataset can be divided into:

$$
\text{Dataset}
\rightarrow
\begin{cases}
\text{Training Set} \\
\text{Validation Set} \\
\text{Test Set}
\end{cases}
$$

Their primary purposes are:

$$
\boxed{
\text{Training} \rightarrow \text{Learn}
}
$$

$$
\boxed{
\text{Validation} \rightarrow \text{Select}
}
$$

$$
\boxed{
\text{Testing} \rightarrow \text{Final Evaluation}
}
$$

The ultimate goal is **generalization**.

A good Machine Learning model should learn useful patterns from the training data and apply those patterns effectively to previously unseen data.

The test set should remain isolated from model development so that it can provide a meaningful final estimate of model performance.
