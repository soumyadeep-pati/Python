# 07 — Unsupervised Learning

## 1. Introduction

**Unsupervised Learning** is a type of Machine Learning in which a model learns from input data without being provided with predefined target values.

In supervised learning, the training data contains input-target pairs:

$$
(X,\mathbf{y})
$$

In unsupervised learning, the learning algorithm receives the input data without corresponding target values:

$$
X
$$

The general idea is:

$$
\boxed{
\text{Input Data}
\rightarrow
\text{Learning Algorithm}
\rightarrow
\text{Discovered Structure}
}
$$

The purpose of unsupervised learning is to discover useful patterns, relationships, representations, or structures within the data.

## 2. Why Is It Called Unsupervised Learning?

The term **unsupervised** refers to the absence of predefined target values during learning.

Consider a dataset containing information about customers:

| Age | Income | Spending |
|---:|---:|---:|
| 22 | 30000 | 25000 |
| 25 | 35000 | 28000 |
| 45 | 90000 | 40000 |
| 48 | 95000 | 42000 |

Suppose the dataset does not contain a column specifying the customer group.

The learning algorithm must discover structure from the available features.

Therefore:

$$
\text{Input Data}
\rightarrow
\text{Learning}
\rightarrow
\text{Discovered Structure}
$$

There is no predefined target telling the algorithm what the correct output should be.

## 3. Supervised vs Unsupervised Learning

The main difference between supervised and unsupervised learning is the availability of target values.

### Supervised Learning

The dataset contains inputs and targets:

$$
(X,\mathbf{y})
$$

The model learns a relationship:

$$
\mathbf{x}
\rightarrow
y
$$

### Unsupervised Learning

The dataset contains input data without predefined targets:

$$
X
$$

The algorithm attempts to discover structure:

$$
X
\rightarrow
\text{Structure}
$$

Therefore:

$$
\boxed{
\text{Supervised Learning}
\rightarrow
\text{Known Targets}
}
$$

$$
\boxed{
\text{Unsupervised Learning}
\rightarrow
\text{No Predefined Targets}
}
$$

## 4. Main Goals of Unsupervised Learning

Unsupervised learning can be used for several purposes.

Important goals include:

1. **Clustering**
2. **Dimensionality Reduction**
3. **Representation Learning**
4. **Pattern Discovery**
5. **Structure Discovery**

These goals are related because the algorithm attempts to learn useful information from the structure of the input data without relying on predefined target values.

## 5. Clustering

**Clustering** is an unsupervised learning task in which observations are grouped according to their similarity.

Suppose we have a dataset containing observations:

$$
X =
\{\mathbf{x}_1,\mathbf{x}_2,\ldots,\mathbf{x}_m\}
$$

A clustering algorithm attempts to divide the observations into groups.

Conceptually:

$$
\text{Data}
\rightarrow
\text{Similarity or Structure}
\rightarrow
\text{Clusters}
$$

Observations within the same cluster are generally intended to be more similar according to the chosen clustering method than observations assigned to different clusters.

## 6. Clustering Example

Consider customers described by:

- Age
- Income
- Spending

Suppose no customer categories are provided.

A clustering algorithm may discover groups such as:

- Customers with relatively low income and low spending
- Customers with relatively high income and high spending
- Customers with different combinations of income and spending

The important point is that these groups are **discovered from the data**.

The categories are not provided as target labels during training.

Conceptually:

$$
\text{Customer Data}
\rightarrow
\text{Clustering Algorithm}
\rightarrow
\text{Customer Groups}
$$

## 7. Clusters and Similarity

Clustering depends on a notion of similarity or distance between observations.

Suppose two observations are represented as:

$$
\mathbf{x}_1=(x_{11},x_{12})
$$

and:

$$
\mathbf{x}_2=(x_{21},x_{22})
$$

A distance measure can be used to quantify how far apart they are.

One common distance measure is Euclidean distance:

$$
d(\mathbf{x}_1,\mathbf{x}_2)
=
\sqrt{
(x_{11}-x_{21})^2
+
(x_{12}-x_{22})^2
}
$$

A clustering algorithm may use a distance or similarity measure to determine which observations should be grouped together.

The choice of representation and distance measure can strongly affect the resulting clusters.

## 8. Dimensionality Reduction

Another important unsupervised learning task is **dimensionality reduction**.

A dataset may contain a large number of features.

For example:

$$
X \in \mathbb{R}^{m \times n}
$$

where \(n\) may be large.

Dimensionality reduction attempts to represent the data using fewer dimensions while preserving useful information.

Conceptually:

$$
\boxed{
\text{High-Dimensional Data}
\rightarrow
\text{Lower-Dimensional Representation}
}
$$

The new representation may be easier to visualize, analyze, store, or use in subsequent learning tasks.

## 9. Example of Dimensionality Reduction

Suppose a dataset contains:

$$
100
$$

features for every sample.

A dimensionality reduction method may transform the data into a representation containing only:

$$
2
$$

or:

$$
10
$$

dimensions.

Conceptually:

$$
\mathbf{x}\in\mathbb{R}^{100}
\rightarrow
\mathbf{z}\in\mathbb{R}^{2}
$$

The reduced representation does not necessarily preserve every detail of the original data.

Instead, the objective is generally to preserve useful information according to the chosen method.

## 10. Visualization Through Dimensionality Reduction

Dimensionality reduction can also help visualize high-dimensional data.

A dataset with many features cannot be directly visualized in ordinary two-dimensional space.

A dimensionality reduction method can create a lower-dimensional representation:

$$
\mathbb{R}^{n}
\rightarrow
\mathbb{R}^{2}
$$

The resulting two-dimensional representation can then be plotted.

Conceptually:

$$
\text{High-Dimensional Data}
\rightarrow
\text{2D Representation}
\rightarrow
\text{Visualization}
$$

This can help us inspect possible patterns or structures in the data.

## 11. Representation Learning

Unsupervised learning can also be used to learn useful representations of data.

A representation is another way of describing the original data.

Let the original input be:

$$
\mathbf{x}
$$

A learning method may transform it into another representation:

$$
\mathbf{z}=g(\mathbf{x})
$$

where:

- \(\mathbf{x}\) is the original representation.
- \(g\) is a learned transformation.
- \(\mathbf{z}\) is the new representation.

The new representation may capture useful characteristics of the original data.

## 12. Pattern Discovery

Unsupervised learning can be used to discover patterns that are not explicitly specified beforehand.

Suppose a dataset contains many observations.

We may not know which observations are similar or which groups exist.

An unsupervised algorithm can attempt to identify structure:

$$
X
\rightarrow
\text{Pattern Discovery}
\rightarrow
\text{Useful Structure}
$$

The discovered structure depends on:

- The data
- The representation
- The algorithm
- The assumptions made by the method
- The similarity or objective used by the algorithm

## 13. Association and Relationships

Unsupervised methods can also be used to discover relationships between variables or events.

For example, a dataset of customer transactions may contain many products purchased together.

An analysis may reveal that certain items frequently occur together.

The general idea is:

$$
\text{Transaction Data}
\rightarrow
\text{Relationship Discovery}
$$

This type of analysis can be useful for understanding patterns in collections of observations.

Association analysis is one of the topics that can be studied within the broader area of unsupervised learning.

## 15. K-Means as a Clustering Method

**K-Means** is a commonly used clustering algorithm.

Given a chosen number of clusters \(k\), the algorithm attempts to assign observations to clusters and identify a representative center for each cluster.

Let the cluster centers be:

$$
\boldsymbol{\mu}_1,
\boldsymbol{\mu}_2,
\ldots,
\boldsymbol{\mu}_k
$$

The basic objective can be expressed as minimizing the within-cluster squared distance:

$$
\min
\sum_{i=1}^{m}
\left\|
\mathbf{x}_i-\boldsymbol{\mu}_{c_i}
\right\|^2
$$

where \(c_i\) represents the cluster assigned to observation \(i\).

K-Means will be studied in detail later.

## 16. Principal Component Analysis

**Principal Component Analysis (PCA)** is a dimensionality reduction technique.

PCA transforms the original features into a new set of directions called principal components.

The first principal component captures the direction of greatest variance according to the PCA objective.

Conceptually:

$$
\text{Original Features}
\rightarrow
\text{Principal Components}
\rightarrow
\text{Reduced Representation}
$$

For an original feature space of dimension \(n\), we may select only the first \(k\) components:

$$
k<n
$$

resulting in a lower-dimensional representation.

PCA will be studied in detail later.

## 17. Advantages of Unsupervised Learning

Unsupervised learning has several important advantages.

### 1. Does Not Require Predefined Targets

The algorithm can work with data for which target labels are unavailable.

### 2. Useful for Exploration

It can help reveal patterns and structure that were not known beforehand.

### 3. Useful for Large Unlabeled Datasets

Many real-world datasets contain large amounts of information without manually assigned labels.

### 4. Can Support Other Tasks

Unsupervised methods can sometimes be used for:

- Data exploration
- Visualization
- Feature representation
- Preprocessing
- Segmentation
- Anomaly analysis

The usefulness of an unsupervised method depends on whether the discovered structure is meaningful for the problem.

## 18. Limitations of Unsupervised Learning

Unsupervised learning also has important limitations.

### 1. No Direct Target

There is no predefined target that directly tells the algorithm what the correct output should be.

### 2. Evaluation Can Be Difficult

It may be difficult to determine whether the discovered structure is meaningful.

### 3. Results Depend on Assumptions

Different algorithms or distance measures can produce different structures.

### 4. Discovered Patterns May Not Be Useful

An algorithm can find patterns that exist mathematically in the data but have little practical meaning.

Therefore:

$$
\boxed{
\text{Discovered Structure}
\neq
\text{Automatically Useful Structure}
}
$$

Interpretation and domain knowledge can be important when evaluating unsupervised learning results.

## 19. Supervised vs Unsupervised Learning

| Aspect | Supervised Learning | Unsupervised Learning |
|---|---|---|
| Target values | Available | Not predefined |
| Main goal | Predict targets | Discover structure |
| Training data | Labeled | Unlabeled |
| Common tasks | Classification, Regression | Clustering, Dimensionality Reduction |
| Evaluation | Often directly compared with targets | Often more dependent on the objective and interpretation |
| Example | Predict house price | Group similar customers |

The fundamental distinction is:

$$
\boxed{
\text{Supervised}
\rightarrow
\text{Learn From Known Targets}
}
$$

$$
\boxed{
\text{Unsupervised}
\rightarrow
\text{Discover Structure Without Known Targets}
}
$$

## 20. Complete Unsupervised Learning Workflow

A simplified unsupervised learning workflow is:

$$
\text{Unlabeled Data}
\rightarrow
\text{Data Preparation}
\rightarrow
\text{Unsupervised Algorithm}
\rightarrow
\text{Discovered Structure}
\rightarrow
\text{Analysis}
$$

A practical workflow may include:

1. Obtain the data.
2. Understand the features.
3. Prepare the data.
4. Select an appropriate unsupervised learning method.
5. Train or apply the method.
6. Analyze the resulting structure.
7. Evaluate whether the discovered structure is useful.
8. Interpret the results in the context of the problem.

## 21. Important Point About Unsupervised Learning

Unsupervised learning does not mean that the algorithm produces completely arbitrary results.

The algorithm still follows a defined objective, mathematical formulation, or set of assumptions.

For example, a clustering algorithm may attempt to minimize distances within clusters.

A dimensionality reduction method may attempt to preserve certain properties of the original data.

Therefore:

$$
\boxed{
\text{No Target}
\neq
\text{No Objective}
}
$$

Unsupervised learning still requires a well-defined learning method and objective.
