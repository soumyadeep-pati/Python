# 09 — Reinforcement Learning

## 1. Introduction

**Reinforcement Learning (RL)** is a type of Machine Learning in which an agent learns how to make decisions by interacting with an environment.

Unlike supervised learning, the agent is not provided with a correct target for every situation.

Instead, the agent:

1. Observes the current state.
2. Selects an action.
3. Receives feedback from the environment.
4. Observes the resulting state.
5. Uses the feedback to improve future decisions.

The basic interaction is:

$$
\boxed{
\text{State}
\rightarrow
\text{Action}
\rightarrow
\text{Reward}
\rightarrow
\text{Learning}
}
$$

The objective is to learn a strategy that produces desirable outcomes over time.

## 2. Why Is It Called Reinforcement Learning?

The term **reinforcement** refers to learning through feedback.

An agent performs an action and receives a reward or penalty from the environment.

A positive reward can indicate that an action was useful, while a negative reward can indicate an undesirable outcome.

The agent uses this feedback to improve its future behavior.

Conceptually:

$$
\text{Action}
\rightarrow
\text{Feedback}
\rightarrow
\text{Learning}
$$

The agent is therefore not simply memorizing predefined answers.

It is learning from the consequences of its actions.

## 3. Agent and Environment

The two central components of Reinforcement Learning are:

### Agent

The **agent** is the learner or decision-making system.

It observes the environment, selects actions, and receives feedback.

### Environment

The **environment** is the system with which the agent interacts.

It receives an action from the agent and responds by producing a new state and a reward.

The interaction can be represented as:

$$
\boxed{
\text{Agent}
\leftrightarrow
\text{Environment}
}
$$

The agent learns by repeatedly interacting with the environment.

## 4. State

A **state** represents the situation of the environment at a particular time.

A state can be represented as:

$$
s_t
$$

where \(t\) represents the current time step.

For example, in a game, the state may contain information about:

- The current position of the player.
- The positions of other objects.
- The current score.
- Other relevant information.

The state should contain information that is useful for deciding what action to take.

Conceptually:

$$
\text{Current Situation}
\rightarrow
s_t
$$

## 5. Action

An **action** is a decision or operation selected by the agent.

An action at time \(t\) can be represented as:

$$
a_t
$$

For example, in a simple navigation problem, the available actions may be:

$$
\mathcal{A}
=
\{\text{Up},\text{Down},\text{Left},\text{Right}\}
$$

The agent observes the current state and selects an action from the available action space.

Conceptually:

$$
s_t
\rightarrow
a_t
$$

## 6. Reward

A **reward** is feedback provided by the environment after an action.

The reward at time \(t\) can be represented as:

$$
r_{t+1}
$$

A reward can be:

- Positive
- Zero
- Negative

The meaning of the reward depends on the problem.

For example, an environment may provide:

$$
r = +1
$$

for achieving a desired outcome, and:

$$
r = -1
$$

for an undesirable outcome.

The reward provides information about the consequences of the agent's action.

## 7. State Transition

After the agent selects an action, the environment changes.

Suppose the current state is:

$$
s_t
$$

and the agent selects:

$$
a_t
$$

The environment then produces a new state:

$$
s_{t+1}
$$

and a reward:

$$
r_{t+1}
$$

The interaction can therefore be represented as:

$$
\boxed{
(s_t,a_t)
\rightarrow
(s_{t+1},r_{t+1})
}
$$

The agent observes this outcome and continues interacting with the environment.

## 8. The Reinforcement Learning Loop

The basic Reinforcement Learning loop is:

$$
s_t
\rightarrow
a_t
\rightarrow
r_{t+1},s_{t+1}
\rightarrow
a_{t+1}
\rightarrow
r_{t+2},s_{t+2}
\rightarrow
\cdots
$$

In words:

1. Observe the current state.
2. Select an action.
3. Receive a reward.
4. Observe the new state.
5. Select the next action.
6. Continue the interaction.

Through repeated interactions, the agent attempts to improve its decision-making behavior.

## 9. Policy

A **policy** describes how an agent chooses actions based on the current state.

A policy can be represented as:

$$
\pi(a\mid s)
$$

This represents the behavior of the agent when it is in state \(s\).

For a deterministic policy, we can write:

$$
a = \pi(s)
$$

This means that the policy directly determines the action for a given state.

A policy is therefore a central object that the agent attempts to learn or improve.

## 10. Objective of Reinforcement Learning

The objective of Reinforcement Learning is generally not to maximize only the immediate reward.

The agent often needs to consider rewards that may be received in the future.

Suppose the agent receives a sequence of rewards:

$$
r_1,r_2,r_3,\ldots
$$

The agent attempts to maximize the cumulative return.

A discounted return can be represented as:

$$
G_t
=
r_{t+1}
+
\gamma r_{t+2}
+
\gamma^2 r_{t+3}
+
\cdots
$$

where:

$$
0 \leq \gamma \leq 1
$$

is the discount factor.

The discount factor controls the importance assigned to future rewards.

## 11. Immediate vs Future Rewards

An important characteristic of Reinforcement Learning is that an action can have consequences beyond the immediate next step.

Suppose an action produces:

$$
r_{t+1}=1
$$

but leads to a state from which the agent can obtain much larger rewards later.

Another action may produce:

$$
r_{t+1}=5
$$

but lead to poor future outcomes.

A good policy should consider the long-term consequences of actions.

Therefore:

$$
\boxed{
\text{Good Action}
\neq
\text{Always the Action With the Largest Immediate Reward}
}
$$

The objective is generally to maximize cumulative future reward.

## 12. Value Function

A **value function** estimates how useful it is for an agent to be in a particular state.

The value of a state can be represented as:

$$
V^\pi(s)
$$

where \(\pi\) is the policy being followed.

Conceptually:

$$
V^\pi(s)
=
\text{Expected Future Return From State }s
$$

A state with a high value is expected to lead to better future outcomes under the policy.

The exact mathematical definition depends on the reinforcement learning formulation.

## 13. Action-Value Function

The **action-value function** evaluates a state-action pair.

It can be represented as:

$$
Q^\pi(s,a)
$$

It represents the expected future return when:

1. The agent is in state \(s\).
2. The agent takes action \(a\).
3. The agent then follows policy \(\pi\).

Conceptually:

$$
Q^\pi(s,a)
=
\text{Expected Future Return}
\mid
(s,a)
$$

The Q-function is particularly important in many reinforcement learning algorithms.

## 14. Exploration and Exploitation

A reinforcement learning agent faces an important decision between:

### Exploration

Trying actions that are not well understood in order to discover potentially better strategies.

### Exploitation

Choosing actions that are already known to produce good results.

The two can be represented as:

$$
\boxed{
\text{Exploration}
\leftrightarrow
\text{Exploitation}
}
$$

If an agent always exploits, it may never discover a better strategy.

If it explores too much, it may fail to use actions that are already known to work well.

A good learning strategy must balance these two behaviors.

## 15. Example: Grid World

Consider a simple grid environment.

The agent starts at one location and must reach a goal.

The available actions may be:

$$
\mathcal{A}
=
\{\text{Up},\text{Down},\text{Left},\text{Right}\}
$$

The environment may provide:

$$
r=-1
$$

for each unnecessary step and:

$$
r=+10
$$

when the agent reaches the goal.

The agent must learn which sequence of actions leads to a high cumulative reward.

Conceptually:

$$
\text{Start}
\rightarrow
\text{Actions}
\rightarrow
\text{Goal}
$$

The agent learns through repeated interaction with the environment.

## 16. Example: Game Playing

Reinforcement Learning can be used in environments such as games.

The agent may observe the current game state:

$$
s_t
$$

It selects an action:

$$
a_t
$$

The game environment responds:

$$
(s_t,a_t)
\rightarrow
(s_{t+1},r_{t+1})
$$

The reward may represent the outcome of the action.

For example, the environment might provide a large positive reward for winning and a negative reward for losing.

Through repeated gameplay, the agent can learn a policy that attempts to maximize long-term reward.

## 17. Episodic and Continuing Tasks

Reinforcement Learning problems can be broadly viewed as either episodic or continuing.

### Episodic Task

An episode has a defined starting point and ending point.

For example:

$$
\text{Start}
\rightarrow
\text{Actions}
\rightarrow
\text{Terminal State}
$$

A game that ends when a player wins or loses is an example of an episodic task.

### Continuing Task

A continuing task does not necessarily have a natural terminal state.

The agent may continue interacting with the environment indefinitely.

The distinction affects how returns, objectives, and learning procedures are defined.

## 18. Reinforcement Learning vs Supervised Learning

Reinforcement Learning and Supervised Learning both involve learning, but their feedback mechanisms are different.

| Aspect | Supervised Learning | Reinforcement Learning |
|---|---|---|
| Training information | Input-target pairs | Rewards from interaction |
| Correct action provided? | Usually yes, through target | Not directly |
| Environment interaction | Not required | Central to the setting |
| Feedback | Target-based | Reward-based |
| Main objective | Predict target | Maximize cumulative reward |

The key distinction is:

$$
\boxed{
\text{Supervised Learning}
\rightarrow
\text{Learn From Known Targets}
}
$$

$$
\boxed{
\text{Reinforcement Learning}
\rightarrow
\text{Learn From Consequences of Actions}
}
$$

## 19. Reinforcement Learning vs Unsupervised Learning

Reinforcement Learning and Unsupervised Learning both differ from ordinary supervised learning, but they are not the same.

In Unsupervised Learning:

$$
X
\rightarrow
\text{Discovered Structure}
$$

There is no predefined target or reward-based interaction.

In Reinforcement Learning:

$$
\text{State}
\rightarrow
\text{Action}
\rightarrow
\text{Reward}
\rightarrow
\text{Learning}
$$

The agent receives feedback from its interaction with an environment.

Therefore:

$$
\boxed{
\text{Unsupervised Learning}
\neq
\text{Reinforcement Learning}
}
$$

## 20. Common Reinforcement Learning Concepts

The most important concepts introduced in Reinforcement Learning are:

- **Agent** — the learner or decision maker.
- **Environment** — the system with which the agent interacts.
- **State** — the current situation.
- **Action** — a decision made by the agent.
- **Reward** — feedback from the environment.
- **Policy** — a strategy for selecting actions.
- **Value Function** — an estimate of expected future return from a state.
- **Q-Function** — an estimate of expected future return for a state-action pair.
- **Exploration** — trying actions to discover useful behavior.
- **Exploitation** — selecting actions based on existing knowledge.
- **Episode** — a sequence of interactions that may end in a terminal state.

## 21. Common Reinforcement Learning Approaches

Several important approaches are studied in Reinforcement Learning.

Examples include:

- Q-Learning
- SARSA
- Deep Q-Networks
- Policy Gradient Methods
- Actor-Critic Methods

These approaches differ in how they represent and learn policies, value functions, or both.

The algorithms themselves will be studied separately in the advanced Machine Learning and Reinforcement Learning sections.

## 22. Challenges in Reinforcement Learning

Reinforcement Learning can involve several challenges.

### 1. Exploration vs Exploitation

The agent must decide when to explore new actions and when to use known actions.

### 2. Delayed Rewards

An action may produce an important consequence only after many later steps.

### 3. Large State Spaces

Complex environments can contain a very large number of possible states.

### 4. Large Action Spaces

Some environments may provide many possible actions.

### 5. Sample Efficiency

Learning may require many interactions with the environment.

These challenges make Reinforcement Learning different from many standard supervised learning problems.

## 23. Complete Reinforcement Learning Workflow

A simplified Reinforcement Learning process is:

$$
\text{State}
\rightarrow
\text{Action}
\rightarrow
\text{Environment}
\rightarrow
\text{Reward + New State}
\rightarrow
\text{Learning}
$$

Repeated interaction produces a sequence:

$$
s_0,a_0,r_1,s_1,a_1,r_2,s_2,\ldots
$$

The agent uses this experience to improve its policy or other learned components.

The overall objective is to learn behavior that produces high cumulative reward.

## 24. Important Point About Reinforcement Learning

Reinforcement Learning is not simply a system that receives rewards and automatically becomes intelligent.

The learning process depends on:

- The design of the environment.
- The state representation.
- The available actions.
- The reward function.
- The learning algorithm.
- The exploration strategy.
- The amount and quality of experience.

A poorly designed reward function can encourage undesirable behavior.

Therefore:

$$
\boxed{
\text{Reward Design}
\text{ is an Important Part of Reinforcement Learning}
}
$$

## 25. Summary

Reinforcement Learning is a type of Machine Learning in which an agent learns through interaction with an environment.

The fundamental interaction is:

$$
\boxed{
s_t
\rightarrow
a_t
\rightarrow
r_{t+1},s_{t+1}
}
$$

The most important components are:

- Agent
- Environment
- State
- Action
- Reward
- Policy
- Value Function
- Q-Function
- Exploration
- Exploitation

The agent attempts to learn behavior that maximizes cumulative reward:

$$
G_t
=
r_{t+1}
+
\gamma r_{t+2}
+
\gamma^2 r_{t+3}
+
\cdots
$$

The central idea is:

$$
\boxed{
\text{Learn Better Actions Through Interaction and Feedback}
}
$$

Reinforcement Learning provides the foundation for studying algorithms such as Q-Learning, SARSA, Deep Q-Networks, and policy-based methods.
